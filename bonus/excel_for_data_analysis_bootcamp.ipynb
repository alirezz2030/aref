## 1. Spreadsheet Structure and Clean Tables

An Excel workbook can contain multiple worksheets. A clean analysis table should have one header row, one record per row, one variable per column, and no merged cells inside the data range.

The companion `.xlsx` workbook contains six sheets: a guide, sales data, lookups, dashboard, exercises, and a formula reference. This lesson demonstrates the same data model in a small Python example; the actual formulas are practiced in Excel.

```python
# Sample records that mirror the companion Excel workbook.
sales = [  # Each dictionary represents one order record.
    {"order_id": 1, "region": "North", "product": "Laptop", "units": 2, "unit_price": 900, "discount": 0.05},  # Add order 1.
    {"order_id": 2, "region": "South", "product": "Phone", "units": 3, "unit_price": 500, "discount": 0.00},  # Add order 2.
    {"order_id": 3, "region": "North", "product": "Tablet", "units": 4, "unit_price": 300, "discount": 0.10},  # Add order 3.
    {"order_id": 4, "region": "West", "product": "Monitor", "units": 1, "unit_price": 220, "discount": 0.00},  # Add order 4.
]  # Finish the sample table.

# In Excel, a structured data table can be created with Ctrl+T.
# Example Excel formula in a table: =[@Units]*[@[Unit Price]]
print("Number of records:", len(sales))  # Count the sample order rows.
print("Columns:", list(sales[0].keys()))  # Display the field names shared by each record.
```

## 2. Basic Arithmetic Formulas

Excel arithmetic operators include `+`, `-`, `*`, `/`, and `^` (power). A formula begins with `=` and can reference cells so results update when inputs change.

For a sales table, `=E2*I2` might calculate gross revenue from units and unit price. Cell locations depend on your worksheet layout.

```python
# Example Excel formulas:
# Gross revenue in J2: =E2*I2
# Revenue after discount in K2: =J2*(1-F2)
units = 2  # Sample order quantity.
unit_price = 900  # Sample price per unit.
discount = 0.05  # Sample discount rate, equal to 5%.
gross_revenue = units * unit_price  # Mirror Excel's multiplication formula.
net_revenue = gross_revenue * (1 - discount)  # Mirror Excel's after-discount formula.
print("Gross revenue:", gross_revenue)  # Display revenue before discount.
print("Net revenue:", net_revenue)  # Display revenue after discount.
```

## 3. SUM, AVERAGE, MIN, and MAX

These core functions summarize numeric ranges:
- `SUM(range)` adds values.
- `AVERAGE(range)` calculates the arithmetic mean.
- `MIN(range)` returns the smallest value.
- `MAX(range)` returns the largest value.

Excel generally ignores blank cells in these numeric aggregations, but zeros still count as numeric values.

```python
values = [125, 200, 175, 300, 100]  # Create five example sales values.
# Excel formulas: =SUM(A2:A6), =AVERAGE(A2:A6), =MIN(A2:A6), =MAX(A2:A6)
print("SUM:", sum(values))  # Calculate the total as a Python equivalent of SUM.
print("AVERAGE:", sum(values) / len(values))  # Calculate the mean as a Python equivalent of AVERAGE.
print("MIN:", min(values))  # Find the smallest value.
print("MAX:", max(values))  # Find the largest value.
```

## 4. COUNT, COUNTA, and COUNTBLANK

Use `COUNT` to count numeric cells, `COUNTA` to count non-empty cells, and `COUNTBLANK` to count blank cells.

These functions help validate imported data. Note that `COUNT` ignores text, while `COUNTA` counts text and numbers. In Excel, a formula returning an empty string may behave differently from a truly empty cell for some checks.

```python
cells = [10, "North", None, 0, "Laptop", None]  # Simulate spreadsheet cells; None represents a blank.
# Excel formulas: =COUNT(A2:A7), =COUNTA(A2:A7), =COUNTBLANK(A2:A7)
count_numeric = sum(isinstance(value, (int, float)) and not isinstance(value, bool) for value in cells if value is not None)  # Count numeric entries, including zero.
count_nonblank = sum(value is not None for value in cells)  # Count populated cells.
count_blank = sum(value is None for value in cells)  # Count blank entries.
print("COUNT equivalent:", count_numeric)  # Display the numeric-cell count.
print("COUNTA equivalent:", count_nonblank)  # Display the non-empty-cell count.
print("COUNTBLANK equivalent:", count_blank)  # Display the blank-cell count.
```

## 5. Relative, Absolute, and Mixed References

When a formula is copied, **relative references** change with position, while `$A$1` is an **absolute reference**. Mixed references such as `$A1` and `A$1` lock only the column or the row.

This is critical when copying a formula down many records or across a grid. In the example workbook, each order's discount is in its own row, so a relative reference such as `F2` is appropriate.

```python
# Formula copied down the rows in Excel:
# Relative references: =E2*I2
# Absolute reference to a fixed discount assumption in B1: =J2*(1-$B$1)
# Mixed reference examples: =$A2*B$1
unit_values = [100, 200, 300]  # Sample values in the order rows.
fixed_discount = 0.10  # A single assumption that stays fixed when formulas are copied.
net_values = [value * (1 - fixed_discount) for value in unit_values]  # Apply the same fixed rate to every value.
print("Values after fixed 10% discount:", net_values)  # Display results that mirror the absolute-reference formula.
```

## 6. IF: Conditional Labels

`IF(condition, value_if_true, value_if_false)` returns one result when a condition is true and another when false.

Analysts use `IF` to flag target achievement, label large orders, or identify exceptions. Write the threshold from a clear business rule rather than choosing an arbitrary number.

```python
revenue_values = [450, 850, 1200, 220]  # Store example order revenues.
# Excel formula in a row: =IF(K2>=1000,"High","Standard")
labels = ["High" if revenue >= 1000 else "Standard" for revenue in revenue_values]  # Apply the same conditional rule in Python.
print(list(zip(revenue_values, labels)))  # Display each example value and its label.
```

## 7. AND, OR, and Nested Conditions

`AND()` is true only when all its conditions are true; `OR()` is true when at least one condition is true. They are frequently combined with `IF()`.

For more than two or three categories, use `IFS()` in supported Excel versions or keep the logic readable with a lookup table.

```python
orders = [  # Create sample order attributes for rule testing.
    {"region": "North", "channel": "Online", "revenue": 1200},  # Example order meeting both criteria.
    {"region": "South", "channel": "Store", "revenue": 600},  # Example order meeting neither region nor channel criteria.
    {"region": "North", "channel": "Store", "revenue": 800},  # Example order meeting only the region criterion.
]  # Finish the test records.
# Excel formulas: =IF(AND(C2="North",H2="Online"),"Yes","No")
# Excel formula using OR: =IF(OR(C2="North",C2="East"),"Priority region","Other")
for order in orders:  # Check the rule for each example order.
    both = order["region"] == "North" and order["channel"] == "Online"  # Mirror the AND condition.
    priority_region = order["region"] in ("North", "East")  # Mirror the OR condition for region.
    print(order, "AND result:", both, "OR result:", priority_region)  # Display both rule outcomes.
```

## 8. SUMIF: Sum with One Condition

`SUMIF(criteria_range, criteria, sum_range)` sums values whose corresponding criteria-range values match one condition.

It is commonly used for revenue by product or sales by region. The criteria range and sum range should cover corresponding rows.

```python
regions = ["North", "South", "North", "West", "South"]  # Store region labels.
net_revenue = [500, 750, 300, 220, 450]  # Store revenue for those same records.
# Excel formula: =SUMIF(C2:C6,"North",K2:K6)
north_total = sum(value for region, value in zip(regions, net_revenue) if region == "North")  # Sum values whose region is North.
print("North revenue:", north_total)  # Display the single-condition sum.
```

## 9. COUNTIF and AVERAGEIF

`COUNTIF(range, criteria)` counts cells matching one criterion. `AVERAGEIF(criteria_range, criteria, average_range)` averages corresponding values matching that criterion.

Criteria can be exact text, a number, a comparison such as `">100"`, or a wildcard pattern. For wildcard text matching, `*` means any sequence and `?` means one character.

```python
products = ["Phone", "Laptop", "Phone", "Tablet", "Phone"]  # Store product names.
units = [2, 1, 4, 3, 5]  # Store quantities aligned to the products.
# Excel formulas: =COUNTIF(D2:D6,"Phone") and =AVERAGEIF(D2:D6,"Phone",E2:E6)
phone_units = [qty for product, qty in zip(products, units) if product == "Phone"]  # Filter quantities associated with phones.
print("Phone count:", len(phone_units))  # Mirror COUNTIF for matching records.
print("Average phone units:", sum(phone_units) / len(phone_units))  # Mirror AVERAGEIF for phone orders.
```

## 10. SUMIFS and COUNTIFS

`SUMIFS` sums values meeting multiple conditions; `COUNTIFS` counts rows meeting multiple conditions.

Unlike `SUMIF`, `SUMIFS` puts the sum range first. Keep criteria ranges the same shape. For an OR condition with `SUMIFS`, you may need multiple formulas or other approaches.

```python
regions = ["North", "North", "South", "North", "South"]  # Store order regions.
channels = ["Online", "Store", "Online", "Online", "Store"]  # Store order channels.
revenue = [500, 600, 800, 300, 400]  # Store each order's revenue.
# Excel formula: =SUMIFS(K2:K6,C2:C6,"North",H2:H6,"Online")
selected = [value for region, channel, value in zip(regions, channels, revenue) if region == "North" and channel == "Online"]  # Select rows meeting both criteria.
print("North Online revenue:", sum(selected))  # Mirror SUMIFS.
print("North Online order count:", len(selected))  # Mirror COUNTIFS on the same rows.
```

## 11. IFERROR: Handle Formula Errors

`IFERROR(value, value_if_error)` returns a fallback when a formula results in an error. It's often paired with lookups to provide a clear “Not found” message.

Use error handling carefully: a fallback can hide genuine issues. During development, investigate unexpected errors instead of automatically replacing all errors with zero.

```python
prices = {"Laptop": 900, "Phone": 500, "Tablet": 300}  # Create a small product-price lookup dictionary.
product_to_find = "Monitor"  # Choose a product that is not in the lookup dictionary.
# Excel formula: =IFERROR(VLOOKUP(D2,Lookups!$A$2:$B$7,2,FALSE),"Not found")
price_result = prices.get(product_to_find, "Not found")  # Mirror a lookup with an explicit fallback.
print("Lookup result:", price_result)  # Display the safe result rather than raising a missing-key error.
```

## 12. VLOOKUP and XLOOKUP

`VLOOKUP` searches the first column of a table and returns a value from another column in the same row. `XLOOKUP` is more flexible and can search one range and return from another; it is available in modern Excel versions.

For exact matching with VLOOKUP, use `FALSE` as the fourth argument. Keep lookup keys unique when the business logic expects only one result.

```python
lookup_prices = {"Laptop": 900, "Phone": 500, "Tablet": 300, "Monitor": 220}  # Define sample product prices.
product = "Phone"  # Choose a product to look up.
# Excel VLOOKUP: =VLOOKUP(D2,Lookups!$A$2:$B$7,2,FALSE)
# Excel XLOOKUP: =XLOOKUP(D2,Lookups!$A$2:$A$7,Lookups!$B$2:$B$7,"Not found")
print("Found price:", lookup_prices.get(product, "Not found"))  # Display the lookup result.
print("Missing product:", lookup_prices.get("Desk", "Not found"))  # Demonstrate the fallback when no match exists.
```

## 13. INDEX + MATCH

`INDEX` returns a value from a range at a given position. `MATCH` finds the position of a value within a range. Together they make a flexible lookup combination that also works in older Excel versions.

Use exact match mode (`0`) in MATCH when searching for an exact product or ID.

```python
products = ["Laptop", "Phone", "Tablet", "Monitor"]  # Store the lookup range in its table order.
prices = [900, 500, 300, 220]  # Store corresponding prices in the same order.
target = "Tablet"  # Choose the product to find.
# Excel formula: =INDEX(B2:B5,MATCH("Tablet",A2:A5,0))
position = products.index(target)  # Find the zero-based position of the exact product match.
price = prices[position]  # Retrieve the value at the matching position.
print(target, "price:", price)  # Display the INDEX/MATCH-style lookup result.
```

## 14. Text Cleaning: TRIM, CLEAN, PROPER

Text fields can contain leading/trailing spaces, repeated spaces, inconsistent capitalization, and non-printing characters.

- `TRIM()` removes extra ordinary spaces.
- `CLEAN()` removes many non-printing characters.
- `PROPER()` capitalizes the first letter of each word.

Check the effect on names, product IDs, and abbreviations before applying these functions to whole columns.

```python
messy_product = "  lAPTOP   "  # Simulate a text label with spaces and inconsistent capitalization.
# Excel formula: =PROPER(TRIM(D2))
trimmed = " ".join(messy_product.split())  # Mirror TRIM by collapsing extra whitespace.
proper_case = trimmed.title()  # Mirror PROPER for this simple product label.
print("Before:", repr(messy_product))  # Show the spaces in the original value.
print("After TRIM + PROPER:", proper_case)  # Display the cleaned label.
```

## 15. LEFT, RIGHT, MID, FIND

Text extraction functions help parse structured strings such as SKU codes, reference IDs, and file names.

- `LEFT(text, n)` returns the first n characters.
- `RIGHT(text, n)` returns the final n characters.
- `MID(text, start, n)` extracts n characters from a 1-based position.
- `FIND(needle, text)` returns a 1-based position and is case-sensitive in Excel.

```python
sku = "LAP-2048-BL"  # Create a sample stock-keeping unit code.
# Excel formulas: =LEFT(A2,3), =RIGHT(A2,2), =MID(A2,5,4), =FIND("-",A2)
left_part = sku[:3]  # Mirror LEFT by selecting the first three characters.
right_part = sku[-2:]  # Mirror RIGHT by selecting the last two characters.
middle_part = sku[4:8]  # Mirror MID for Excel's start position 5 and four characters.
find_position = sku.index("-") + 1  # Convert Python's zero-based position to Excel's one-based position.
print("LEFT:", left_part, "| MID:", middle_part, "| RIGHT:", right_part, "| FIND position:", find_position)  # Show each extracted piece.
```

## 16. Date Functions

Date functions help extract reporting periods and calculate date boundaries.

- `YEAR(date)` and `MONTH(date)` extract components.
- `EOMONTH(date, months)` returns the end of the month a number of months away.
- `TODAY()` returns the current date and updates as the workbook recalculates.

Use real Excel date values rather than text when possible. Date formulas can vary slightly across spreadsheet applications.

```python
from datetime import date  # Import date utilities for this illustrative formula example.
order_date = date(2025, 3, 14)  # Store a sample order date.
# Excel formulas: =YEAR(B2), =MONTH(B2), =EOMONTH(B2,0), =TODAY()
year_part = order_date.year  # Mirror YEAR.
month_part = order_date.month  # Mirror MONTH.
print("Year:", year_part)  # Display the year.
print("Month:", month_part)  # Display the month.
print("Order date:", order_date.isoformat())  # Display the original date in ISO format.
print("In Excel, EOMONTH(B2,0) returns the final date of this month.")  # Explain the end-of-month formula without depending on an Excel engine.
```

## 17. Percentages and Period-over-Period Growth

Percentage change is calculated as `(new value − old value) / old value`. In Excel, percentage cells are usually stored as decimals and formatted as percentages.

A previous value of zero makes a percentage change undefined. Handle zero denominators explicitly rather than allowing misleading or unhandled errors.

```python
previous_month = 10000  # Store the earlier month's revenue.
current_month = 11500  # Store the current month's revenue.
# Excel formula: =IFERROR((B3-B2)/B2,0)
growth = (current_month - previous_month) / previous_month if previous_month != 0 else 0  # Calculate the change safely.
print("Absolute change:", current_month - previous_month)  # Display the amount of revenue increase.
print("Growth rate:", f"{growth:.1%}")  # Format the growth rate as a percentage.
```

## 18. ROUND, RANK.EQ, LARGE, SMALL

`ROUND(value, digits)` rounds a value to a chosen number of decimal places. `RANK.EQ` ranks a value within a range; `LARGE` and `SMALL` return the kth-largest or kth-smallest item.

Ranking direction matters: `RANK.EQ(value, range, 0)` ranks the largest value as 1.

```python
revenues = [750, 1200, 950, 1200, 420]  # Create example revenues with a tie.
target_revenue = 950  # Choose a value to rank.
# Excel formulas: =ROUND(K2,2), =RANK.EQ(K2,$K$2:$K$6,0), =LARGE(K2:K6,2), =SMALL(K2:K6,2)
rounded = round(target_revenue, 2)  # Mirror rounding to two decimal places.
rank = 1 + sum(value > target_revenue for value in revenues)  # Mirror descending RANK.EQ semantics for this sample.
second_largest = sorted(revenues, reverse=True)[1]  # Mirror LARGE(range, 2).
second_smallest = sorted(revenues)[1]  # Mirror SMALL(range, 2).
print("Rounded value:", rounded)  # Display the rounded value.
print("Descending rank (ties share rank):", rank)  # Display the rank of the selected value.
print("Second largest:", second_largest)  # Display the second-largest value.
print("Second smallest:", second_smallest)  # Display the second-smallest value.
```

## 19. Sorting, Filtering, and Excel Tables

Excel's Sort and Filter tools let you inspect subsets without changing the underlying data. Excel Tables (Ctrl+T) provide headers, automatic expansion, and structured references.

A structured reference such as `=SUM(SalesTable[Net Revenue])` is easier to maintain than a fixed range when rows are added to the table.

```python
sales_rows = [  # Define a few example rows with region and revenue.
    {"order_id": 1, "region": "North", "revenue": 1250},  # Example row 1.
    {"order_id": 2, "region": "South", "revenue": 750},  # Example row 2.
    {"order_id": 3, "region": "North", "revenue": 1800},  # Example row 3.
]  # Finish the sample rows.
north_rows = [row for row in sales_rows if row["region"] == "North"]  # Mirror filtering to North in Excel.
ranked = sorted(sales_rows, key=lambda row: row["revenue"], reverse=True)  # Mirror sorting by revenue descending.
# Structured table formula example: =SUM(SalesTable[Net Revenue])
print("North rows:", north_rows)  # Display rows matching the filter.
print("Sorted by revenue:", ranked)  # Display the sorted records.
```

## 20. Removing Duplicates and Data Quality

Duplicate removal can be helpful for repeated imports, but first decide what makes a record a duplicate. The same customer can legitimately have multiple orders.

In Excel, use Data → Remove Duplicates and select the columns that define a duplicate. Consider keeping a raw copy before making irreversible changes.

```python
rows = [  # Create records with one exact duplicate order.
    (1001, "North", 500),  # First record.
    (1002, "South", 800),  # Second record.
    (1001, "North", 500),  # Exact duplicate of the first record.
]  # Finish the sample data.
unique_rows = list(dict.fromkeys(rows))  # Preserve order while removing exact duplicate tuples.
# Excel alternative: Data → Remove Duplicates, then select the key columns.
print("Rows before:", len(rows))  # Count original records.
print("Rows after exact deduplication:", len(unique_rows))  # Count remaining unique records.
print("Unique rows:", unique_rows)  # Display the remaining records.
```

## 21. PivotTables

A PivotTable quickly summarizes large tables without writing formulas for every group. Drag fields into **Rows**, **Columns**, **Values**, and **Filters** to answer questions such as revenue by region and product.

Verify the Values aggregation: numeric fields can default to Sum, but some fields or source data issues may cause Count instead. Refresh the PivotTable when the source data changes.

```python
records = [  # Create small examples of region/product sales.
    ("North", "Laptop", 1800),  # Record one.
    ("North", "Phone", 500),  # Record two.
    ("South", "Laptop", 900),  # Record three.
    ("North", "Laptop", 900),  # Record four.
]  # Finish the sample sales data.
pivot_like = {}  # Prepare a dictionary to mimic PivotTable row/column aggregation.
for region, product, revenue in records:  # Process each transaction.
    pivot_like[(region, product)] = pivot_like.get((region, product), 0) + revenue  # Add revenue to its region/product group.
# In Excel: Insert → PivotTable; Rows=Region, Columns=Product, Values=Sum of Revenue.
for key, value in sorted(pivot_like.items()):  # Print each grouped result.
    print(key, "=>", value)  # Display region, product, and summed revenue.
```

## 22. Conditional Formatting

Conditional formatting applies visual styles when values meet rules, such as highlighting low revenue, duplicate IDs, or dates past due.

Use a small number of meaningful colors and write rules that reflect a business threshold. Color alone should not be the only way to communicate a critical status; consider adding text labels as well.

```python
revenue_values = [120, 250, 80, 1400, 300]  # Store sample order revenue values.
threshold = 250  # Define the business threshold used by the formatting rule.
# Excel rule: Home → Conditional Formatting → Highlight Cells Rules → Less Than → 250
flagged = [value < threshold for value in revenue_values]  # Mirror the condition behind the formatting rule.
print(list(zip(revenue_values, flagged)))  # Show which values would receive highlighting.
```

## 23. Data Validation and Drop-Down Lists

Data validation restricts entries to allowed values or ranges. A dropdown list can prevent inconsistent spelling of categories such as regions or sales channels.

In Excel, Data → Data Validation → List can use a fixed list or a range on a lookup sheet. Validation improves consistency but does not replace downstream data-quality checks.

```python
allowed_channels = ["Online", "Store", "Partner"]  # Define the allowed channel values.
test_entry = "Online"  # Choose an example value to validate.
is_valid = test_entry in allowed_channels  # Check whether the value appears in the approved list.
# Excel list source example: =Lookups!$H$2:$H$4
print("Entered value:", test_entry)  # Show the test value.
print("Passes allowed-value rule:", is_valid)  # Display whether the entry is permitted.
```

## 24. Charts and Chart Selection

Use charts to communicate the result of an analysis: line charts for time trends, bars for category comparisons, scatter plots for numeric relationships, and histograms for distributions.

A chart should have a descriptive title, labeled axes, units, and a scale that does not mislead. Avoid 3D effects and unnecessary decoration.

```python
monthly_revenue = [8200, 9100, 8750, 10200, 11100, 10800]  # Store six monthly revenue totals.
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]  # Store their matching month labels.
# In Excel: select Month and Revenue summary columns → Insert → Line Chart.
# For product comparisons, select Product and Revenue → Insert → Clustered Column Chart.
print("Month/revenue pairs:", list(zip(months, monthly_revenue)))  # Display the data range that would feed a monthly line chart.
print("Highest month:", months[monthly_revenue.index(max(monthly_revenue))])  # Identify the month with the highest revenue for annotation.
```

## 25. Final Project: Regional Sales Dashboard

A useful dashboard combines a few key performance indicators with focused summaries and charts. For example: total net revenue, total units, revenue by region, revenue by product, and a monthly trend.

Use formulas for repeatable metrics, keep raw data separate from summaries, label units, and check that the underlying definitions match the business question. The companion workbook includes a working formula-driven dashboard and exercises.

```python
records = [  # Create a final mini dataset for the dashboard exercise.
    {"region": "North", "product": "Laptop", "units": 2, "price": 900, "discount": 0.05},  # Add one North order.
    {"region": "South", "product": "Phone", "units": 3, "price": 500, "discount": 0.00},  # Add one South order.
    {"region": "North", "product": "Tablet", "units": 4, "price": 300, "discount": 0.10},  # Add another North order.
    {"region": "West", "product": "Monitor", "units": 1, "price": 220, "discount": 0.00},  # Add one West order.
]  # Finish the mini project data.
for row in records:  # Calculate net revenue for each record.
    row["net_revenue"] = row["units"] * row["price"] * (1 - row["discount"])  # Mirror the Excel net-revenue formula.
total_net_revenue = sum(row["net_revenue"] for row in records)  # Calculate the total KPI.
units_sold = sum(row["units"] for row in records)  # Calculate total units sold.
region_totals = {}  # Prepare revenue totals by region.
for row in records:  # Aggregate the rows by region.
    region_totals[row["region"]] = region_totals.get(row["region"], 0) + row["net_revenue"]  # Add each order to its region total.
top_region = max(region_totals, key=region_totals.get)  # Find the region with the largest total.
# Excel dashboard formulas: =SUM(SalesTable[Net Revenue]) and =SUM(SalesTable[Units])
print("Total net revenue:", round(total_net_revenue, 2))  # Show the overall revenue KPI.
print("Units sold:", units_sold)  # Show the total quantity KPI.
print("Revenue by region:", region_totals)  # Show the region summary.
print("Top region:", top_region)  # State the highest-revenue region.
```
