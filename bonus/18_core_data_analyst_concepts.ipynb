## 2. Inspecting Data Before Analysis

Before analyzing data, check how many rows and columns exist, which data types are present, and whether any fields contain missing values.

Useful commands include `head()`, `shape`, `columns`, `dtypes`, `info()`, and `isna().sum()`. These checks help catch problems before they affect results.

```python
print(df.head(3))  # Display the first three records.
print("Shape:", df.shape)  # Show the number of rows and columns.
print("Columns:", df.columns.tolist())  # Display column names as a Python list.
print("\nData types:")  # Add a label before the data-type report.
print(df.dtypes)  # Show each column's current data type.
print("\nMissing values per column:")  # Add a label before the missing-value report.
print(df.isna().sum())  # Count missing cells in every column.
```

## 3. Selecting Columns and Filtering Rows

Filtering is similar to SQL's `WHERE` clause. Selecting columns is similar to choosing fields in `SELECT`.

Use `df[["col1", "col2"]]` to select columns and a Boolean condition such as `df[df["region"] == "North"]` to filter rows. Combine conditions with `&` for AND and `|` for OR; wrap each condition in parentheses.

```python
north_orders = df[df["region"] == "North"]  # Keep only rows where region is North.
print(north_orders[["order_id", "product", "units"]])  # Display selected columns for those rows.

high_quantity = df[(df["units"] >= 3) & (df["product"] == "Phone")]  # Filter phones with at least three units.
print(high_quantity[["order_id", "region", "units"]])  # Show the orders matching both conditions.
```

## 4. Handling Missing Values

Missing values can distort calculations or cause some operations to fail. First identify them, then decide whether to remove rows, fill values, or investigate the source.

Here, missing `units` are filled with the median, and missing `region` values are labelled `Unknown`. In real projects, the best choice depends on why the values are missing and what the field means.

```python
clean_df = df.copy()  # Copy the raw data so the original remains available.
print(clean_df.isna().sum())  # Count missing values before cleaning.
units_median = clean_df["units"].median()  # Calculate the median of the available unit counts.
clean_df["units"] = clean_df["units"].fillna(units_median)  # Fill missing unit counts with the median.
clean_df["region"] = clean_df["region"].fillna("Unknown")  # Label missing regions instead of guessing them.
print("\nMissing values after cleaning:")  # Print a label for the validation result.
print(clean_df.isna().sum())  # Check whether missing values remain.
```

## 5. Removing Duplicate Records

Duplicate rows can cause double-counting. `duplicated()` flags repeated rows; `drop_duplicates()` removes repeated rows while keeping the first occurrence by default.

Do not automatically remove records merely because one field repeats. For example, repeated customer IDs may be perfectly valid in a transaction table. Decide which columns define a true duplicate.

```python
duplicate_mask = df.duplicated()  # Mark rows that exactly repeat an earlier row.
print("Duplicate rows:", int(duplicate_mask.sum()))  # Count the duplicate rows.
dedup_df = df.drop_duplicates().copy()  # Create a new DataFrame without exact duplicate rows.
print("Before:", len(df), "rows")  # Display the original row count.
print("After:", len(dedup_df), "rows")  # Display the row count after removing duplicates.
```

## 6. Converting Data Types

Dates and numeric values should have appropriate data types. Correct types make sorting, date calculations, arithmetic, and plotting more reliable.

`pd.to_datetime()` converts date text to datetime values. `pd.to_numeric()` converts values to numbers; `errors="coerce"` turns unparseable values into missing values, which should then be investigated.

```python
typed_df = clean_df.drop_duplicates().copy()  # Start with cleaned data and remove exact duplicates.
typed_df["date"] = pd.to_datetime(typed_df["date"], errors="coerce")  # Convert date strings into datetime values.
typed_df["units"] = pd.to_numeric(typed_df["units"], errors="coerce")  # Ensure units are numeric.
typed_df["unit_price"] = pd.to_numeric(typed_df["unit_price"], errors="coerce")  # Ensure prices are numeric.
print(typed_df[["date", "units", "unit_price"]].dtypes)  # Verify the converted data types.
print(typed_df[["date", "units", "unit_price"]].head())  # Preview the converted columns.
```

## 7. Creating Calculated Columns and KPIs

Analysts often create metrics from existing fields. For example, revenue can be calculated as `units × unit_price`.

A **KPI** (key performance indicator) measures a business outcome, such as revenue, units sold, average order value, or conversion rate. Always confirm the business definition before presenting a calculated metric.

```python
analysis_df = typed_df.copy()  # Create a working DataFrame for analysis.
analysis_df["revenue"] = analysis_df["units"] * analysis_df["unit_price"]  # Calculate revenue per order.
analysis_df["average_unit_price"] = analysis_df["unit_price"]  # Keep a clearly named price metric for later analysis.
total_revenue = analysis_df["revenue"].sum()  # Add up revenue across orders with available values.
average_order_revenue = analysis_df["revenue"].mean()  # Calculate mean revenue per order.
print("Total revenue:", round(total_revenue, 2))  # Display total revenue.
print("Average order revenue:", round(average_order_revenue, 2))  # Display average revenue per recorded order.
```

## 8. Sorting and Finding Top Performers

Sorting helps identify the highest- or lowest-performing records. Use `sort_values()` with `ascending=False` for descending order and `head(n)` to keep only the top n rows.

Top results are useful for questions like “Which orders generated the most revenue?” or “Which products sold the most units?” Choose the metric that answers the actual business question.

```python
top_orders = analysis_df.sort_values("revenue", ascending=False).head(5)  # Keep the five orders with the highest revenue.
print(top_orders[["order_id", "region", "product", "revenue"]])  # Show their key details.

product_units = analysis_df.groupby("product", as_index=False)["units"].sum()  # Add units sold for each product.
top_products = product_units.sort_values("units", ascending=False)  # Rank products by total units.
print("\nProducts ranked by units sold:")  # Label the ranking output.
print(top_products)  # Display all products in rank order.
```

## 9. Grouping and Aggregation

`groupby()` splits records into groups and calculates summaries for each group. It plays a role similar to SQL's `GROUP BY`.

Common aggregations include `sum`, `mean`, `median`, `count`, `min`, and `max`. Multiple metrics can be calculated together with `.agg()`.

```python
region_summary = analysis_df.groupby("region", as_index=False).agg(  # Group orders by region and calculate multiple metrics.
    orders=("order_id", "nunique"),  # Count unique order IDs in each region.
    units_sold=("units", "sum"),  # Add units sold in each region.
    revenue=("revenue", "sum"),  # Add revenue in each region.
    average_order_revenue=("revenue", "mean"),  # Calculate mean order revenue in each region.
)  # Finish the aggregation definition.
region_summary = region_summary.sort_values("revenue", ascending=False)  # Rank regions by total revenue.
print(region_summary)  # Display the regional summary.
```

## 10. Joining Tables

Real datasets are often split across multiple tables. A join combines them using a shared key, much like SQL's `JOIN`.

Pandas supports `inner`, `left`, `right`, and `outer` joins through `merge()`. Check key uniqueness and unmatched rows so that a join does not unexpectedly multiply records or lose data.

```python
region_targets = pd.DataFrame({  # Create a small reference table with regional targets.
    "region": ["North", "South", "East", "West", "Unknown"],  # Define join keys.
    "revenue_target": [2500, 3000, 2200, 1800, 500],  # Set a target for each region.
})  # Finish the reference table.
joined = region_summary.merge(region_targets, on="region", how="left", validate="one_to_one")  # Join each region to its target.
joined["target_gap"] = joined["revenue"] - joined["revenue_target"]  # Calculate the amount above or below target.
joined["target_status"] = np.where(joined["target_gap"] >= 0, "Met", "Not met")  # Label target performance.
print(joined[["region", "revenue", "revenue_target", "target_gap", "target_status"]])  # Display target comparisons.
```

## 11. Pivot Tables

A pivot table summarizes one metric across two categorical dimensions. It is useful for comparing products across regions or months.

`pivot_table()` supports aggregation, and `fill_value=0` can make absent combinations easier to read. Use zero only when a missing combination truly means no recorded activity—not when data is unknown.

```python
pivot = pd.pivot_table(  # Create a cross-tabulated summary table.
    analysis_df,  # Provide the data to summarize.
    index="region",  # Put each region on a row.
    columns="product",  # Put each product in its own column.
    values="revenue",  # Summarize the revenue field.
    aggfunc="sum",  # Add revenue for each region-product combination.
    fill_value=0,  # Show zero for combinations with no recorded revenue.
)  # Finish the pivot-table definition.
print(pivot)  # Display revenue by region and product.
```

## 12. Date and Time Analysis

Date analysis helps reveal seasonality and trends. Once dates are parsed as datetimes, extract fields like year and month or resample a time-indexed table.

Monthly totals answer questions such as “How did revenue change each month?” Make sure the time zone and date format are correct when working with real event or transaction timestamps.

```python
dated_df = analysis_df.dropna(subset=["date"]).copy()  # Exclude rows where a date could not be parsed.
dated_df["month"] = dated_df["date"].dt.to_period("M").astype(str)  # Create a year-month label.
monthly_revenue = dated_df.groupby("month", as_index=False)["revenue"].sum()  # Sum revenue in each month.
monthly_revenue = monthly_revenue.sort_values("month")  # Sort the month labels chronologically.
print(monthly_revenue)  # Display monthly revenue totals.

monthly_indexed = dated_df.set_index("date").resample("MS")["revenue"].sum()  # Resample revenue by month start.
print("\nMonthly resampled totals:")  # Add a label before the time-series output.
print(monthly_indexed)  # Display the monthly time series.
```

## 13. Descriptive Statistics

Descriptive statistics summarize the center and spread of the data. The **mean** is sensitive to extreme values; the **median** is often more robust. Standard deviation describes dispersion around the mean.

`describe()` gives a quick summary. For skewed business data, compare mean and median and inspect percentiles rather than relying on one number.

```python
print(analysis_df[["units", "unit_price", "revenue"]].describe())  # Show count, mean, standard deviation, quartiles, and extremes.
revenue_mean = analysis_df["revenue"].mean()  # Calculate the average revenue.
revenue_median = analysis_df["revenue"].median()  # Calculate the middle revenue value.
revenue_std = analysis_df["revenue"].std()  # Calculate sample standard deviation.
print("Revenue mean:", round(revenue_mean, 2))  # Display the mean.
print("Revenue median:", round(revenue_median, 2))  # Display the median.
print("Revenue standard deviation:", round(revenue_std, 2))  # Display the standard deviation.
```

## 14. Detecting Potential Outliers with IQR

The **interquartile range (IQR)** is `Q3 − Q1`. A common screening rule flags values below `Q1 − 1.5 × IQR` or above `Q3 + 1.5 × IQR`.

This rule identifies potential outliers, not necessarily errors. An unusually large sale may be valid. Investigate flagged records using business context before removing or changing them.

```python
revenue_values = analysis_df["revenue"].dropna()  # Use only available revenue values.
q1 = revenue_values.quantile(0.25)  # Calculate the first quartile (25th percentile).
q3 = revenue_values.quantile(0.75)  # Calculate the third quartile (75th percentile).
iqr = q3 - q1  # Calculate the interquartile range.
lower_limit = q1 - 1.5 * iqr  # Calculate the lower screening threshold.
upper_limit = q3 + 1.5 * iqr  # Calculate the upper screening threshold.
outliers = analysis_df[(analysis_df["revenue"] < lower_limit) | (analysis_df["revenue"] > upper_limit)]  # Filter values outside the thresholds.
print("Lower threshold:", round(lower_limit, 2))  # Display the lower boundary.
print("Upper threshold:", round(upper_limit, 2))  # Display the upper boundary.
print(outliers[["order_id", "product", "revenue"]])  # Inspect flagged orders instead of deleting them automatically.
```

## 15. Correlation

Correlation measures the strength and direction of a linear relationship between two numeric variables. Pearson correlation ranges from `-1` to `1`.

Correlation does **not** prove causation. It can also be misleading when relationships are nonlinear, outliers are present, or important variables are missing. Avoid interpreting a correlation calculated from very few observations as strong evidence.

```python
numeric_data = analysis_df[["units", "unit_price", "revenue"]].dropna()  # Keep numeric columns and complete rows only.
correlation_matrix = numeric_data.corr(numeric_only=True)  # Calculate pairwise Pearson correlation coefficients.
print(correlation_matrix.round(2))  # Display the matrix rounded to two decimal places.
print("\nCorrelation between units and revenue:", round(numeric_data["units"].corr(numeric_data["revenue"]), 2))  # Show one selected relationship.
```

## 16. Random Sampling

Sampling selects part of a dataset for review or analysis. It is useful when a dataset is large or when manually checking a subset of rows.

Set `random_state` to make the selection reproducible. A simple random sample is not automatically representative if the data has important subgroups or time patterns; stratified sampling may be more appropriate in those cases.

```python
sample_size = min(5, len(analysis_df))  # Avoid requesting more rows than the dataset contains.
sample_df = analysis_df.sample(n=sample_size, random_state=42)  # Draw a reproducible random sample.
print(sample_df[["order_id", "date", "region", "product", "revenue"]])  # Display the sampled records.
```

## 17. Conversion Rate and Funnel Metrics

A funnel tracks how many users progress through stages such as visit → add to cart → purchase. A conversion rate is usually `conversions ÷ eligible users`.

Define the denominator precisely: purchase rate among all visitors is different from purchase rate among users who added to cart. Funnel metrics help analysts find where users drop off.

```python
funnel = pd.DataFrame({  # Create a small illustrative funnel summary.
    "stage": ["Visitors", "Added to cart", "Purchased"],  # Name the funnel stages.
    "users": [1000, 280, 120],  # Provide the number of users reaching each stage.
})  # Finish the funnel table.
visitors = funnel.loc[funnel["stage"] == "Visitors", "users"].iloc[0]  # Read the visitor count.
cart_users = funnel.loc[funnel["stage"] == "Added to cart", "users"].iloc[0]  # Read the add-to-cart count.
buyers = funnel.loc[funnel["stage"] == "Purchased", "users"].iloc[0]  # Read the purchaser count.
cart_rate = cart_users / visitors  # Calculate the share of visitors who added to cart.
purchase_rate = buyers / visitors  # Calculate the share of visitors who purchased.
cart_to_purchase_rate = buyers / cart_users  # Calculate purchase rate among cart users.
print(f"Visitor-to-cart rate: {cart_rate:.1%}")  # Display the visitor-to-cart conversion rate.
print(f"Visitor-to-purchase rate: {purchase_rate:.1%}")  # Display the overall conversion rate.
print(f"Cart-to-purchase rate: {cart_to_purchase_rate:.1%}")  # Display the cart conversion rate.
```

## 18. A/B Test: Compare Two Conversion Rates

An **A/B test** compares an outcome between two groups. For conversion data, start by calculating each group's conversion rate and the difference between them.

The code below computes the observed difference and an approximate standard error using a common independent-proportions formula. It is a teaching example, not a complete experiment analysis: check random assignment, sample size, independence, test design, and a suitable confidence interval or statistical test before making decisions.

```python
a_conversions, a_users = 45, 500  # Record conversions and total users for variant A.
b_conversions, b_users = 60, 500  # Record conversions and total users for variant B.
rate_a = a_conversions / a_users  # Calculate the conversion rate for variant A.
rate_b = b_conversions / b_users  # Calculate the conversion rate for variant B.
difference = rate_b - rate_a  # Calculate the observed B-minus-A difference.
standard_error = np.sqrt(rate_a * (1 - rate_a) / a_users + rate_b * (1 - rate_b) / b_users)  # Approximate the standard error of the difference.
ci_low = difference - 1.96 * standard_error  # Calculate the lower endpoint of an approximate 95% interval.
ci_high = difference + 1.96 * standard_error  # Calculate the upper endpoint of an approximate 95% interval.
print(f"Variant A conversion: {rate_a:.2%}")  # Display A's conversion rate.
print(f"Variant B conversion: {rate_b:.2%}")  # Display B's conversion rate.
print(f"Difference (B - A): {difference:.2%}")  # Display the observed absolute difference in percentage points.
print(f"Approx. 95% interval: [{ci_low:.2%}, {ci_high:.2%}]")  # Display the approximate interval for the difference.
```
