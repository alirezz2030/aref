## 1. Core Data Analysis Vocabulary

Start with the words that appear in analyst job descriptions and day-to-day work.

- **Dataset:** a collection of related data.
- **Record / row:** one entry in a table.
- **Field / column:** one variable in a table.
- **Metric:** a measure used to track performance.
- **KPI:** a key performance indicator linked to an objective.
- **Insight:** a meaningful interpretation of evidence.
- **Finding:** an observation supported by the analysis.
- **Recommendation:** an action proposed based on findings.

**Language pattern:** “The dataset contains 12,000 records and 14 columns.”

```python
# Read each term aloud, then create one sentence from your own project.
vocabulary = {  # Keep the core terms and simple meanings together.
    "dataset": "a collection of related data",  # A complete body of data.
    "record": "one entry or row in a table",  # A single observation.
    "field": "one variable or column in a table",  # An attribute recorded for each row.
    "metric": "a measurement used to track something",  # A measure such as revenue.
    "KPI": "a key measure linked to a business objective",  # A metric important to success.
    "finding": "an observation supported by evidence",  # Something the analysis shows.
    "insight": "an interpretation that helps explain what matters",  # Meaning drawn from findings.
    "recommendation": "an action suggested by the analysis",  # A proposed next step.
}  # Finish the vocabulary dictionary.
for term, meaning in vocabulary.items():  # Go through each term and its meaning.
    print(f"{term.title()}: {meaning}")  # Display the vocabulary in a study-friendly format.
print("\nPractice: Explain the difference between a finding and an insight in your own words.")  # Give a speaking prompt.
```

## 4. Explaining Metrics and KPIs

Use clear formulas and name the denominator when describing a rate.

Sentence patterns:
- “We calculate conversion rate as purchases divided by eligible visitors.”
- “Gross margin is gross profit divided by revenue.”
- “The metric increased from 12% to 15%.”
- “The numerator is… and the denominator is…”
- “This KPI measures…”
- “We exclude cancelled orders from this calculation.”
- “The definition should be consistent across reporting periods.”

Avoid saying “sales went up by 3%” when the change is from 12% to 15%; that is a **3 percentage-point increase** and a **25% relative increase**.

```python
metric_examples = [  # Model clear descriptions of common business metrics.
    "Conversion rate = purchases / eligible visitors.",  # State a rate's denominator.
    "Gross margin = gross profit / revenue.",  # Define margin.
    "Average order value = revenue / number of distinct orders.",  # Explain a ratio using the correct order count.
    "Customer churn rate = customers lost / customers active at the start of the period.",  # State an example churn definition.
    "We exclude cancelled orders and refunds according to the agreed reporting rules.",  # Mention metric rules.
]  # Finish the metric examples.
for sentence in metric_examples:  # Loop through the definitions.
    print("-", sentence)  # Display one definition per line.
old_rate, new_rate = 0.12, 0.15  # Define example rates before and after a change.
print(f"Percentage-point change: {(new_rate-old_rate)*100:.1f} points")  # Calculate the absolute difference in percentage points.
print(f"Relative change: {(new_rate/old_rate-1):.1%}")  # Calculate the relative percentage change.
```

## 6. Describing Trends Over Time

Use trend verbs carefully:

- **increase / rise / grow:** move upward.
- **decrease / fall / decline:** move downward.
- **fluctuate:** move up and down.
- **peak:** reach the highest point.
- **bottom out:** reach the lowest point.
- **stabilize / level off:** stop changing significantly.
- **recover:** rise after a decline.
- **remain flat:** show little or no change.

Useful sentence: “Revenue rose steadily in the first quarter, peaked in March, and declined slightly in April.”

```python
trend_sentences = [  # Practice common ways to describe time-series charts.
    "Revenue rose steadily during the first quarter.",  # Describe a consistent increase.
    "Orders fluctuated throughout the month.",  # Describe movement in both directions.
    "Sales peaked in November.",  # Identify the high point.
    "Customer activity bottomed out in February.",  # Identify the low point.
    "The metric stabilized toward the end of the period.",  # Describe leveling off.
    "Sales recovered after a decline in the previous month.",  # Describe a rebound.
]  # Finish the trend sentence list.
for sentence in trend_sentences:  # Display each example.
    print("-", sentence)  # Print the phrase.
print("\nExercise: Describe a line chart in three sentences: beginning, peak or low, and end.")  # Encourage a structured chart description.
```

## 9. Explaining What a Chart Means

Move from description to interpretation:

1. **Observation:** what the chart directly shows.
2. **Comparison:** how it differs from a target or baseline.
3. **Interpretation:** what the pattern may mean.
4. **Caveat:** what remains uncertain.

Example: “Online revenue increased by 10%, while gross margin declined by 2 percentage points. This may indicate a shift toward discounted products. We should check product mix and returns before changing the promotion strategy.”

```python
chart_interpretation = {  # Separate observation from interpretation.
    "Observation": "Online revenue increased by 10%, while gross margin declined by 2 percentage points.",  # State what the metrics show.
    "Comparison": "Revenue grew, but margin moved in the opposite direction.",  # Describe the contrast.
    "Possible interpretation": "A change in discounts or product mix may explain the pattern.",  # Offer a hypothesis, not a proven cause.
    "Next check": "Compare discounts, product mix, refunds, and variable fulfillment costs.",  # Identify evidence to examine.
}  # Finish the interpretation template.
for stage, sentence in chart_interpretation.items():  # Display each stage.
    print(f"{stage}: {sentence}")  # Print the stage and sentence.
```

## 10. Correlation, Causation, and Uncertainty

Use careful language when evidence does not prove cause:

- “The variables are positively correlated.”
- “We observed an association between…”
- “The data do not establish causation.”
- “One possible explanation is…”
- “The result is consistent with the hypothesis, but it does not prove it.”
- “Additional data are needed to determine whether…”
- “The estimate is uncertain because…”

Avoid saying “X caused Y” based only on a correlation or a comparison between groups that may differ in other ways.

```python
careful_language = {  # Contrast stronger-than-justified claims with better wording.
    "Avoid": "The discount caused revenue to increase.",  # This is a causal claim that may be unsupported.
    "Prefer": "Higher discounts were associated with higher revenue in this dataset.",  # State the observed association.
    "Add caveat": "We need to control for product mix and test the change before claiming a causal effect.",  # Name a validation step.
}  # Finish the phrasing comparison.
for label, sentence in careful_language.items():  # Iterate through the wording examples.
    print(f"{label}: {sentence}")  # Display the careful-language guide.
```

## 13. Explaining Statistics

Statistical English should be precise:
- “The median is less affected by extreme values than the mean.”
- “The 95% confidence interval ranges from… to…”
- “The p-value is… under the null hypothesis.”
- “The result was not statistically significant at the preselected threshold.”
- “The effect may be practically important even if the estimate is uncertain.”
- “The sample may not represent the target population.”

Do not say a p-value is the probability that the null hypothesis is true.

```python
statistics_phrases = [  # Collect precise language for statistical reporting.
    "The median is less affected by extreme values than the mean.",  # Explain a robust statistic.
    "The estimated difference is 3 percentage points, with a 95% confidence interval from 1 to 5 points.",  # Describe an estimate and interval.
    "The result was not statistically significant at the preselected 5% threshold.",  # Use neutral test language.
    "The analysis shows an association, but it does not establish causation.",  # Avoid causal overclaiming.
    "The sample may not represent all customers.",  # State a sampling limitation.
]  # Finish the language list.
for sentence in statistics_phrases:  # Display each sentence.
    print("-", sentence)  # Print the phrase.
```

- Recommendation: “Review discount depth and product mix before expanding online promotions.”

Keep these categories separate so stakeholders can see which statements are evidence and which are interpretation or proposed action.

## 15. Writing an Executive Summary

An executive summary should be understandable without reading the entire notebook. Lead with the answer, then the strongest evidence, business implication, recommendation, and limitation.

Useful opening phrases:
- “The analysis found that…”
- “The main driver appears to be…”
- “Compared with the previous period…”
- “We recommend…”
- “The expected benefit is…, subject to…”
- “The main limitation is…”

Avoid burying the conclusion after several paragraphs of methodology.

```python
executive_summary_template = {  # Create a reusable executive-summary structure.
    "Conclusion": "The analysis found that [main finding].",  # Lead with the answer.
    "Evidence": "This is supported by [metric, comparison, and time period].",  # Give concrete evidence.
    "Business implication": "This matters because [impact on revenue, cost, risk, or customers].",  # Explain importance.
    "Recommendation": "We recommend [specific action] and measuring [success metric].",  # Propose an action and test.
    "Limitation": "The main limitation is [assumption, missing data, or uncertainty].",  # State the caveat.
}  # Finish the template.
for section, sentence in executive_summary_template.items():  # Iterate through the summary structure.
    print(f"{section}: {sentence}")  # Display the section and starter sentence.
```

## 17. Discussing Limitations

A limitation explains what the analysis cannot reliably answer. Common limitations include missing data, limited coverage, small samples, changing definitions, selection bias, unmeasured confounders, and data that are synthetic or out of date.

Useful phrases:
- “The results should be interpreted with caution because…”
- “The dataset does not include…”
- “This analysis is descriptive, not causal.”
- “We could not verify…”
- “The finding may not generalize to…”
- “Further validation is needed before…”

A limitation does not make analysis useless; it helps people use the result appropriately.

```python
limitations = [  # List examples of transparent limitation statements.
    "The results should be interpreted with caution because the dataset covers only six months.",  # Limit time coverage.
    "The dataset does not include marketing costs, so channel profitability is incomplete.",  # Identify a missing business driver.
    "This analysis is descriptive and does not establish causation.",  # Avoid causal overclaiming.
    "The finding may not generalize to other regions or customer groups.",  # Limit generalizability.
    "The source data could not be fully reconciled with the finance system.",  # State validation uncertainty.
]  # Finish the list.
for sentence in limitations:  # Loop through the examples.
    print("-", sentence)  # Display a professional caveat.
```

## 24. Role-Play: Revenue Is Down

In a case interview, do not jump immediately to “increase advertising.” First clarify the metric, period, and comparison; verify the data; then decompose revenue into drivers and segment the decline.

Use phrases such as “I would first validate…,” “I would break this down by…,” and “One hypothesis to test is…”. End with an action, a success metric, and a caveat.

## 25. Practice: Rewrite Weak Analyst English

Clear business English is specific and evidence-based. Avoid vague phrases such as “the chart is going up a lot,” “the data says the campaign works,” or “sales are better.”

Prefer: “Revenue increased by 12% compared with the previous month.” Or: “The treatment group had a higher conversion rate, but we should review the confidence interval and experiment quality before making a rollout decision.”

```python
weak_to_strong = {  # Store examples of clearer professional wording.
    "Weak: The chart is going up a lot.": "Stronger: Revenue increased by 12% compared with the previous month.",  # Replace vague wording with a metric and baseline.
    "Weak: The data says the campaign works.": "Stronger: The campaign group had a higher conversion rate; the result needs uncertainty and experiment-quality checks.",  # Avoid overclaiming causation.
    "Weak: Sales are better.": "Stronger: Revenue rose, although gross margin remained unchanged.",  # Distinguish measures.
    "Weak: We have some bad data.": "Stronger: The customer ID field has missing values in 4% of records.",  # Quantify a quality issue.
    "Weak: I did a dashboard.": "Stronger: I built a Power BI dashboard tracking revenue, gross profit, and margin by region and channel.",  # State output and scope.
}  # Finish the rewrite examples.
for weak, strong in weak_to_strong.items():  # Review each pair.
    print(weak)  # Display the less precise sentence.
    print("  ", strong, "\n")  # Display the clearer version.
```
