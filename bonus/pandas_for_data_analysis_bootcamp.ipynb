## 2. Inspecting a Dataset

Always inspect a dataset before analyzing it. `head()`, `tail()`, `shape`, `columns`, `dtypes`, `info()`, and `describe()` reveal its structure, data types, and basic numeric summary.

`isna().sum()` counts missing values per column. Early inspection can prevent incorrect calculations caused by missing or wrongly typed fields.

```python
print(df.head(3))  # Display the first three rows.
print("\nShape:", df.shape)  # Show how many rows and columns exist.
print("\nColumn names:", df.columns.tolist())  # Convert the column index to a readable list.
print("\nData types:\n", df.dtypes)  # Show the type of each column.
print("\nMissing values:\n", df.isna().sum())  # Count missing values in each column.
print("\nNumeric summary:\n", df.describe())  # Summarize numeric columns with count, mean, quartiles, and extremes.
```

## 3. Selecting Columns

Use `df["column"]` to select one column as a Series, and `df[["a", "b"]]` to select multiple columns as a DataFrame.

Selecting only the fields needed for a task makes analysis easier to read and can reduce unnecessary memory use on larger datasets.

```python
product_column = df["product"]  # Select the product column as a Series.
small_table = df[["order_id", "region", "product", "units"]]  # Select four columns as a smaller DataFrame.
print(product_column.head())  # Display the first five product values.
print("\nSelected table:\n", small_table.head())  # Display the first rows of the smaller table.
print("\nSingle-column type:", type(product_column))  # Show that one column returns a Series.
print("Multiple-column type:", type(small_table))  # Show that multiple columns return a DataFrame.
```

## 4. Selecting Rows with loc and iloc

`loc` selects by labels (index labels and column names); `iloc` selects by integer position.

- `df.loc[rows, columns]` uses labels and label-based slices include the ending label.
- `df.iloc[rows, columns]` uses zero-based integer positions and standard Python slice rules.

These are two of the most important indexing methods in pandas.

```python
first_row_by_label = df.loc[0, ["order_id", "product"]]  # Select index label 0 and two named columns.
first_three_rows = df.iloc[0:3, :]  # Select the first three rows by integer position.
specific_cells = df.iloc[1:4, 0:3]  # Select rows 1 through 3 and columns 0 through 2 by position.
print("One row by label:\n", first_row_by_label)  # Display the selected Series.
print("\nFirst three rows:\n", first_three_rows)  # Display the first three records.
print("\nSelected slice:\n", specific_cells)  # Display the selected rectangular subset.
```

## 5. Filtering Rows

Filtering is similar to SQL's `WHERE` clause. A condition creates a Boolean mask, which can select matching rows.

Use `&` for AND and `|` for OR between pandas conditions. Put every comparison in parentheses. Use `.isin([...])` to test membership in a list of values.

```python
north_orders = df[df["region"] == "North"]  # Keep rows whose region equals North.
high_quantity = df[(df["units"] >= 3) & (df["product"].str.strip().str.lower() == "phone")]  # Keep phone orders with at least three units.
selected_regions = df[df["region"].isin(["North", "East"])]  # Keep orders from either North or East.
print("North orders:\n", north_orders[["order_id", "product", "units"]])  # Display key details for North orders.
print("\nHigh-quantity phone orders:\n", high_quantity[["order_id", "region", "units"]])  # Display rows matching both conditions.
print("\nNorth or East rows:", len(selected_regions))  # Count rows from the selected regions.
```

## 6. Sorting Data

`sort_values()` sorts rows by one or more columns. Use `ascending=False` to sort from largest to smallest, and pass a list of columns to apply multi-column sorting.

Sorting helps rank orders, customers, products, or regions. Choose the sort key that matches the question you are trying to answer.

```python
sorted_by_price = df.sort_values("unit_price", ascending=False)  # Sort rows from highest unit price to lowest.
sorted_by_region_date = df.sort_values(["region", "date"], ascending=[True, True], na_position="last")  # Sort by region, then date, with missing regions last.
print(sorted_by_price[["order_id", "product", "unit_price"]].head(5))  # Display the five most expensive records.
print("\nSorted by region and date:\n", sorted_by_region_date[["region", "date", "order_id"]].head(8))  # Preview the multi-column sorting result.
```

## 7. Handling Missing Values

Missing data requires a reasoned choice. You can remove rows with `dropna()`, fill values with `fillna()`, or investigate why a value is absent.

This example fills missing `units` with the median and missing regions with `"Unknown"`. These are demonstration choices, not universal rules; the right approach depends on the meaning and origin of the missing data.

```python
clean_df = df.copy()  # Copy the raw dataset so it remains unchanged.
print("Missing before cleaning:\n", clean_df.isna().sum())  # Count missing values before filling.
units_median = clean_df["units"].median()  # Calculate the median of available unit values.
clean_df["units"] = clean_df["units"].fillna(units_median)  # Fill missing unit counts with the median.
clean_df["region"] = clean_df["region"].fillna("Unknown")  # Mark missing regions as Unknown instead of guessing.
print("\nMissing after cleaning:\n", clean_df.isna().sum())  # Verify which missing values remain.
```

## 8. Removing Duplicates

`duplicated()` flags rows repeated earlier, while `drop_duplicates()` removes duplicates. By default, pandas compares all columns and keeps the first copy.

In real datasets, decide what defines a duplicate. Repeated customer IDs may be normal in a transaction table, while a duplicated transaction ID may indicate repeated data.

```python
duplicate_rows = clean_df[clean_df.duplicated()]  # Select exact duplicate rows.
print("Exact duplicate count:", len(duplicate_rows))  # Count exact duplicate records.
dedup_df = clean_df.drop_duplicates().copy()  # Remove exact duplicate rows and make a clean copy.
print("Rows before:", len(clean_df))  # Display the original number of rows.
print("Rows after:", len(dedup_df))  # Display the number of rows after deduplication.
print("Duplicate rows remaining:", int(dedup_df.duplicated().sum()))  # Verify that exact duplicates were removed.
```

## 9. Converting Data Types

Dates stored as strings cannot use all datetime functionality. Numeric values stored as text can also cause calculation problems.

Use `pd.to_datetime()` for date values and `pd.to_numeric()` for numeric columns. `errors="coerce"` converts unparseable values to missing values, which should then be reviewed.

```python
typed_df = dedup_df.copy()  # Start from the de-duplicated dataset.
typed_df["date"] = pd.to_datetime(typed_df["date"], errors="coerce")  # Convert date strings to pandas datetime values.
typed_df["units"] = pd.to_numeric(typed_df["units"], errors="coerce")  # Ensure units are numeric.
typed_df["unit_price"] = pd.to_numeric(typed_df["unit_price"], errors="coerce")  # Ensure unit prices are numeric.
print(typed_df[["date", "units", "unit_price"]].dtypes)  # Verify the converted column types.
print("\nConverted values:\n", typed_df[["date", "units", "unit_price"]].head())  # Preview the converted columns.
```

## 10. Cleaning Text Columns

Text fields often contain extra spaces, inconsistent capitalization, or spelling variants. The pandas `.str` accessor supports vectorized string methods such as `strip()`, `lower()`, `upper()`, `title()`, and `replace()`.

Standardizing categories before grouping prevents values such as `"Phone"`, `"PHONE"`, and `" phone "` from being counted as separate products.

```python
text_df = typed_df.copy()  # Create a new working copy for text cleaning.
text_df["product"] = text_df["product"].str.strip().str.title()  # Remove surrounding spaces and standardize capitalization.
text_df["customer_type"] = text_df["customer_type"].str.strip().str.title()  # Normalize customer-type text.
print("Product labels:", sorted(text_df["product"].unique()))  # Show the unique product labels after cleaning.
print("Customer types:", sorted(text_df["customer_type"].unique()))  # Show standardized customer types.
print(text_df[["order_id", "product", "customer_type"]].head())  # Preview the cleaned text columns.
```

## 11. Working with Dates

Once a column is datetime-typed, the `.dt` accessor extracts date components such as year, month, day, weekday, or quarter.

Date features help answer questions about seasonality, monthly performance, and differences across weekdays. Ensure the source date format and time zone are correctly understood before analyzing real timestamps.

```python
date_df = text_df.copy()  # Make a working copy for date-based analysis.
date_df["year"] = date_df["date"].dt.year  # Extract the year component.
date_df["month"] = date_df["date"].dt.month  # Extract the numeric month.
date_df["month_name"] = date_df["date"].dt.month_name()  # Extract the month name for reports.
date_df["weekday"] = date_df["date"].dt.day_name()  # Extract the day-of-week name.
date_df["quarter"] = date_df["date"].dt.quarter  # Extract the calendar quarter.
print(date_df[["date", "year", "month", "month_name", "weekday", "quarter"]].head())  # Display the derived date features.
```

## 12. Creating Calculated Columns

A calculated column derives a new field from existing values. Vectorized expressions are usually clearer and faster than looping over rows.

For sales data, revenue often begins as `units × unit_price`. A total such as this should reflect the organization's definition of revenue, including whether discounts, taxes, refunds, or shipping apply.

```python
sales_df = date_df.copy()  # Create a working table for calculated business metrics.
sales_df["revenue"] = sales_df["units"] * sales_df["unit_price"]  # Calculate gross revenue per order before any discounts or taxes.
sales_df["revenue_per_unit"] = sales_df["revenue"] / sales_df["units"]  # Calculate average revenue per unit for each order.
sales_df["is_large_order"] = sales_df["revenue"] >= 1500  # Flag orders with revenue of at least 1500.
print(sales_df[["order_id", "units", "unit_price", "revenue", "is_large_order"]].head())  # Preview the new calculated columns.
```

## 13. Mapping and Replacing Values

`map()` replaces values according to a dictionary and returns missing values for unmapped keys. `replace()` can replace particular values while leaving other values intact.

Use these methods for category mapping, code translation, and standardizing known variants. Always check for unmapped or unexpected categories after applying a mapping.

```python
category_map = {"Laptop": "Electronics", "Phone": "Electronics", "Tablet": "Electronics"}  # Define product-to-category mapping.
sales_df["category"] = sales_df["product"].map(category_map)  # Assign a category based on the product name.
sales_df["customer_type"] = sales_df["customer_type"].replace({"New": "First-time", "Returning": "Repeat"})  # Rename customer-type labels.
print(sales_df[["product", "category", "customer_type"]].drop_duplicates())  # Display unique combinations after mapping.
print("Unmapped products:", sales_df.loc[sales_df["category"].isna(), "product"].unique())  # Check for product names missing from the mapping.
```

## 14. Counting Categories with value_counts

`value_counts()` counts the frequency of each distinct value. It is often one of the quickest ways to understand categorical columns.

Use `normalize=True` to display proportions instead of raw counts, and `dropna=False` when you want missing values included in the count.

```python
print("Orders by product:\n", sales_df["product"].value_counts())  # Count orders by standardized product label.
print("\nProduct proportions:\n", sales_df["product"].value_counts(normalize=True).round(3))  # Convert counts to proportions.
print("\nRegions including missing values:\n", sales_df["region"].value_counts(dropna=False))  # Count regions while retaining missing values.
print("\nOrders by customer type:\n", sales_df["customer_type"].value_counts())  # Count orders for each customer group.
```

## 15. Grouping and Aggregation with groupby

`groupby()` divides records into groups and calculates statistics for each group. It plays a role similar to SQL's `GROUP BY`.

With `.agg()`, you can name multiple metrics and compute them in one operation. Check whether you need to count records (`count`) or distinct identifiers (`nunique`), as these answer different questions.

```python
region_summary = sales_df.groupby("region", dropna=False, as_index=False).agg(  # Group records by region and calculate summary metrics.
    orders=("order_id", "nunique"),  # Count distinct order IDs for each region.
    units_sold=("units", "sum"),  # Add units sold within each region.
    total_revenue=("revenue", "sum"),  # Sum revenue within each region.
    average_order_revenue=("revenue", "mean"),  # Calculate average revenue per record within each region.
)  # Finish the aggregation specification.
region_summary = region_summary.sort_values("total_revenue", ascending=False)  # Rank regions by their revenue totals.
print(region_summary)  # Display the regional performance summary.
```

## 16. Filtering with query()

The `query()` method filters a DataFrame using a readable expression string. It can be convenient for simple conditions on columns.

For more complex expressions, ordinary Boolean indexing is often easier to debug. Avoid inserting untrusted strings into expressions, and use normal indexing when column names contain spaces or unusual characters.

```python
high_value_orders = sales_df.query("revenue >= 1000 and units >= 2")  # Filter for revenue of at least 1000 and two or more units.
north_or_east = sales_df.query("region in ['North', 'East']")  # Select records from either North or East.
print(high_value_orders[["order_id", "region", "product", "units", "revenue"]])  # Display the high-value records.
print("\nNorth/East order count:", len(north_or_east))  # Count the rows from the two selected regions.
```

## 17. Pivot Tables

A pivot table summarizes one or more measures across two categorical dimensions. For example, it can show revenue by region and product.

`pivot_table()` supports aggregation through `aggfunc`. Use `fill_value=0` only when a missing combination should be interpreted as zero activity, not as unknown data.

```python
pivot = pd.pivot_table(  # Build a cross-tabulated summary.
    sales_df,  # Provide the source records.
    index="region",  # Put each region on a row.
    columns="product",  # Create one column per product.
    values="revenue",  # Summarize the revenue field.
    aggfunc="sum",  # Add revenue for each region-product combination.
    fill_value=0,  # Display zero for combinations with no recorded revenue.
)  # Finish the pivot-table specification.
print(pivot)  # Show revenue by region and product.
print("\nGrand totals by product:\n", pivot.sum(axis=0))  # Sum each product column across regions.
```

## 18. Cross-Tabulation with crosstab

`pd.crosstab()` counts combinations of categories. It is useful for checking patterns such as customer type by region or product by weekday.

Use `normalize=` to calculate proportions. Choose the normalization direction carefully because proportions of all records differ from proportions within each row or column.

```python
counts = pd.crosstab(sales_df["region"], sales_df["customer_type"])  # Count records for each region/customer-type combination.
row_percentages = pd.crosstab(sales_df["region"], sales_df["customer_type"], normalize="index")  # Calculate customer-type proportions within each region.
print("Raw counts:\n", counts)  # Display the cross-tabulated counts.
print("\nProportions within each region:\n", row_percentages.round(2))  # Display proportions that sum to one per region when data exists.
```

## 19. Merging DataFrames

`merge()` combines tables using shared keys, much like SQL joins. Common join types include `inner`, `left`, `right`, and `outer`.

Check key uniqueness before merging. If a key occurs multiple times on both sides, a many-to-many join can multiply rows unexpectedly. The `validate` parameter can help catch incorrect join assumptions.

```python
targets = pd.DataFrame({  # Create a reference table with business targets.
    "region": ["North", "South", "East", "West", "Unknown"],  # Define one key row per region.
    "revenue_target": [2500, 3000, 2200, 1800, 500],  # Define the target revenue for each region.
})  # Finish the target table.
joined_summary = region_summary.merge(targets, on="region", how="left", validate="one_to_one")  # Join each regional summary to its target.
joined_summary["target_gap"] = joined_summary["total_revenue"] - joined_summary["revenue_target"]  # Calculate revenue minus target.
joined_summary["target_met"] = joined_summary["target_gap"] >= 0  # Identify regions at or above target.
print(joined_summary[["region", "total_revenue", "revenue_target", "target_gap", "target_met"]])  # Display target performance.
```

## 20. Concatenating DataFrames

`pd.concat()` stacks DataFrames vertically or side-by-side. Vertical concatenation is useful when combining monthly exports that share the same columns.

By default, pandas aligns columns by name. Use `ignore_index=True` when a fresh sequential index is needed. Check for schema differences before appending files from different sources.

```python
january = sales_df.iloc[:4].copy()  # Take a small slice to represent a January export.
february = sales_df.iloc[4:8].copy()  # Take a second slice to represent another export.
combined = pd.concat([january, february], ignore_index=True)  # Stack rows and assign a fresh index.
side_by_side = pd.concat([january[["order_id"]].reset_index(drop=True), january[["revenue"]].reset_index(drop=True)], axis=1)  # Combine two aligned columns horizontally.
print("Vertically combined shape:", combined.shape)  # Show the combined row and column count.
print(combined[["order_id", "date", "product", "revenue"]])  # Preview the combined records.
print("\nHorizontal combination:\n", side_by_side.head())  # Preview the two side-by-side columns.
```

## 21. Binning Numeric Values with cut

`pd.cut()` divides continuous numeric values into intervals. Analysts use bins to create groups such as low, medium, and high order values.

Choose boundaries that fit the business problem. Binning loses information compared with retaining the original numeric values, so keep the original measure if it may be needed later.

```python
sales_df["revenue_band"] = pd.cut(  # Create a categorical revenue-band column.
    sales_df["revenue"],  # Use order revenue as the value to categorize.
    bins=[-np.inf, 500, 1000, 2000, np.inf],  # Define the upper boundaries for four intervals.
    labels=["Low", "Medium", "High", "Very High"],  # Label the resulting intervals.
    right=True,  # Include each interval's right boundary.
)  # Finish the binning operation.
print(sales_df[["order_id", "revenue", "revenue_band"]].head(10))  # Preview the assigned revenue bands.
print("\nOrders per band:\n", sales_df["revenue_band"].value_counts(sort=False))  # Count records in each band in interval order.
```

## 22. Applying Functions: apply()

`apply()` runs a function along a Series or DataFrame axis. It is flexible, but for simple arithmetic or conditional rules, vectorized pandas operations are often faster and easier to read.

This example uses `apply()` to label a single revenue value. For larger workflows, prefer vectorized expressions such as `np.where()` or `pd.cut()` when they can express the same rule.

```python
def label_revenue(value):  # Define a function that labels a revenue value.
    if pd.isna(value):  # Check whether the revenue value is missing.
        return "Unknown"  # Keep missing revenue clearly labeled.
    if value >= 1500:  # Test the highest revenue threshold first.
        return "High"  # Label large revenue values High.
    if value >= 700:  # Test the middle threshold.
        return "Medium"  # Label medium revenue values Medium.
    return "Low"  # Label all remaining non-missing values Low.

sales_df["revenue_label"] = sales_df["revenue"].apply(label_revenue)  # Apply the function to each revenue value.
print(sales_df[["order_id", "revenue", "revenue_label"]].head(10))  # Display example labels.
print("\nLabel counts:\n", sales_df["revenue_label"].value_counts(dropna=False))  # Count records in each label.
```

## 23. Renaming and Dropping Columns

Use `rename()` to change column names and `drop()` to remove fields or rows. Keeping column names consistent can make multi-source analysis easier.

Prefer creating a transformed copy when you still need the original raw dataset. Document why a column is removed so the analysis remains reproducible.

```python
renamed_df = sales_df.rename(columns={"unit_price": "price_per_unit", "customer_type": "customer_segment"})  # Rename selected columns in a new DataFrame.
reduced_df = renamed_df.drop(columns=["revenue_per_unit", "is_large_order"])  # Remove derived columns that are not needed for this view.
print("Renamed columns:", renamed_df.columns.tolist())  # Display the renamed schema.
print("Reduced columns:", reduced_df.columns.tolist())  # Display the schema after dropping the selected columns.
print("\nPreview:\n", reduced_df.head(3))  # Preview the transformed table.
```

## 24. Setting and Resetting an Index

A DataFrame index labels rows. It may be a simple sequence, an identifier, or a datetime index for time-series analysis.

`set_index()` moves a column into the index; `reset_index()` returns the index to a regular column and creates a default integer index. An index is not automatically a unique key unless you verify it.

```python
indexed_df = sales_df.set_index("order_id")  # Use order_id as the row index.
selected_order = indexed_df.loc[2005]  # Retrieve a record by its order ID label.
reset_df = indexed_df.reset_index()  # Move order_id back into a regular column.
print("Selected order:\n", selected_order[["date", "region", "product", "revenue"]])  # Display selected details.
print("\nIndex name:", indexed_df.index.name)  # Display the index label.
print("Reset table columns:", reset_df.columns.tolist())  # Verify that order_id is a regular column again.
```

## 25. Rolling Calculations

A rolling calculation summarizes a moving window of nearby observations, such as a three-day average. It can smooth short-term fluctuations and reveal trends.

Sort data chronologically before applying rolling calculations. Decide what to do when the window has too few observations; `min_periods` controls the minimum required count.

```python
daily = sales_df.groupby("date", as_index=False)["revenue"].sum()  # Aggregate all orders into one revenue total per date.
daily = daily.sort_values("date")  # Sort dates from earliest to latest.
daily["rolling_3_day_mean"] = daily["revenue"].rolling(window=3, min_periods=1).mean()  # Compute the mean of the current and previous two date rows.
print(daily[["date", "revenue", "rolling_3_day_mean"]])  # Display daily revenue and its rolling average.
```

## 26. Resampling Time Series

Resampling converts datetime-indexed observations to a new time frequency. For example, daily records can be aggregated into monthly totals.

The date column must be converted to datetime and set as the index before calling `resample()`. Choose a suitable aggregation: `sum` for totals, `mean` for averages, or another method matching the metric's meaning.

```python
dated_sales = sales_df.dropna(subset=["date"]).set_index("date").sort_index()  # Use valid dates as a sorted datetime index.
monthly_revenue = dated_sales["revenue"].resample("MS").sum()  # Sum revenue by calendar month beginning.
monthly_orders = dated_sales["order_id"].resample("MS").nunique()  # Count distinct orders in each month.
monthly_report = pd.DataFrame({"revenue": monthly_revenue, "orders": monthly_orders})  # Combine monthly metrics in one table.
print(monthly_report)  # Display revenue and order counts by month.
```

## 27. Correlation Between Numeric Columns

`corr()` computes pairwise correlations between numeric columns. Pearson correlation measures linear association and ranges from -1 to 1.

Correlation does not prove causation. Interpret results with domain knowledge, sample size, possible confounders, and visual checks. Correlations among derived variables can be mechanically high because they share components.

```python
numeric_columns = sales_df[["units", "unit_price", "revenue"]]  # Select numeric columns for the example.
correlation_matrix = numeric_columns.corr()  # Calculate the pairwise Pearson correlation coefficients.
print(correlation_matrix.round(2))  # Display the matrix rounded to two decimal places.
print("\nUnits vs revenue:", round(sales_df["units"].corr(sales_df["revenue"]), 3))  # Display one selected correlation.
print("Reminder: correlation is association, not proof of cause.")  # Keep the interpretation appropriately cautious.
```

## 28. Ranking Records

`rank()` assigns relative ranks to values. You can rank from largest to smallest, choose how ties are treated, and rank within groups using `groupby()`.

Ranking is useful for leaderboards, identifying top customers, and comparing products within a region. If a dataset contains ties, choose a tie method that matches the reporting requirement.

```python
sales_df["revenue_rank"] = sales_df["revenue"].rank(method="dense", ascending=False)  # Rank all records by revenue with dense ranks.
sales_df["region_revenue_rank"] = sales_df.groupby("region", dropna=False)["revenue"].rank(method="min", ascending=False)  # Rank each order within its region.
top_orders = sales_df.sort_values(["revenue", "order_id"], ascending=[False, True]).head(5)  # Sort records and keep the five highest-revenue orders.
print(top_orders[["order_id", "region", "product", "revenue", "revenue_rank"]])  # Display the top orders and their overall ranks.
print("\nRegion ranks preview:\n", sales_df[["order_id", "region", "revenue", "region_revenue_rank"]].head())  # Display region-specific rankings.
```

This mini project combines cleaning, calculated columns, grouping, ranking, joining targets, and filtering into one business-oriented workflow.

A concise sales report might answer:
- How much revenue did each region generate?
- How many distinct orders and units were recorded?
- Which regions met their targets?
- Which region had the highest revenue?

The sample dataset is small and educational; real reports should validate definitions and data quality first.

```python
project_df = df.copy()  # Start again from the raw data to show a complete workflow.
project_df = project_df.drop_duplicates().copy()  # Remove exact duplicate rows.
project_df["date"] = pd.to_datetime(project_df["date"], errors="coerce")  # Convert dates to datetime values.
project_df["region"] = project_df["region"].fillna("Unknown")  # Label missing regions explicitly.
project_df["units"] = project_df["units"].fillna(project_df["units"].median())  # Impute missing unit counts using the median for this demonstration.
project_df["product"] = project_df["product"].str.strip().str.title()  # Standardize product labels.
project_df["revenue"] = project_df["units"] * project_df["unit_price"]  # Calculate revenue for every order.
report = project_df.groupby("region", as_index=False).agg(  # Summarize the cleaned records by region.
    orders=("order_id", "nunique"),  # Count distinct orders.
    units_sold=("units", "sum"),  # Sum units sold.
    revenue=("revenue", "sum"),  # Sum revenue.
    avg_order_revenue=("revenue", "mean"),  # Calculate mean revenue per order record.
)  # Finish the regional aggregation.
targets = pd.DataFrame({"region": ["North", "South", "East", "West", "Unknown"], "target": [2500, 3000, 2200, 1800, 500]})  # Create an example target table.
report = report.merge(targets, on="region", how="left", validate="one_to_one")  # Attach the target for each region.
report["gap_to_target"] = report["revenue"] - report["target"]  # Calculate the amount above or below target.
report["target_status"] = np.where(report["gap_to_target"] >= 0, "Met", "Not met")  # Label target attainment.
report = report.sort_values("revenue", ascending=False)  # Rank regions by revenue.
print(report.round(2))  # Display the final regional report rounded to two decimal places.
print("\nHighest-revenue region:", report.iloc[0]["region"])  # Name the region at the top of the report.
print("Overall revenue:", round(project_df["revenue"].sum(), 2))  # Display overall revenue across cleaned orders.
```
