## 1. Linear Search

**Idea:** Check each element one by one until the target is found.

- **Time complexity:** `O(n)` in the worst case.
- **Space complexity:** `O(1)` for this iterative implementation.
- **Best for:** Small or unsorted lists.

Linear search works even when the list is not sorted.

```python
def linear_search(items, target):
    for index, item in enumerate(items):
        if item == target:
            return index
    return -1

numbers = [12, 5, 8, 20, 3]
print(linear_search(numbers, 20))  # 3
print(linear_search(numbers, 99))  # -1
```

## 2. Binary Search

**Idea:** Repeatedly divide the search range in half.

- **Requirement:** The list must already be sorted.
- **Time complexity:** `O(log n)`.
- **Space complexity:** `O(1)` for this iterative version.

At each step, compare the target with the middle element and discard the half that cannot contain it.

```python
def binary_search(sorted_items, target):
    left, right = 0, len(sorted_items) - 1

    while left <= right:
        middle = (left + right) // 2

        if sorted_items[middle] == target:
            return middle
        if sorted_items[middle] < target:
            left = middle + 1
        else:
            right = middle - 1

    return -1

numbers = [3, 5, 8, 12, 20, 27, 31]  # sorted
print(binary_search(numbers, 20))  # 4
print(binary_search(numbers, 99))  # -1
```

## 3. Bubble Sort

**Idea:** Compare neighboring elements and swap them when they are in the wrong order. Large values gradually move toward the end.

- **Time complexity:** `O(n²)` in the average and worst cases.
- **Space complexity:** `O(1)` auxiliary space.
- **Use:** Mainly for learning; it is inefficient for large lists.

The `swapped` flag lets the algorithm stop early if a pass makes no swaps.

```python
def bubble_sort(items):
    result = items.copy()
    n = len(result)

    for end in range(n - 1, 0, -1):
        swapped = False
        for i in range(end):
            if result[i] > result[i + 1]:
                result[i], result[i + 1] = result[i + 1], result[i]
                swapped = True
        if not swapped:
            break

    return result

print(bubble_sort([5, 1, 4, 2, 8]))  # [1, 2, 4, 5, 8]
```

## 4. Selection Sort

**Idea:** Find the smallest element in the unsorted part and swap it into the next position.

- **Time complexity:** `O(n²)`.
- **Space complexity:** `O(1)` auxiliary space if sorting in place.
- **Use:** Useful for understanding sorting, but usually not for production.

This example returns a sorted copy so the original list stays unchanged.

```python
def selection_sort(items):
    result = items.copy()

    for start in range(len(result)):
        smallest = start
        for i in range(start + 1, len(result)):
            if result[i] < result[smallest]:
                smallest = i
        result[start], result[smallest] = result[smallest], result[start]

    return result

print(selection_sort([64, 25, 12, 22, 11]))  # [11, 12, 22, 25, 64]
```

## 5. Insertion Sort

**Idea:** Build a sorted section one item at a time by inserting each new item into its correct position.

- **Time complexity:** `O(n²)` average/worst case; `O(n)` best case for already sorted data.
- **Space complexity:** `O(1)` auxiliary space for in-place sorting.
- **Best for:** Small lists or nearly sorted data.

This implementation sorts a copy of the input.

```python
def insertion_sort(items):
    result = items.copy()

    for i in range(1, len(result)):
        current = result[i]
        j = i - 1

        while j >= 0 and result[j] > current:
            result[j + 1] = result[j]
            j -= 1

        result[j + 1] = current

    return result

print(insertion_sort([9, 5, 1, 4, 3]))  # [1, 3, 4, 5, 9]
```

## 6. Merge Sort

**Idea:** Divide the list into halves, sort each half recursively, then merge the sorted halves.

- **Time complexity:** `O(n log n)`.
- **Space complexity:** Typically `O(n)` additional space.
- **Approach:** Divide and conquer.

Merge sort gives predictable performance and is useful for learning recursive algorithms.

```python
def merge_sort(items):
    if len(items) <= 1:
        return items.copy()

    middle = len(items) // 2
    left = merge_sort(items[:middle])
    right = merge_sort(items[middle:])

    merged = []
    i = j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            merged.append(left[i])
            i += 1
        else:
            merged.append(right[j])
            j += 1

    merged.extend(left[i:])
    merged.extend(right[j:])
    return merged

print(merge_sort([38, 27, 43, 3, 9, 82, 10]))
```

## 7. Quick Sort

**Idea:** Choose a pivot, split values into those smaller than or equal to the pivot and those larger, then sort the parts.

- **Average time complexity:** `O(n log n)`.
- **Worst-case time complexity:** `O(n²)` for poor pivot choices.
- **Space:** This simple recursive version creates new lists, so it uses additional memory.

This readable version prioritizes clarity over in-place performance.

```python
def quick_sort(items):
    if len(items) <= 1:
        return items.copy()

    pivot = items[len(items) // 2]
    smaller = [x for x in items if x < pivot]
    equal = [x for x in items if x == pivot]
    larger = [x for x in items if x > pivot]

    return quick_sort(smaller) + equal + quick_sort(larger)

print(quick_sort([10, 7, 8, 9, 1, 5, 5]))  # [1, 5, 5, 7, 8, 9, 10]
```

## 8. Recursion

**Idea:** A function solves a problem by calling itself on a smaller version of that problem.

Every recursive function needs:
1. A **base case** that stops the recursion.
2. A **recursive case** that makes progress toward the base case.

Factorial example: `5! = 5 × 4 × 3 × 2 × 1 = 120`.

- **Time complexity:** `O(n)` for this factorial implementation.
- **Space complexity:** `O(n)` call-stack space.

```python
def factorial(n):
    if n < 0:
        raise ValueError("n must be non-negative")
    if n in (0, 1):  # base case
        return 1
    return n * factorial(n - 1)  # recursive case

print(factorial(5))  # 120
print(factorial(0))  # 1
```

## 9. Breadth-First Search (BFS)

**Idea:** Visit graph nodes level by level, using a queue.

- **Time complexity:** `O(V + E)` with an adjacency-list graph.
- **Space complexity:** `O(V)`.
- **Useful for:** Finding the shortest path in an unweighted graph.

`V` is the number of vertices and `E` is the number of edges. This example returns nodes in BFS visitation order.

```python
from collections import deque

def bfs(graph, start):
    visited = {start}
    queue = deque([start])
    order = []

    while queue:
        node = queue.popleft()
        order.append(node)

        for neighbor in graph.get(node, []):
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)

    return order

graph = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": [],
    "E": [],
    "F": []
}
print(bfs(graph, "A"))  # ['A', 'B', 'C', 'D', 'E', 'F']
```

## 10. Depth-First Search (DFS)

**Idea:** Explore one branch as far as possible before returning to explore another. This iterative version uses a stack.

- **Time complexity:** `O(V + E)`.
- **Space complexity:** `O(V)`.
- **Useful for:** Reachability, connected components, and graph exploration.

The reversed neighbor order helps preserve the left-to-right order shown in the example.

```python
def dfs(graph, start):
    visited = set()
    stack = [start]
    order = []

    while stack:
        node = stack.pop()
        if node in visited:
            continue

        visited.add(node)
        order.append(node)

        for neighbor in reversed(graph.get(node, [])):
            if neighbor not in visited:
                stack.append(neighbor)

    return order

graph = {
    "A": ["B", "C"],
    "B": ["D", "E"],
    "C": ["F"],
    "D": [],
    "E": [],
    "F": []
}
print(dfs(graph, "A"))
```

## 11. Hashing

**Idea:** A hash function maps a key to a hash value. Python dictionaries and sets use hashing internally to support fast average-case lookup.

- **Average dictionary lookup/insert:** `O(1)`.
- **Worst case:** Can be `O(n)`.
- **Requirement:** Dictionary keys must be hashable, such as strings, numbers, or tuples of hashable values.

In everyday Python, use `dict` and `set` rather than implementing a hash table yourself.

```python
# Count how often each word appears.
words = ["python", "sql", "python", "data", "sql", "python"]
counts = {}

for word in words:
    counts[word] = counts.get(word, 0) + 1

print(counts)  # {'python': 3, 'sql': 2, 'data': 1}

# A set stores unique hashable values.
unique_words = set(words)
print(unique_words)
print("python" in counts)  # Average-case O(1) lookup
```

## 12. Two Pointers

**Idea:** Keep two indexes and move them strategically to solve a problem efficiently.

This example checks whether a **sorted** list contains two numbers whose sum equals the target.

- **Time complexity:** `O(n)`.
- **Space complexity:** `O(1)`.
- **Requirement for this approach:** Input is sorted in ascending order.

If the sum is too small, move the left pointer right; if too large, move the right pointer left.

```python
def pair_with_sum(sorted_numbers, target):
    left, right = 0, len(sorted_numbers) - 1

    while left < right:
        total = sorted_numbers[left] + sorted_numbers[right]

        if total == target:
            return sorted_numbers[left], sorted_numbers[right]
        if total < target:
            left += 1
        else:
            right -= 1

    return None

numbers = [1, 2, 4, 6, 9, 11]
print(pair_with_sum(numbers, 15))  # (4, 11)
print(pair_with_sum(numbers, 100))  # None
```

## 13. Sliding Window

**Idea:** Maintain a moving range of elements instead of recalculating each range from scratch.

This example finds the maximum sum of any contiguous window of exactly `k` numbers.

- **Time complexity:** `O(n)`.
- **Space complexity:** `O(1)`.
- **Useful for:** Contiguous subarray and substring problems.

The first window is calculated once; then each step removes the value leaving the window and adds the new value entering it.

```python
def max_window_sum(numbers, k):
    if k <= 0 or k > len(numbers):
        raise ValueError("k must be between 1 and len(numbers)")

    window_sum = sum(numbers[:k])
    best = window_sum

    for right in range(k, len(numbers)):
        window_sum += numbers[right] - numbers[right - k]
        best = max(best, window_sum)

    return best

print(max_window_sum([2, 1, 5, 1, 3, 2], 3))  # 9 (5 + 1 + 3)
```

## 14. Greedy Algorithms

**Idea:** At each step, choose the option that looks best locally, hoping these choices produce a globally good solution.

This example solves the **activity-selection problem**: choose the maximum number of non-overlapping activities. Each activity is `(start, finish)`.

- **Time complexity:** `O(n log n)` because of sorting.
- **Key idea:** Always choose the available activity that finishes earliest.
- **Important:** Greedy strategies do not solve every optimization problem optimally; correctness depends on the problem's structure.

```python
def select_activities(activities):
    # Sort by finish time.
    activities = sorted(activities, key=lambda activity: activity[1])
    selected = []
    last_finish = float("-inf")

    for start, finish in activities:
        if start >= last_finish:
            selected.append((start, finish))
            last_finish = finish

    return selected

activities = [(1, 3), (2, 4), (3, 5), (0, 6), (5, 7), (8, 9)]
print(select_activities(activities))  # [(1, 3), (3, 5), (5, 7), (8, 9)]
```

## 15. Dynamic Programming

**Idea:** Solve overlapping subproblems once and store their answers so they can be reused.

This example computes Fibonacci numbers using **bottom-up tabulation**.

- **Time complexity:** `O(n)`.
- **Space complexity:** `O(n)` for the table.
- **Why not plain recursion?** The naive recursive version repeats the same calculations many times.

Dynamic programming is especially useful when a problem has overlapping subproblems and optimal substructure.

```python
def fibonacci(n):
    if n < 0:
        raise ValueError("n must be non-negative")
    if n <= 1:
        return n

    dp = [0] * (n + 1)
    dp[1] = 1

    for i in range(2, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]

    return dp[n]

print(fibonacci(10))  # 55
print([fibonacci(i) for i in range(8)])  # [0, 1, 1, 2, 3, 5, 8, 13]
```
