## 1. Two Sum (Hash Map)

**Pattern:** Hash map lookup. Find two values that add up to a target in one pass.
- Time: `O(n)` average.
- Space: `O(n)`.
- Returns the two indices, or `None` if no pair exists.

```python
def two_sum(nums, target):  # Define a function that receives a list and the target sum.
    seen = {}  # Map each value already visited to its index.
    for i, num in enumerate(nums):  # Visit every number together with its index.
        need = target - num  # Calculate the value needed to reach the target.
        if need in seen:  # Check whether the needed value appeared earlier.
            return seen[need], i  # Return the earlier index and the current index.
        seen[num] = i  # Remember this number and its index for later numbers.
    return None  # Report that no valid pair was found.

print(two_sum([2, 7, 11, 15], 9))  # Demonstrate the pair at indices 0 and 1.
```

## 2. Contains Duplicate (Set)

**Pattern:** Set membership. A set stores unique values and offers average `O(1)` membership checks.
- Time: `O(n)` average.
- Space: `O(n)`.

```python
def contains_duplicate(nums):  # Define a function that checks a list for repeated values.
    seen = set()  # Create an empty set of values already encountered.
    for num in nums:  # Inspect each value in the list.
        if num in seen:  # Check whether this value has appeared before.
            return True  # A repeated value means the list contains a duplicate.
        seen.add(num)  # Record the value for future checks.
    return False  # No value was repeated.

print(contains_duplicate([1, 2, 3, 1]))  # Print True because 1 appears twice.
```

## 3. Frequency Counter

**Pattern:** Count occurrences with a dictionary.
- Time: `O(n)` average.
- Space: `O(k)`, where `k` is the number of distinct values.

```python
def frequency_count(items):  # Define a function to count item occurrences.
    counts = {}  # Create a dictionary for item-to-count mappings.
    for item in items:  # Process each item once.
        counts[item] = counts.get(item, 0) + 1  # Read its count (or zero), then add one.
    return counts  # Return all computed frequencies.

print(frequency_count(["a", "b", "a", "c", "b", "a"]))  # Show each letter's count.
```

## 4. Anagram Check

**Pattern:** Frequency counting. Two strings are anagrams when their character counts match.
- Time: `O(n + m)`.
- Space: `O(k)` for distinct characters.

```python
from collections import Counter  # Import a standard-library tool for counting values.

def are_anagrams(first, second):  # Define a function for comparing two strings.
    return Counter(first) == Counter(second)  # Compare the character-count mappings.

print(are_anagrams("listen", "silent"))  # Print True because both have the same letters.
```

## 5. Valid Parentheses (Stack)

**Pattern:** Stack. The most recently opened bracket must be closed first.
- Time: `O(n)`.
- Space: `O(n)`.

```python
def valid_parentheses(text):  # Define a function that validates bracket order.
    pairs = {")": "(", "]": "[", "}": "{"}  # Map each closing bracket to its opener.
    stack = []  # Store opening brackets that have not been closed.
    for char in text:  # Read the string from left to right.
        if char in "([{":  # Check whether the current character opens a bracket.
            stack.append(char)  # Push the opening bracket onto the stack.
        elif char in pairs:  # Check whether the character is a closing bracket.
            if not stack or stack.pop() != pairs[char]:  # Reject missing or mismatched openers.
                return False  # The bracket sequence is invalid.
    return not stack  # Valid only if no opening brackets remain.

print(valid_parentheses("([]){}"))  # Print True for correctly nested brackets.
```

## 6. Reverse a Linked List (Iterative)

**Pattern:** Pointer reversal. Change each node's `next` reference to point backward.
- Time: `O(n)`.
- Space: `O(1)` auxiliary space.
This example defines a tiny node class so it can run independently.

```python
class ListNode:  # Define one linked-list node.
    def __init__(self, value, next_node=None):  # Initialize the node and optional next pointer.
        self.value = value  # Store the node's value.
        self.next = next_node  # Link to the next node, or None.

def reverse_list(head):  # Receive the first node of a linked list.
    previous = None  # The reversed part starts empty.
    current = head  # Begin at the original first node.
    while current is not None:  # Continue until every node has been processed.
        following = current.next  # Save the next node before changing the link.
        current.next = previous  # Reverse the current node's pointer.
        previous = current  # Move the previous pointer forward.
        current = following  # Continue with the saved next node.
    return previous  # The previous node is now the new head.

head = ListNode(1, ListNode(2, ListNode(3)))  # Build the list 1 -> 2 -> 3.
head = reverse_list(head)  # Reverse the list and save its new head.
while head:  # Visit the reversed list.
    print(head.value, end=" -> ")  # Print the current node's value.
    head = head.next  # Advance to the next node.
print("None")  # Mark the end of the linked list.
```

## 7. Binary Search

**Pattern:** Divide and conquer on a sorted list.
- Requirement: input is sorted.
- Time: `O(log n)`.
- Space: `O(1)`.

```python
def binary_search(nums, target):  # Search for target in an ascending sorted list.
    left, right = 0, len(nums) - 1  # Set inclusive boundaries for the search range.
    while left <= right:  # Continue while the range contains at least one element.
        mid = (left + right) // 2  # Find the middle index without using floating point.
        if nums[mid] == target:  # Check whether the middle item is the target.
            return mid  # Return the target's index when it is found.
        if nums[mid] < target:  # If the middle item is too small,
            left = mid + 1  # discard the left half, including the middle item.
        else:  # Otherwise the middle item is too large.
            right = mid - 1  # discard the right half, including the middle item.
    return -1  # Return -1 when the target is absent.

print(binary_search([1, 3, 5, 7, 9], 7))  # Print index 3.
```

## 8. Lower Bound (First Value >= Target)

**Pattern:** Binary search on a boundary. Return the first index whose value is at least the target.
- Time: `O(log n)`.
- Returns `len(nums)` if every value is smaller.

```python
def lower_bound(nums, target):  # Find the first position where target could be inserted.
    left, right = 0, len(nums)  # Use a half-open range [left, right).
    while left < right:  # Stop when the candidate range is empty.
        mid = (left + right) // 2  # Select the middle candidate index.
        if nums[mid] < target:  # If this value is too small,
            left = mid + 1  # the answer must be after this index.
        else:  # This value may be the answer.
            right = mid  # Keep the middle index and search the left side.
    return left  # Return the first valid index, or len(nums).

print(lower_bound([1, 3, 3, 5, 8], 3))  # Print 1, the first index containing 3.
```

## 9. Merge Two Sorted Lists

**Pattern:** Two pointers. Compare the front items and append the smaller one.
- Time: `O(n + m)`.
- Space: `O(n + m)` for the output.

```python
def merge_sorted(a, b):  # Merge two lists that are already sorted.
    i = j = 0  # Start one pointer at the beginning of each list.
    result = []  # Store the merged output.
    while i < len(a) and j < len(b):  # Continue while both lists have items left.
        if a[i] <= b[j]:  # Compare the current items.
            result.append(a[i])  # Append the smaller item from list a.
            i += 1  # Advance pointer i.
        else:  # List b has the smaller current item.
            result.append(b[j])  # Append that item.
            j += 1  # Advance pointer j.
    result.extend(a[i:])  # Append any unprocessed items from a.
    result.extend(b[j:])  # Append any unprocessed items from b.
    return result  # Return the sorted merged list.

print(merge_sorted([1, 4, 7], [2, 3, 8]))  # Print [1, 2, 3, 4, 7, 8].
```

## 10. Merge Sort

**Pattern:** Divide and conquer. Split, sort each half, then merge.
- Time: `O(n log n)`.
- Space: `O(n)` for this implementation.

```python
def merge_sort(nums):  # Define a sorting function using divide and conquer.
    if len(nums) <= 1:  # A list with zero or one item is already sorted.
        return nums.copy()  # Return a copy so the input is not changed.
    mid = len(nums) // 2  # Find the midpoint.
    left = merge_sort(nums[:mid])  # Recursively sort the left half.
    right = merge_sort(nums[mid:])  # Recursively sort the right half.
    return merge_sorted(left, right)  # Merge the two sorted halves.

print(merge_sort([8, 3, 5, 1, 9, 2]))  # Print the numbers in ascending order.
```

## 11. Quick Sort

**Pattern:** Partition around a pivot. This clear version creates new lists.
- Average time: `O(n log n)`.
- Worst-case time: `O(n²)`.
- Additional memory is used by the partitions.

```python
def quick_sort(nums):  # Sort a list recursively using a pivot.
    if len(nums) <= 1:  # Lists of length zero or one need no sorting.
        return nums.copy()  # Return a copy of the base case.
    pivot = nums[len(nums) // 2]  # Choose the middle value as the pivot.
    lower = [x for x in nums if x < pivot]  # Collect values smaller than the pivot.
    equal = [x for x in nums if x == pivot]  # Collect values equal to the pivot.
    upper = [x for x in nums if x > pivot]  # Collect values larger than the pivot.
    return quick_sort(lower) + equal + quick_sort(upper)  # Sort partitions and join them.

print(quick_sort([4, 1, 7, 4, 2, 9]))  # Print [1, 2, 4, 4, 7, 9].
```

## 12. Insertion Sort

**Pattern:** Grow a sorted prefix by inserting each next item into its correct place.
- Worst-case time: `O(n²)`.
- Best-case time: `O(n)` for already sorted input.
- Auxiliary space: `O(1)` for in-place sorting; this example sorts a copy.

```python
def insertion_sort(nums):  # Sort a copy of the input list.
    result = nums.copy()  # Keep the original list unchanged.
    for i in range(1, len(result)):  # Start at the second item.
        current = result[i]  # Save the value that needs to be inserted.
        j = i - 1  # Start comparing with the last item in the sorted prefix.
        while j >= 0 and result[j] > current:  # Shift larger values one position right.
            result[j + 1] = result[j]  # Move the larger value to make room.
            j -= 1  # Compare with the previous item.
        result[j + 1] = current  # Insert current into the open position.
    return result  # Return the sorted copy.

print(insertion_sort([5, 2, 4, 1]))  # Print [1, 2, 4, 5].
```

## 13. Bubble Sort

**Pattern:** Repeatedly swap adjacent out-of-order values.
- Average/worst time: `O(n²)`.
- Auxiliary space: `O(1)` if sorting in place.
Mostly useful for learning rather than production code.

```python
def bubble_sort(nums):  # Sort a copy using adjacent swaps.
    result = nums.copy()  # Copy the input to avoid modifying it.
    for end in range(len(result) - 1, 0, -1):  # Shrink the unsorted suffix after each pass.
        swapped = False  # Track whether this pass changed the list.
        for i in range(end):  # Compare adjacent pairs up to the current boundary.
            if result[i] > result[i + 1]:  # Check whether the pair is out of order.
                result[i], result[i + 1] = result[i + 1], result[i]  # Swap the pair.
                swapped = True  # Record that a swap happened.
        if not swapped:  # If no swaps happened, the list is already sorted.
            break  # Stop early instead of doing unnecessary passes.
    return result  # Return the sorted copy.

print(bubble_sort([4, 2, 5, 1]))  # Print [1, 2, 4, 5].
```

## 14. Prefix Sums

**Pattern:** Precompute cumulative totals to answer range-sum queries quickly.
- Build time: `O(n)`.
- Each range query: `O(1)`.
- Space: `O(n)`.

```python
def build_prefix(nums):  # Build cumulative sums with a leading zero.
    prefix = [0]  # prefix[i] will represent the sum before index i.
    for num in nums:  # Process each value in order.
        prefix.append(prefix[-1] + num)  # Add this value to the previous cumulative sum.
    return prefix  # Return the prefix-sum array.

def range_sum(prefix, left, right):  # Sum inclusive indices left through right.
    return prefix[right + 1] - prefix[left]  # Subtract the sum before left from the ending total.

values = [2, 4, 6, 8, 10]  # Create sample data.
prefix = build_prefix(values)  # Precompute the cumulative sums once.
print(range_sum(prefix, 1, 3))  # Print 4 + 6 + 8 = 18.
```

## 15. Sliding Window Maximum Sum

**Pattern:** Maintain a fixed-size contiguous window.
- Time: `O(n)`.
- Extra space: `O(1)`.
Useful for contiguous subarray problems.

```python
def max_window_sum(nums, k):  # Find the maximum sum of k consecutive values.
    if k < 1 or k > len(nums):  # Validate the requested window size.
        raise ValueError("k must be between 1 and len(nums)")  # Explain invalid input.
    window = sum(nums[:k])  # Calculate the first window's sum.
    best = window  # Treat the first window as the best so far.
    for right in range(k, len(nums)):  # Move the window's right edge forward.
        window += nums[right] - nums[right - k]  # Add entering value and remove leaving value.
        best = max(best, window)  # Save the largest sum seen so far.
    return best  # Return the maximum window sum.

print(max_window_sum([2, 1, 5, 1, 3, 2], 3))  # Print 9.
```

## 16. Longest Substring Without Repeating Characters

**Pattern:** Variable-size sliding window with a last-seen index map.
- Time: `O(n)` average.
- Space: `O(k)` for distinct characters in the window/map.

```python
def longest_unique_substring(text):  # Find the length of the longest unique-character substring.
    last_seen = {}  # Map each character to its most recent index.
    left = 0  # Mark the left edge of the current window.
    best = 0  # Store the maximum length found so far.
    for right, char in enumerate(text):  # Expand the window one character at a time.
        if char in last_seen and last_seen[char] >= left:  # Check for a repeat inside this window.
            left = last_seen[char] + 1  # Move left past the previous occurrence.
        last_seen[char] = right  # Update the character's latest index.
        best = max(best, right - left + 1)  # Update the best window length.
    return best  # Return the longest length.

print(longest_unique_substring("abcabcbb"))  # Print 3 for "abc".
```

## 17. Maximum Subarray (Kadane's Algorithm)

**Pattern:** Dynamic programming with a running best subarray ending at each position.
- Time: `O(n)`.
- Space: `O(1)`.
Requires a non-empty list in this version.

```python
def max_subarray_sum(nums):  # Find the largest sum of a non-empty contiguous subarray.
    if not nums:  # Check that at least one value exists.
        raise ValueError("nums must not be empty")  # Reject an undefined empty-input case.
    current = best = nums[0]  # Initialize both values from the first element.
    for num in nums[1:]:  # Process each remaining value.
        current = max(num, current + num)  # Start fresh or extend the current subarray.
        best = max(best, current)  # Save the best sum found anywhere.
    return best  # Return the maximum subarray sum.

print(max_subarray_sum([-2, 1, -3, 4, -1, 2, 1, -5, 4]))  # Print 6.
```

## 18. Best Time to Buy and Sell Stock

**Pattern:** One-pass minimum tracking.
- Time: `O(n)`.
- Space: `O(1)`.
Find the maximum profit from one buy followed by one sell.

```python
def max_stock_profit(prices):  # Find the best profit from one buy and one later sell.
    min_price = float("inf")  # Start with a value larger than any normal price.
    best_profit = 0  # Zero profit is possible by making no trade.
    for price in prices:  # Visit prices in chronological order.
        min_price = min(min_price, price)  # Remember the cheapest price seen so far.
        best_profit = max(best_profit, price - min_price)  # Evaluate selling at today's price.
    return best_profit  # Return the best possible profit.

print(max_stock_profit([7, 1, 5, 3, 6, 4]))  # Print 5 (buy at 1, sell at 6).
```

## 19. Product of Array Except Self

**Pattern:** Prefix and suffix products without division.
- Time: `O(n)`.
- Output space: `O(n)`.
Works when the input contains zero values.

```python
def product_except_self(nums):  # Return each position's product excluding itself.
    result = [1] * len(nums)  # Initialize each output to the multiplicative identity.
    prefix = 1  # Store the product of values to the left.
    for i, num in enumerate(nums):  # Move from left to right.
        result[i] = prefix  # Save the product of all earlier values.
        prefix *= num  # Include the current value for the next position.
    suffix = 1  # Store the product of values to the right.
    for i in range(len(nums) - 1, -1, -1):  # Move from right to left.
        result[i] *= suffix  # Combine the left product with the right product.
        suffix *= nums[i]  # Include the current value for the next position.
    return result  # Return products excluding each element.

print(product_except_self([1, 2, 3, 4]))  # Print [24, 12, 8, 6].
```

## 20. Rotate Array

**Pattern:** Reverse segments. Rotate right by `k` positions in `O(n)` time.
- Time: `O(n)`.
- Space: `O(n)` because this version returns a new list.

```python
def rotate_right(nums, k):  # Rotate the list right by k positions.
    if not nums:  # Handle an empty list separately.
        return []  # There is nothing to rotate.
    k %= len(nums)  # Reduce k so it fits within the list length.
    return nums[-k:] + nums[:-k] if k else nums.copy()  # Join the moved suffix and prefix.

print(rotate_right([1, 2, 3, 4, 5], 2))  # Print [4, 5, 1, 2, 3].
```

## 21. Move Zeroes

**Pattern:** In-place stable compaction.
- Time: `O(n)`.
- Extra space: `O(1)`.
Preserves the relative order of non-zero elements.

```python
def move_zeroes(nums):  # Move zero values to the end of the list in place.
    write = 0  # Point to the next position for a non-zero value.
    for value in nums:  # Read each original value from left to right.
        if value != 0:  # Keep only non-zero values in the first pass.
            nums[write] = value  # Write this value into the compacted prefix.
            write += 1  # Advance the next write position.
    while write < len(nums):  # Fill every remaining position.
        nums[write] = 0  # Put a zero at the current position.
        write += 1  # Advance to the next position.
    return nums  # Return the same, now-modified list.

print(move_zeroes([0, 1, 0, 3, 12]))  # Print [1, 3, 12, 0, 0].
```

## 22. Three Sum

**Pattern:** Sort + fixed first element + two pointers.
- Time: `O(n²)`.
- Space: depends on sorting and output size.
Returns unique triplets that sum to zero.

```python
def three_sum(nums):  # Find unique triplets whose sum is zero.
    nums = sorted(nums)  # Sort values so two pointers can be used.
    result = []  # Store the unique triplets.
    for i in range(len(nums) - 2):  # Choose the first value of each triplet.
        if i > 0 and nums[i] == nums[i - 1]:  # Skip duplicate first values.
            continue  # Avoid producing duplicate triplets.
        left, right = i + 1, len(nums) - 1  # Search the remaining range with two pointers.
        while left < right:  # Continue while two distinct positions remain.
            total = nums[i] + nums[left] + nums[right]  # Calculate the current triplet sum.
            if total == 0:  # Check whether this triplet matches the target.
                result.append([nums[i], nums[left], nums[right]])  # Store the triplet.
                left += 1  # Move the left pointer to search for another pair.
                right -= 1  # Move the right pointer inward as well.
                while left < right and nums[left] == nums[left - 1]:  # Skip repeated left values.
                    left += 1  # Advance past this duplicate.
            elif total < 0:  # A sum that is too small needs a larger value.
                left += 1  # Move left toward larger values.
            else:  # A sum that is too large needs a smaller value.
                right -= 1  # Move right toward smaller values.
    return result  # Return all unique zero-sum triplets.

print(three_sum([-1, 0, 1, 2, -1, -4]))  # Print [[-1, -1, 2], [-1, 0, 1]].
```

## 23. Binary Tree Inorder Traversal

**Pattern:** Tree DFS. Inorder visits left subtree, node, then right subtree.
- Time: `O(n)`.
- Space: `O(h)` recursion stack, where `h` is tree height.

```python
class TreeNode:  # Define a node for a binary tree.
    def __init__(self, value, left=None, right=None):  # Accept a value and optional children.
        self.value = value  # Store this node's value.
        self.left = left  # Link to the left child.
        self.right = right  # Link to the right child.

def inorder(root):  # Return values in left-node-right order.
    if root is None:  # An empty subtree contains no values.
        return []  # Return an empty result for this subtree.
    return inorder(root.left) + [root.value] + inorder(root.right)  # Combine left, current, and right.

tree = TreeNode(2, TreeNode(1), TreeNode(3))  # Build a small tree with values 1, 2, and 3.
print(inorder(tree))  # Print [1, 2, 3].
```

## 24. Binary Tree Level Order Traversal

**Pattern:** BFS with a queue.
- Time: `O(n)`.
- Space: `O(n)` for the queue and output.
Visits each tree level from left to right.

```python
from collections import deque  # Import a queue with efficient front removal.

def level_order(root):  # Traverse a binary tree one level at a time.
    if root is None:  # Check for an empty tree.
        return []  # An empty tree has no levels.
    queue = deque([root])  # Start the queue with the root node.
    levels = []  # Store one list of values per level.
    while queue:  # Continue until no nodes remain.
        level_values = []  # Start collecting the current level.
        for _ in range(len(queue)):  # Process only the nodes currently in the queue.
            node = queue.popleft()  # Remove the oldest node.
            level_values.append(node.value)  # Record its value.
            if node.left:  # Check whether a left child exists.
                queue.append(node.left)  # Schedule the left child for the next level.
            if node.right:  # Check whether a right child exists.
                queue.append(node.right)  # Schedule the right child for the next level.
        levels.append(level_values)  # Save the completed level.
    return levels  # Return all levels in order.

tree = TreeNode(1, TreeNode(2), TreeNode(3))  # Build a tree with two children.
print(level_order(tree))  # Print [[1], [2, 3]].
```

## 25. Maximum Depth of a Binary Tree

**Pattern:** Recursive DFS.
- Time: `O(n)`.
- Space: `O(h)` recursion stack.
The depth of an empty tree is zero.

```python
def max_depth(root):  # Compute the longest root-to-leaf path in number of nodes.
    if root is None:  # An empty subtree has depth zero.
        return 0  # Provide the recursion's base case.
    left_depth = max_depth(root.left)  # Calculate the left subtree's depth.
    right_depth = max_depth(root.right)  # Calculate the right subtree's depth.
    return 1 + max(left_depth, right_depth)  # Count this node plus the deeper subtree.

tree = TreeNode(1, TreeNode(2, TreeNode(4)), TreeNode(3))  # Build a tree of depth 3.
print(max_depth(tree))  # Print 3.
```

## 26. Graph BFS

**Pattern:** Queue-based graph traversal.
- Time: `O(V + E)` using an adjacency list.
- Space: `O(V)`.
Useful for shortest paths in unweighted graphs.

```python
from collections import deque  # Import an efficient queue.

def graph_bfs(graph, start):  # Traverse a graph from a starting node.
    visited = {start}  # Mark the start node immediately to prevent re-adding it.
    queue = deque([start])  # Schedule the start node for processing.
    order = []  # Store the visit order.
    while queue:  # Continue while nodes are waiting.
        node = queue.popleft()  # Process the earliest scheduled node.
        order.append(node)  # Add the node to the result.
        for neighbor in graph.get(node, []):  # Inspect the node's adjacent neighbors.
            if neighbor not in visited:  # Only schedule unvisited nodes.
                visited.add(neighbor)  # Mark the neighbor before adding it to the queue.
                queue.append(neighbor)  # Schedule the neighbor for later processing.
    return order  # Return the traversal order.

graph = {"A": ["B", "C"], "B": ["D"], "C": [], "D": []}  # Define an adjacency list.
print(graph_bfs(graph, "A"))  # Print ['A', 'B', 'C', 'D'].
```

## 27. Graph DFS

**Pattern:** Stack-based graph traversal.
- Time: `O(V + E)`.
- Space: `O(V)`.
An iterative stack avoids recursive-call depth limits.

```python
def graph_dfs(graph, start):  # Traverse a graph depth-first.
    visited = set()  # Track nodes that have already been processed.
    stack = [start]  # Put the starting node on the stack.
    order = []  # Store the traversal order.
    while stack:  # Continue while work remains.
        node = stack.pop()  # Take the most recently added node.
        if node in visited:  # Ignore nodes already processed.
            continue  # Move to the next stack item.
        visited.add(node)  # Mark this node as visited.
        order.append(node)  # Record the visit.
        for neighbor in reversed(graph.get(node, [])):  # Push neighbors in reverse order.
            if neighbor not in visited:  # Avoid adding already visited neighbors.
                stack.append(neighbor)  # Explore this neighbor in a later iteration.
    return order  # Return the DFS order.

graph = {"A": ["B", "C"], "B": ["D"], "C": [], "D": []}  # Define a sample graph.
print(graph_dfs(graph, "A"))  # Print one valid DFS order.
```

## 28. Detect Cycle in a Directed Graph

**Pattern:** DFS with three states: unseen, active, finished.
- Time: `O(V + E)`.
- Space: `O(V)`.
An edge to an active node indicates a directed cycle.

```python
def has_directed_cycle(graph):  # Determine whether a directed graph contains a cycle.
    state = {}  # Store 0/unseen, 1/active, or 2/finished for each node.
    def visit(node):  # Define recursive DFS for one node.
        state[node] = 1  # Mark this node active on the current recursion path.
        for neighbor in graph.get(node, []):  # Examine each outgoing edge.
            if state.get(neighbor, 0) == 1:  # An edge to an active node closes a cycle.
                return True  # Report that a cycle exists.
            if state.get(neighbor, 0) == 0 and visit(neighbor):  # Explore an unseen neighbor.
                return True  # Propagate a cycle found deeper in the graph.
        state[node] = 2  # Mark the node finished after exploring all neighbors.
        return False  # No cycle was found from this node.
    for node in graph:  # Start DFS from each disconnected component.
        if state.get(node, 0) == 0 and visit(node):  # Search any node not visited yet.
            return True  # Return as soon as a cycle is found.
    return False  # Every component was explored without finding a cycle.

print(has_directed_cycle({"A": ["B"], "B": ["C"], "C": ["A"]}))  # Print True.
```

## 29. Topological Sort (Kahn's Algorithm)

**Pattern:** Indegree counting + queue.
- Time: `O(V + E)`.
- Space: `O(V)`.
Works for directed acyclic graphs (DAGs); fewer output nodes means a cycle exists.

```python
from collections import deque  # Import a queue for zero-indegree nodes.

def topological_sort(graph):  # Return a topological ordering or None if a cycle exists.
    nodes = set(graph)  # Start with all nodes explicitly listed as keys.
    for neighbors in graph.values():  # Inspect every adjacency list.
        nodes.update(neighbors)  # Include destination-only nodes as well.
    indegree = {node: 0 for node in nodes}  # Initialize every node's incoming-edge count.
    for node in graph:  # Visit every source node.
        for neighbor in graph[node]:  # Inspect each directed edge.
            indegree[neighbor] += 1  # Count this incoming edge.
    queue = deque(node for node in nodes if indegree[node] == 0)  # Start with nodes having no prerequisites.
    order = []  # Store the topological ordering.
    while queue:  # Process nodes whose prerequisites are complete.
        node = queue.popleft()  # Remove a ready node.
        order.append(node)  # Place it into the ordering.
        for neighbor in graph.get(node, []):  # Remove the effect of its outgoing edges.
            indegree[neighbor] -= 1  # One prerequisite has now been processed.
            if indegree[neighbor] == 0:  # Check whether the neighbor is now ready.
                queue.append(neighbor)  # Schedule the newly ready node.
    return order if len(order) == len(nodes) else None  # Reject a cycle if nodes remain.

print(topological_sort({"cook": ["eat"], "shop": ["cook"], "eat": []}))  # Print a valid ordering.
```

## 30. Dijkstra's Shortest Path

**Pattern:** Greedy shortest path with a min-heap.
- Time: roughly `O((V + E) log V)` with an adjacency list and heap.
- Requirement: edge weights must be non-negative.
Graph format: `node: [(neighbor, weight), ...]`.

```python
import heapq  # Import Python's min-heap implementation.

def dijkstra(graph, start):  # Compute shortest distances from start.
    distances = {start: 0}  # The distance from start to itself is zero.
    heap = [(0, start)]  # Store (distance, node) candidates, smallest first.
    while heap:  # Continue while candidate paths remain.
        distance, node = heapq.heappop(heap)  # Get the currently shortest candidate.
        if distance != distances.get(node, float("inf")):  # Ignore outdated heap entries.
            continue  # A better path to this node was already found.
        for neighbor, weight in graph.get(node, []):  # Inspect every outgoing weighted edge.
            new_distance = distance + weight  # Calculate the path through this node.
            if new_distance < distances.get(neighbor, float("inf")):  # Check for improvement.
                distances[neighbor] = new_distance  # Save the shorter known distance.
                heapq.heappush(heap, (new_distance, neighbor))  # Schedule the improved path.
    return distances  # Return distances for reachable nodes.

graph = {"A": [("B", 4), ("C", 1)], "C": [("B", 2)], "B": []}  # Define non-negative weighted edges.
print(dijkstra(graph, "A"))  # Print distances A:0, C:1, B:3.
```

## 31. Union-Find / Disjoint Set

**Pattern:** Track connected groups with path compression and union by size.
- Near-constant amortized operation time: `O(α(n))`.
Useful in connectivity problems and Kruskal's minimum spanning tree algorithm.

```python
class UnionFind:  # Define a data structure for disjoint sets.
    def __init__(self, n):  # Initialize n separate groups numbered 0 through n-1.
        self.parent = list(range(n))  # Each element initially points to itself.
        self.size = [1] * n  # Every initial group has size one.
    def find(self, x):  # Find the representative of x's group.
        if self.parent[x] != x:  # Check whether x is not a root.
            self.parent[x] = self.find(self.parent[x])  # Compress the path recursively.
        return self.parent[x]  # Return the root representative.
    def union(self, a, b):  # Merge the groups containing a and b.
        root_a = self.find(a)  # Find a's representative.
        root_b = self.find(b)  # Find b's representative.
        if root_a == root_b:  # Check whether both are already connected.
            return False  # No merge was needed.
        if self.size[root_a] < self.size[root_b]:  # Keep the larger group as the root.
            root_a, root_b = root_b, root_a  # Swap the roots if necessary.
        self.parent[root_b] = root_a  # Attach the smaller group to the larger one.
        self.size[root_a] += self.size[root_b]  # Update the merged group's size.
        return True  # Report that a merge occurred.

uf = UnionFind(5)  # Create five separate groups.
uf.union(0, 1)  # Connect elements 0 and 1.
uf.union(1, 2)  # Connect element 2 to their group.
print(uf.find(0) == uf.find(2))  # Print True because 0 and 2 are connected.
```

## 32. Backtracking: Generate Subsets

**Pattern:** Choose or skip each element, then undo the choice when returning.
- Time: `O(2^n)` subsets, ignoring the cost of copying each subset.
- Output space: `O(n · 2^n)` to store all subsets.

```python
def subsets(nums):  # Generate every subset of nums.
    result = []  # Store each completed subset.
    path = []  # Hold the subset currently being built.
    def backtrack(index):  # Decide what to do with nums[index].
        if index == len(nums):  # Check whether every item has been considered.
            result.append(path.copy())  # Save a copy of the current subset.
            return  # Finish this branch of the search.
        backtrack(index + 1)  # First branch: skip the current item.
        path.append(nums[index])  # Choose the current item for the second branch.
        backtrack(index + 1)  # Explore subsets that include this item.
        path.pop()  # Undo the choice before returning to the caller.
    backtrack(0)  # Start with the first index and an empty subset.
    return result  # Return all generated subsets.

print(subsets([1, 2]))  # Print [], [2], [1], and [1, 2] in traversal order.
```

## 33. Backtracking: Permutations

**Pattern:** Build an ordering one unused element at a time.
- Time: `O(n · n!)` including output construction.
- Space: `O(n)` auxiliary recursion/path space, excluding output.

```python
def permutations(nums):  # Generate every ordering of the input values.
    result = []  # Store completed permutations.
    path = []  # Store the current partial permutation.
    used = [False] * len(nums)  # Track which input positions are already selected.
    def backtrack():  # Extend the current partial permutation.
        if len(path) == len(nums):  # Check whether all positions have been selected.
            result.append(path.copy())  # Save a copy of this complete ordering.
            return  # Stop this branch.
        for i, value in enumerate(nums):  # Try each input position.
            if used[i]:  # Skip a position already in the current path.
                continue  # Try the next value instead.
            used[i] = True  # Mark this position as selected.
            path.append(value)  # Add the selected value to the path.
            backtrack()  # Recursively choose the next position.
            path.pop()  # Remove the last value to restore the previous state.
            used[i] = False  # Mark the position available for another branch.
    backtrack()  # Start with an empty path.
    return result  # Return every permutation.

print(permutations([1, 2, 3]))  # Print all six possible orderings.
```

## 34. N-Queens (Backtracking)

**Pattern:** Backtracking with sets for occupied columns and diagonals.
- Exponential search in general.
- This example returns the number of solutions, not the boards.

```python
def count_n_queens(n):  # Count placements of n queens on an n-by-n board.
    columns = set()  # Track columns already occupied by queens.
    diagonal_down = set()  # Track row-column differences for one diagonal direction.
    diagonal_up = set()  # Track row+column sums for the other diagonal direction.
    def place(row):  # Place a queen in the given row.
        if row == n:  # All rows have a queen when row reaches n.
            return 1  # Count one complete solution.
        total = 0  # Count solutions reachable from this row.
        for col in range(n):  # Try each column in this row.
            down = row - col  # Identify one diagonal using row minus column.
            up = row + col  # Identify the other diagonal using row plus column.
            if col in columns or down in diagonal_down or up in diagonal_up:  # Reject attacks.
                continue  # Try another column.
            columns.add(col)  # Mark this column occupied.
            diagonal_down.add(down)  # Mark the first diagonal occupied.
            diagonal_up.add(up)  # Mark the second diagonal occupied.
            total += place(row + 1)  # Count solutions after placing this queen.
            columns.remove(col)  # Undo the column choice.
            diagonal_down.remove(down)  # Undo the first diagonal choice.
            diagonal_up.remove(up)  # Undo the second diagonal choice.
        return total  # Return the number of valid placements from this row.
    return place(0)  # Start placing queens in row zero.

print(count_n_queens(4))  # Print 2 solutions for a 4-by-4 board.
```

## 35. Fibonacci with Memoization

**Pattern:** Top-down dynamic programming. Cache results so each subproblem is solved once.
- Time: `O(n)`.
- Space: `O(n)` for the cache and recursion stack.

```python
from functools import lru_cache  # Import a decorator that caches function results.

@lru_cache(maxsize=None)  # Keep every computed result in an unbounded cache.
def fibonacci_memo(n):  # Define Fibonacci recursively with memoization.
    if n < 0:  # Reject negative indices.
        raise ValueError("n must be non-negative")  # Explain the valid input range.
    if n < 2:  # The first two Fibonacci values are their own indices.
        return n  # Return 0 for n=0 and 1 for n=1.
    return fibonacci_memo(n - 1) + fibonacci_memo(n - 2)  # Reuse cached smaller results.

print(fibonacci_memo(40))  # Compute the 40th Fibonacci number efficiently.
```

## 36. Coin Change (Minimum Coins)

**Pattern:** Bottom-up dynamic programming.
- Time: `O(amount × number_of_coins)`.
- Space: `O(amount)`.
Returns `-1` if the amount cannot be formed. Coin values must be positive.

```python
def min_coins(coins, amount):  # Find the fewest coins needed to make amount.
    if amount < 0 or any(coin <= 0 for coin in coins):  # Validate the problem inputs.
        raise ValueError("amount must be non-negative and coins positive")  # Explain invalid input.
    unreachable = amount + 1  # Use a number larger than any possible optimal coin count.
    dp = [0] + [unreachable] * amount  # dp[x] stores the fewest coins needed for value x.
    for value in range(1, amount + 1):  # Solve every amount from 1 upward.
        for coin in coins:  # Try each available coin denomination.
            if coin <= value:  # A coin can be used only if it fits this amount.
                dp[value] = min(dp[value], dp[value - coin] + 1)  # Improve the answer using this coin.
    return dp[amount] if dp[amount] != unreachable else -1  # Convert unreachable to -1.

print(min_coins([1, 3, 4], 6))  # Print 2 because 3 + 3 uses two coins.
```

## 37. 0/1 Knapsack

**Pattern:** Dynamic programming over capacity. Each item can be taken at most once.
- Time: `O(n × capacity)`.
- Space: `O(capacity)` with a one-dimensional table.

```python
def knapsack(weights, values, capacity):  # Maximize value without exceeding capacity.
    dp = [0] * (capacity + 1)  # dp[c] stores the best value for capacity c.
    for weight, value in zip(weights, values):  # Process each item exactly once.
        for cap in range(capacity, weight - 1, -1):  # Iterate backward to avoid reusing this item.
            dp[cap] = max(dp[cap], dp[cap - weight] + value)  # Compare skipping and taking the item.
    return dp[capacity]  # Return the best value at full capacity.

print(knapsack([2, 3, 4], [4, 5, 7], 5))  # Print 9 by taking the first two items.
```

## 38. Longest Common Subsequence

**Pattern:** Two-dimensional dynamic programming.
- Time: `O(n × m)`.
- Space: `O(n × m)`.
A subsequence preserves order but does not need to be contiguous.

```python
def lcs_length(a, b):  # Compute the longest common subsequence length.
    rows, cols = len(a), len(b)  # Store the lengths of the two strings.
    dp = [[0] * (cols + 1) for _ in range(rows + 1)]  # Initialize empty-prefix answers to zero.
    for i in range(1, rows + 1):  # Consider prefixes of string a.
        for j in range(1, cols + 1):  # Consider prefixes of string b.
            if a[i - 1] == b[j - 1]:  # If the current characters match,
                dp[i][j] = dp[i - 1][j - 1] + 1  # extend the previous common subsequence.
            else:  # If the characters differ,
                dp[i][j] = max(dp[i - 1][j], dp[i][j - 1])  # Keep the better previous answer.
    return dp[rows][cols]  # Return the answer for both full strings.

print(lcs_length("abcde", "ace"))  # Print 3 for the subsequence "ace".
```

## 39. Edit Distance

**Pattern:** Dynamic programming for insertions, deletions, and replacements.
- Time: `O(n × m)`.
- Space: `O(n × m)`.
Returns the minimum number of edits needed to transform one string into another.

```python
def edit_distance(a, b):  # Find the minimum edits to transform a into b.
    n, m = len(a), len(b)  # Store the lengths of both strings.
    dp = [[0] * (m + 1) for _ in range(n + 1)]  # Allocate the edit-distance table.
    for i in range(n + 1):  # Initialize comparisons against an empty second string.
        dp[i][0] = i  # Deleting i characters costs i edits.
    for j in range(m + 1):  # Initialize comparisons against an empty first string.
        dp[0][j] = j  # Inserting j characters costs j edits.
    for i in range(1, n + 1):  # Process each prefix of a.
        for j in range(1, m + 1):  # Process each prefix of b.
            cost = 0 if a[i - 1] == b[j - 1] else 1  # Matching characters cost zero; replacement costs one.
            dp[i][j] = min(dp[i - 1][j] + 1, dp[i][j - 1] + 1, dp[i - 1][j - 1] + cost)  # Choose delete, insert, or replace.
    return dp[n][m]  # Return the distance for the complete strings.

print(edit_distance("kitten", "sitting"))  # Print 3.
```

## 40. Greedy Activity Selection

**Pattern:** Sort by finish time and always choose the next compatible activity.
- Time: `O(n log n)` due to sorting.
- Greedy is optimal for this specific activity-selection problem.

```python
def select_activities(activities):  # Select the maximum number of non-overlapping intervals.
    activities = sorted(activities, key=lambda item: item[1])  # Sort intervals by finish time.
    selected = []  # Store activities chosen by the greedy rule.
    last_finish = float("-inf")  # Initially, no activity has finished.
    for start, finish in activities:  # Inspect activities in finish-time order.
        if start >= last_finish:  # Choose an activity if it starts after the last selected one ends.
            selected.append((start, finish))  # Record the compatible activity.
            last_finish = finish  # Update the end time of the latest selected activity.
    return selected  # Return the chosen non-overlapping intervals.

print(select_activities([(1, 3), (2, 4), (3, 5), (5, 7)]))  # Print a maximum-size compatible selection.
```

## 41. Interval Merge

**Pattern:** Sort intervals, then merge overlaps.
- Time: `O(n log n)`.
- Space: `O(n)` for the result.

```python
def merge_intervals(intervals):  # Merge overlapping closed intervals.
    if not intervals:  # Handle an empty collection.
        return []  # There are no intervals to merge.
    intervals = sorted(intervals, key=lambda item: item[0])  # Sort by each interval's start.
    merged = [list(intervals[0])]  # Start with a mutable copy of the first interval.
    for start, end in intervals[1:]:  # Process every remaining interval.
        if start <= merged[-1][1]:  # Check whether it overlaps the most recent merged interval.
            merged[-1][1] = max(merged[-1][1], end)  # Extend the merged interval if needed.
        else:  # This interval starts after the previous interval ends.
            merged.append([start, end])  # Start a new merged interval.
    return merged  # Return all merged intervals.

print(merge_intervals([(1, 3), (2, 6), (8, 10), (9, 12)]))  # Print [[1, 6], [8, 12]].
```

## 42. Heap: K Largest Values

**Pattern:** Keep a min-heap of size `k`.
- Time: `O(n log k)`.
- Space: `O(k)`.
Useful when `k` is much smaller than the total number of values.

```python
import heapq  # Import the min-heap functions.

def k_largest(nums, k):  # Return the k largest values, sorted descending.
    if k < 0:  # Reject a negative requested count.
        raise ValueError("k must be non-negative")  # Explain the valid range.
    if k == 0:  # No values are requested.
        return []  # Return an empty result.
    heap = []  # Keep a min-heap of the best candidates.
    for num in nums:  # Inspect each input value.
        heapq.heappush(heap, num)  # Add the value to the heap.
        if len(heap) > k:  # Keep at most k values in the heap.
            heapq.heappop(heap)  # Remove the smallest candidate.
    return sorted(heap, reverse=True)  # Return retained values from largest to smallest.

print(k_largest([4, 1, 9, 7, 3, 8], 3))  # Print [9, 8, 7].
```

## 43. Heap: Merge K Sorted Lists

**Pattern:** Min-heap of the current smallest remaining item from each list.
- Time: `O(N log k)` for N total items and k lists.
- Space: `O(k)` heap space, excluding output.

```python
import heapq  # Import priority-queue operations.

def merge_k_sorted(lists):  # Merge multiple already-sorted lists.
    heap = []  # Store (value, list_index, item_index) candidates.
    for list_index, values in enumerate(lists):  # Visit each input list.
        if values:  # Ignore empty lists.
            heapq.heappush(heap, (values[0], list_index, 0))  # Add its first item to the heap.
    result = []  # Store the globally sorted output.
    while heap:  # Continue while any list has an unprocessed item.
        value, list_index, item_index = heapq.heappop(heap)  # Extract the smallest current item.
        result.append(value)  # Add the extracted value to the result.
        next_index = item_index + 1  # Calculate the next position in that same list.
        if next_index < len(lists[list_index]):  # Check whether another item remains in that list.
            next_value = lists[list_index][next_index]  # Read the next value.
            heapq.heappush(heap, (next_value, list_index, next_index))  # Add it as a new candidate.
    return result  # Return the fully merged sequence.

print(merge_k_sorted([[1, 4, 7], [2, 5], [3, 6, 8]]))  # Print values 1 through 8.
```

## 44. Monotonic Stack: Next Greater Element

**Pattern:** Keep a stack of indices whose next greater value is not known yet.
- Time: `O(n)` because each index is pushed and popped at most once.
- Space: `O(n)`.

```python
def next_greater(nums):  # Find the next greater value to the right for each element.
    answer = [-1] * len(nums)  # Use -1 when no greater value exists.
    stack = []  # Store indices waiting for a greater value.
    for i, num in enumerate(nums):  # Process values from left to right.
        while stack and nums[stack[-1]] < num:  # Resolve previous values smaller than this one.
            index = stack.pop()  # Remove the index whose answer is now known.
            answer[index] = num  # Record this value as its next greater element.
        stack.append(i)  # Wait for a future greater value for this index.
    return answer  # Return the answers in original index order.

print(next_greater([2, 1, 2, 4, 3]))  # Print [4, 2, 4, -1, -1].
```

## 45. Monotonic Stack: Largest Histogram Rectangle

**Pattern:** Increasing stack of bar indices.
- Time: `O(n)`.
- Space: `O(n)`.
Each bar is evaluated when a shorter bar closes its possible rectangle.

```python
def largest_rectangle(heights):  # Find the largest rectangle under a histogram.
    stack = []  # Keep indices with increasing bar heights.
    best = 0  # Store the largest area found.
    extended = heights + [0]  # Add a zero-height bar to flush the stack at the end.
    for i, height in enumerate(extended):  # Process every bar including the sentinel.
        while stack and extended[stack[-1]] > height:  # Close bars taller than the current bar.
            h = extended[stack.pop()]  # Use the popped bar's height for a rectangle.
            left_smaller = stack[-1] if stack else -1  # Find the previous smaller bar boundary.
            width = i - left_smaller - 1  # Calculate how far this height can extend.
            best = max(best, h * width)  # Update the best rectangle area.
        stack.append(i)  # Push the current bar index onto the stack.
    return best  # Return the maximum area.

print(largest_rectangle([2, 1, 5, 6, 2, 3]))  # Print 10.
```

## 46. KMP String Search

**Pattern:** Prefix-function preprocessing avoids restarting comparisons from scratch.
- Time: `O(n + m)`.
- Space: `O(m)`.
Find the first occurrence of a pattern in a text; return `-1` if absent.

```python
def kmp_search(text, pattern):  # Find pattern in text using Knuth-Morris-Pratt.
    if pattern == "":  # Define the empty pattern as matching at the start.
        return 0  # The first valid position is index zero.
    lps = [0] * len(pattern)  # lps[i] stores the longest proper prefix/suffix length.
    length = 0  # Track the current candidate prefix length.
    i = 1  # Start building the prefix table from pattern index one.
    while i < len(pattern):  # Process each pattern position.
        if pattern[i] == pattern[length]:  # Extend the matching prefix when characters match.
            length += 1  # Increase the prefix length.
            lps[i] = length  # Save the length for this position.
            i += 1  # Move to the next pattern position.
        elif length:  # If there is a mismatch but a shorter prefix may still match,
            length = lps[length - 1]  # Fall back using the prefix table.
        else:  # No shorter prefix can be used.
            lps[i] = 0  # Record that no proper prefix/suffix matches here.
            i += 1  # Continue building the table.
    i = j = 0  # Start text and pattern pointers at zero.
    while i < len(text):  # Scan the text once.
        if text[i] == pattern[j]:  # Advance both pointers on a character match.
            i += 1  # Move through the text.
            j += 1  # Move through the pattern.
            if j == len(pattern):  # Check whether the whole pattern matched.
                return i - j  # Return the match's starting index.
        elif j:  # If a mismatch occurs after some matching characters,
            j = lps[j - 1]  # Reuse the longest prefix/suffix information.
        else:  # The first pattern character did not match.
            i += 1  # Advance the text pointer.
    return -1  # No complete match was found.

print(kmp_search("ababcabcabababd", "ababd"))  # Print 10.
```

## 47. Euclidean GCD

**Pattern:** Repeated remainder. The GCD of `a` and `b` equals the GCD of `b` and `a % b`.
- Time: `O(log(min(a, b)))`.
- Space: `O(1)`.

```python
def gcd(a, b):  # Find the greatest common divisor of two integers.
    a, b = abs(a), abs(b)  # Use non-negative magnitudes.
    while b != 0:  # Continue until the remainder becomes zero.
        a, b = b, a % b  # Replace the pair with the divisor and remainder.
    return a  # The final non-zero value is the greatest common divisor.

print(gcd(48, 18))  # Print 6.
```

## 48. Sieve of Eratosthenes

**Pattern:** Mark composite numbers instead of testing each number separately.
- Time: `O(n log log n)`.
- Space: `O(n)`.
Returns all prime numbers up to and including `limit`.

```python
def primes_up_to(limit):  # Generate primes no greater than limit.
    if limit < 2:  # Values below two contain no primes.
        return []  # Return an empty list.
    is_prime = [True] * (limit + 1)  # Initially assume each number is prime.
    is_prime[0] = is_prime[1] = False  # Zero and one are not prime numbers.
    p = 2  # Start with the smallest prime candidate.
    while p * p <= limit:  # Only process candidates up to the square root.
        if is_prime[p]:  # Mark multiples only when p is prime.
            for multiple in range(p * p, limit + 1, p):  # Start at p squared and step by p.
                is_prime[multiple] = False  # Mark this multiple as composite.
        p += 1  # Try the next candidate.
    return [number for number, prime in enumerate(is_prime) if prime]  # Collect the remaining primes.

print(primes_up_to(30))  # Print primes from 2 through 29.
```

## 49. Fast Exponentiation

**Pattern:** Exponentiation by squaring.
- Time: `O(log exponent)` multiplications.
- Space: `O(1)` for this iterative version.
This version supports non-negative integer exponents.

```python
def fast_power(base, exponent):  # Compute base raised to a non-negative integer exponent.
    if exponent < 0:  # This implementation intentionally excludes negative exponents.
        raise ValueError("exponent must be non-negative")  # Explain the limitation.
    result = 1  # Any number to the power zero is one.
    while exponent > 0:  # Process exponent bits until none remain.
        if exponent % 2 == 1:  # Check whether the current exponent is odd.
            result *= base  # Include the current base factor in the result.
        base *= base  # Square the base for the next binary exponent position.
        exponent //= 2  # Shift the exponent one bit to the right.
    return result  # Return the computed power.

print(fast_power(2, 10))  # Print 1024.
```

## 50. Binary Search on Answer: Integer Square Root

**Pattern:** Search a monotonic condition. Find the largest integer `x` where `x*x <= n`.
- Time: `O(log n)`.
- Space: `O(1)`.
Avoids floating-point rounding errors.

```python
def integer_sqrt(n):  # Return floor(sqrt(n)) for a non-negative integer.
    if n < 0:  # Real integer square root is undefined here.
        raise ValueError("n must be non-negative")  # Reject negative input.
    left, right = 0, n  # The answer must lie between zero and n for n >= 1.
    answer = 0  # Store the best valid candidate found.
    while left <= right:  # Search while the candidate interval is not empty.
        mid = (left + right) // 2  # Test the middle candidate.
        if mid * mid <= n:  # Check whether the candidate square fits within n.
            answer = mid  # Save this valid candidate.
            left = mid + 1  # Search for a potentially larger valid answer.
        else:  # The candidate square is too large.
            right = mid - 1  # Search smaller candidates.
    return answer  # Return the largest valid integer.

print(integer_sqrt(27))  # Print 5 because 5² <= 27 < 6².
```

## 51. Sliding Window: Minimum Size Subarray Sum

**Pattern:** Shrink a variable-size window while its sum meets a target.
- Time: `O(n)`.
- Requirement: positive numbers, so expanding/shrinking behaves monotonically.
Returns zero if no qualifying subarray exists.

```python
def min_subarray_len(target, nums):  # Find the shortest subarray with sum at least target.
    left = 0  # Mark the start of the current window.
    total = 0  # Store the current window sum.
    best = float("inf")  # Begin with no known qualifying window.
    for right, num in enumerate(nums):  # Expand the window one value at a time.
        total += num  # Include the new rightmost value.
        while total >= target:  # Shrink while the current window still qualifies.
            best = min(best, right - left + 1)  # Record the shortest length so far.
            total -= nums[left]  # Remove the leftmost value from the window sum.
            left += 1  # Move the window's left edge rightward.
    return 0 if best == float("inf") else best  # Return zero when no valid window exists.

print(min_subarray_len(7, [2, 3, 1, 2, 4, 3]))  # Print 2 for [4, 3].
```

## 52. Longest Increasing Subsequence

**Pattern:** Binary search over tails. This version finds the LIS length, not the sequence itself.
- Time: `O(n log n)`.
- Space: `O(n)`.
The tails list stores the smallest possible tail for each subsequence length.

```python
from bisect import bisect_left  # Import binary search for insertion positions.

def lis_length(nums):  # Find the longest strictly increasing subsequence length.
    tails = []  # tails[k] is the smallest tail found for a subsequence of length k+1.
    for num in nums:  # Process every value.
        index = bisect_left(tails, num)  # Find the first tail greater than or equal to num.
        if index == len(tails):  # If num exceeds every existing tail,
            tails.append(num)  # extend the longest subsequence length.
        else:  # Otherwise, num can improve a tail at this length.
            tails[index] = num  # Keep the smaller tail to allow future extensions.
    return len(tails)  # The number of tail entries is the LIS length.

print(lis_length([10, 9, 2, 5, 3, 7, 101, 18]))  # Print 4.
```

## 53. Flood Fill

**Pattern:** DFS/BFS on a grid. Replace a connected region of the same starting color.
- Time: `O(rows × columns)`.
- Space: up to `O(rows × columns)` for the queue/stack and visited region.

```python
def flood_fill(image, start_row, start_col, new_color):  # Recolor a connected region in a grid.
    if not image or not image[0]:  # Handle an empty grid.
        return image  # Nothing can be changed.
    old_color = image[start_row][start_col]  # Read the starting pixel's original color.
    if old_color == new_color:  # Avoid repeatedly processing an unchanged region.
        return image  # The desired color is already present.
    rows, cols = len(image), len(image[0])  # Store the grid dimensions.
    stack = [(start_row, start_col)]  # Start DFS from the selected pixel.
    image[start_row][start_col] = new_color  # Mark the start immediately to avoid revisiting it.
    while stack:  # Process pixels until the region is exhausted.
        row, col = stack.pop()  # Take one pixel from the stack.
        for dr, dc in ((1, 0), (-1, 0), (0, 1), (0, -1)):  # Define four orthogonal directions.
            nr, nc = row + dr, col + dc  # Calculate a neighboring position.
            if 0 <= nr < rows and 0 <= nc < cols and image[nr][nc] == old_color:  # Check bounds and color.
                image[nr][nc] = new_color  # Recolor the neighbor before scheduling it.
                stack.append((nr, nc))  # Continue the fill from this neighbor.
    return image  # Return the modified grid.

grid = [[1, 1, 1], [1, 1, 0], [1, 0, 1]]  # Create a sample image as a 2D list.
print(flood_fill(grid, 0, 0, 2))  # Recolor the connected top-left region with 2.
```

## 54. Number of Islands

**Pattern:** Grid traversal + visited marking.
- Time: `O(rows × columns)`.
- Space: `O(rows × columns)` in the worst case.
This version treats `'1'` as land and `'0'` as water and does not mutate the input.

```python
def count_islands(grid):  # Count connected groups of land in a character grid.
    if not grid or not grid[0]:  # Handle an empty grid.
        return 0  # An empty grid contains no islands.
    rows, cols = len(grid), len(grid[0])  # Store grid dimensions.
    visited = set()  # Remember land cells already explored.
    islands = 0  # Count connected land components.
    for r in range(rows):  # Visit each row.
        for c in range(cols):  # Visit each column in this row.
            if grid[r][c] != "1" or (r, c) in visited:  # Skip water and previously explored land.
                continue  # Continue scanning the grid.
            islands += 1  # A new unvisited land cell starts a new island.
            stack = [(r, c)]  # Begin DFS for this island.
            visited.add((r, c))  # Mark the starting cell immediately.
            while stack:  # Explore the entire connected component.
                cr, cc = stack.pop()  # Take one land cell.
                for dr, dc in ((1, 0), (-1, 0), (0, 1), (0, -1)):  # Check four neighbors.
                    nr, nc = cr + dr, cc + dc  # Calculate neighbor coordinates.
                    if 0 <= nr < rows and 0 <= nc < cols and grid[nr][nc] == "1" and (nr, nc) not in visited:  # Check valid unvisited land.
                        visited.add((nr, nc))  # Mark the neighbor before pushing it.
                        stack.append((nr, nc))  # Explore the neighbor later.
    return islands  # Return the number of islands.

sample = [["1", "1", "0"], ["0", "1", "0"], ["1", "0", "1"]]  # Define a grid with three islands.
print(count_islands(sample))  # Print 3.
```

## 55. A* Search (Grid Pathfinding)

**Pattern:** Best-first graph search using `f = g + h`.
- `g`: known cost from start; `h`: estimated cost to goal.
- Manhattan distance is suitable for four-direction movement with equal step costs.
- This example returns a path or `None`.

```python
import heapq  # Import the min-heap used to select the lowest estimated total cost.

def astar_grid(blocked, start, goal, rows, cols):  # Find a path through a four-direction grid.
    def heuristic(cell):  # Estimate the remaining distance to the goal.
        return abs(cell[0] - goal[0]) + abs(cell[1] - goal[1])  # Manhattan distance.
    frontier = [(heuristic(start), 0, start)]  # Store (estimated total cost, actual cost, cell).
    came_from = {start: None}  # Record each cell's predecessor for path reconstruction.
    cost_so_far = {start: 0}  # Store the best known cost to each discovered cell.
    while frontier:  # Continue while there are candidate cells.
        _, cost, current = heapq.heappop(frontier)  # Expand the candidate with the lowest estimated cost.
        if current == goal:  # Check whether the goal has been reached.
            path = []  # Prepare to reconstruct the path backward.
            while current is not None:  # Follow predecessors until the start's predecessor.
                path.append(current)  # Add this cell to the reversed path.
                current = came_from[current]  # Move to the previous cell.
            return path[::-1]  # Reverse the collected cells to get start-to-goal order.
        if cost != cost_so_far[current]:  # Ignore outdated candidates for this cell.
            continue  # A better route has already been found.
        for dr, dc in ((1, 0), (-1, 0), (0, 1), (0, -1)):  # Explore four possible moves.
            neighbor = (current[0] + dr, current[1] + dc)  # Calculate the neighbor coordinates.
            r, c = neighbor  # Unpack the neighbor's row and column.
            if not (0 <= r < rows and 0 <= c < cols) or neighbor in blocked:  # Reject off-grid or blocked cells.
                continue  # Try another direction.
            new_cost = cost + 1  # Each move has cost one.
            if new_cost < cost_so_far.get(neighbor, float("inf")):  # Check whether this route is better.
                cost_so_far[neighbor] = new_cost  # Save the improved cost.
                came_from[neighbor] = current  # Save the predecessor for path reconstruction.
                heapq.heappush(frontier, (new_cost + heuristic(neighbor), new_cost, neighbor))  # Schedule the neighbor.
    return None  # No path reaches the goal.

print(astar_grid({(1, 1)}, (0, 0), (2, 2), 3, 3))  # Find a path around the blocked cell.
```

## 56. Trie: Prefix Search

**Pattern:** Prefix tree. Each edge represents a character; terminal nodes mark complete words.
- Insert/search: `O(L)` average for word length L.
- Space: proportional to stored characters.

```python
class TrieNode:  # Define one node in the prefix tree.
    def __init__(self):  # Initialize a node with no children or complete word.
        self.children = {}  # Map characters to child nodes.
        self.is_word = False  # Track whether a word ends at this node.

class Trie:  # Define the trie data structure.
    def __init__(self):  # Initialize an empty trie.
        self.root = TrieNode()  # Create the root node for all words.
    def insert(self, word):  # Add a word to the trie.
        node = self.root  # Start at the root.
        for char in word:  # Process every character in the word.
            if char not in node.children:  # Check whether this character edge exists.
                node.children[char] = TrieNode()  # Create a child node when needed.
            node = node.children[char]  # Follow the edge for this character.
        node.is_word = True  # Mark the final node as a complete word.
    def search(self, word):  # Check whether the full word exists.
        node = self.root  # Start at the root.
        for char in word:  # Follow the word's character path.
            if char not in node.children:  # Stop if an edge is missing.
                return False  # The word is not stored.
            node = node.children[char]  # Move to the next character node.
        return node.is_word  # Only return True if a complete word ends here.
    def starts_with(self, prefix):  # Check whether any word begins with a prefix.
        node = self.root  # Start at the root.
        for char in prefix:  # Follow each prefix character.
            if char not in node.children:  # Check whether the prefix path breaks.
                return False  # No stored word can have this prefix.
            node = node.children[char]  # Continue along the prefix path.
        return True  # Every prefix character existed.

trie = Trie()  # Create an empty trie.
trie.insert("python")  # Insert one word.
trie.insert("pycharm")  # Insert another word sharing a prefix.
print(trie.search("python"), trie.starts_with("py"))  # Print True True.
```

## 57. LRU Cache

**Pattern:** Hash map + doubly linked ordering is the usual `O(1)` design.
Python's `OrderedDict` provides a compact standard-library implementation.
- Average get/set: `O(1)`.

```python
from collections import OrderedDict  # Import an ordered dictionary with move-to-end support.

class LRUCache:  # Define a least-recently-used cache.
    def __init__(self, capacity):  # Set the maximum number of cached entries.
        if capacity < 0:  # Reject an invalid capacity.
            raise ValueError("capacity must be non-negative")  # Explain the constraint.
        self.capacity = capacity  # Store the maximum allowed number of entries.
        self.data = OrderedDict()  # Keep keys ordered from least to most recently used.
    def get(self, key):  # Retrieve a cached value.
        if key not in self.data:  # Check whether the key is absent.
            return -1  # Use -1 to represent a cache miss in this example.
        self.data.move_to_end(key)  # Mark this key as the most recently used.
        return self.data[key]  # Return the cached value.
    def put(self, key, value):  # Insert or update a cached value.
        if self.capacity == 0:  # A zero-capacity cache cannot store entries.
            return  # Leave the cache empty.
        self.data[key] = value  # Insert the value or update the existing key.
        self.data.move_to_end(key)  # Mark this key as most recently used.
        if len(self.data) > self.capacity:  # Check whether capacity was exceeded.
            self.data.popitem(last=False)  # Remove the least recently used entry.

cache = LRUCache(2)  # Create a cache that can hold two entries.
cache.put("a", 1)  # Store key a.
cache.put("b", 2)  # Store key b.
cache.get("a")  # Access a, making it the most recently used entry.
cache.put("c", 3)  # Evict b because it is now least recently used.
print(cache.get("b"), cache.get("c"))  # Print -1 3.
```

## 58. Rate Limiter: Token Bucket

**Pattern:** Track tokens that refill over time. Each request consumes one token.
- Space: `O(1)`.
- Uses a clock, so results depend on elapsed time.
This teaching version is not thread-safe; production systems need synchronization.

```python
import time  # Import a monotonic clock for measuring elapsed time.

class TokenBucket:  # Define a simple token-bucket rate limiter.
    def __init__(self, capacity, refill_rate):  # Set bucket size and tokens added per second.
        self.capacity = capacity  # Store the maximum number of tokens.
        self.refill_rate = refill_rate  # Store how quickly tokens are replenished.
        self.tokens = capacity  # Begin with a full bucket.
        self.updated_at = time.monotonic()  # Record the time of the last refill calculation.
    def allow_request(self):  # Decide whether a request can proceed.
        now = time.monotonic()  # Read the current monotonic time.
        elapsed = now - self.updated_at  # Calculate time since the last update.
        self.tokens = min(self.capacity, self.tokens + elapsed * self.refill_rate)  # Refill without exceeding capacity.
        self.updated_at = now  # Save the timestamp for the next request.
        if self.tokens >= 1:  # Check whether at least one token is available.
            self.tokens -= 1  # Consume one token for this request.
            return True  # Allow the request.
        return False  # Reject the request when no token is available.

limiter = TokenBucket(capacity=3, refill_rate=1)  # Allow bursts of three and refill one token per second.
print(limiter.allow_request())  # Print True for the first request.
```

## 59. Exponential Backoff

**Pattern:** Increase the wait between retries after transient failures.
- Delay often grows as `base_delay × 2^attempt`, usually with jitter.
- This example only computes delays; it does not make network requests.
In production, cap the delay and retry only safe/transient failures.

```python
import random  # Import a random-number generator for jitter.

def backoff_delay(attempt, base_delay=1.0, max_delay=30.0):  # Calculate a retry wait.
    exponential = base_delay * (2 ** attempt)  # Double the base delay for each attempt.
    capped = min(exponential, max_delay)  # Prevent the delay from growing without limit.
    jitter = random.uniform(0, 0.25 * capped)  # Add a random delay to spread concurrent retries.
    return capped + jitter  # Return the capped delay plus jitter.

print(round(backoff_delay(3), 2))  # Print a delay around eight to ten seconds.
```

## 60. Reservoir Sampling

**Pattern:** Choose a uniformly random sample of k items from a stream of unknown length.
- Time: `O(n)` for n items.
- Extra space: `O(k)`.
Each item has an equal chance of being included.

```python
import random  # Import random integer selection.

def reservoir_sample(stream, k):  # Select k stream items uniformly without storing the whole stream.
    if k < 0:  # Reject negative sample sizes.
        raise ValueError("k must be non-negative")  # Explain the valid input.
    reservoir = []  # Store the sample currently selected.
    for i, item in enumerate(stream):  # Process the stream with a zero-based index.
        if i < k:  # Fill the reservoir with the first k items.
            reservoir.append(item)  # Add this item to the initial sample.
        else:  # Consider replacement for later items.
            j = random.randint(0, i)  # Choose uniformly among all positions seen so far.
            if j < k:  # Replace only if the random position falls inside the reservoir.
                reservoir[j] = item  # Replace the selected reservoir entry.
    return reservoir  # Return the sample, which may be smaller than k if input is short.

print(reservoir_sample(range(100), 5))  # Print a random sample of five values.
```

## 61. Majority Element (Boyer-Moore Voting)

**Pattern:** Cancel pairs of different values to find a majority candidate.
- Time: `O(n)`.
- Space: `O(1)`.
A majority element must appear more than n/2 times. This version verifies the candidate.

```python
def majority_element(nums):  # Find a value occurring more than half the time.
    candidate = None  # Store the current majority candidate.
    count = 0  # Track the candidate's balance against other values.
    for num in nums:  # Process each value once.
        if count == 0:  # When no candidate has a positive balance,
            candidate = num  # choose the current value as the new candidate.
        count += 1 if num == candidate else -1  # Match increases balance; mismatch decreases it.
    if nums.count(candidate) > len(nums) // 2:  # Verify that the candidate is a true majority.
        return candidate  # Return the confirmed majority value.
    return None  # Return None when no majority exists.

print(majority_element([2, 2, 1, 2, 3, 2, 2]))  # Print 2.
```

## 62. Missing Number (XOR)

**Pattern:** XOR cancellation. Equal values XOR to zero, and `x ^ 0 == x`.
- Time: `O(n)`.
- Space: `O(1)`.
Input must contain distinct values from 0 through n with one missing.

```python
def missing_number(nums):  # Find the missing value from the range 0 through n.
    result = len(nums)  # Include n in the XOR accumulation.
    for i, num in enumerate(nums):  # Pair each index with the value stored there.
        result ^= i ^ num  # Equal present values cancel each other through XOR.
    return result  # The only remaining value is the missing number.

print(missing_number([3, 0, 1]))  # Print 2.
```

## 63. Single Number (XOR)

**Pattern:** XOR all values; pairs cancel out.
- Time: `O(n)`.
- Space: `O(1)`.
Assumes every value appears twice except one value.

```python
def single_number(nums):  # Find the value that appears exactly once.
    result = 0  # XOR identity: zero does not change any value.
    for num in nums:  # Process each number.
        result ^= num  # Cancel matching pairs and retain the unpaired value.
    return result  # Return the unique value.

print(single_number([4, 1, 2, 1, 2]))  # Print 4.
```

## 64. Palindrome Check (Two Pointers)

**Pattern:** Compare mirrored characters from both ends.
- Time: `O(n)`.
- Space: `O(1)` extra space after normalization; normalization itself creates a string.
This example ignores spaces, punctuation, and letter case.

```python
def is_palindrome(text):  # Check whether normalized text reads the same in both directions.
    normalized = "".join(char.lower() for char in text if char.isalnum())  # Keep lowercase letters and digits.
    left, right = 0, len(normalized) - 1  # Start pointers at both ends.
    while left < right:  # Continue until pointers meet or cross.
        if normalized[left] != normalized[right]:  # Compare the mirrored characters.
            return False  # One mismatch proves it is not a palindrome.
        left += 1  # Move the left pointer inward.
        right -= 1  # Move the right pointer inward.
    return True  # Every mirrored pair matched.

print(is_palindrome("A man, a plan, a canal: Panama"))  # Print True.
```

## 65. Run-Length Encoding

**Pattern:** Group consecutive repeated characters.
- Time: `O(n)`.
- Space: `O(n)` for the encoded output.
This is a simple educational encoding, not a general-purpose compression format.

```python
def run_length_encode(text):  # Encode each run as character followed by its count.
    if not text:  # Handle an empty string.
        return ""  # The encoding of empty text is empty.
    result = []  # Collect encoded runs as strings.
    current = text[0]  # Start the first run with the first character.
    count = 1  # The first run contains one character.
    for char in text[1:]:  # Process the remaining characters.
        if char == current:  # Check whether this character continues the current run.
            count += 1  # Increase the run length.
        else:  # The current run has ended.
            result.append(f"{current}{count}")  # Store the completed run.
            current = char  # Begin a new run with this character.
            count = 1  # Reset the run length.
    result.append(f"{current}{count}")  # Store the final run after the loop.
    return "".join(result)  # Join encoded runs into one string.

print(run_length_encode("aaabbc"))  # Print a3b2c1.
```

## 67. Lowest Common Ancestor in a Binary Tree

**Pattern:** Recursive tree search.

```python
def lowest_common_ancestor(root, p, q):  # Find the lowest node containing both targets in its subtree.
    if root is None or root is p or root is q:  # Stop at an empty node or either target.
        return root  # This node is a candidate ancestor.
    left = lowest_common_ancestor(root.left, p, q)  # Search the left subtree.
    right = lowest_common_ancestor(root.right, p, q)  # Search the right subtree.
    if left and right:  # If targets were found on opposite sides,
        return root  # the current node is their lowest common ancestor.
    return left or right  # Otherwise propagate the result found on one side.

p = TreeNode(4)  # Create target node p.
q = TreeNode(5)  # Create target node q.
root = TreeNode(3, TreeNode(5, p, q), TreeNode(1))  # Build a tree containing both target nodes.
print(lowest_common_ancestor(root, p, q).value)  # Print 5.
```

## 68. Min Stack

**Pattern:** Store the current minimum alongside each stack value.
- Push/pop/top/get-min: `O(1)`.
- Space: `O(n)`.
Each stack entry keeps the minimum up to that entry.

```python
class MinStack:  # Define a stack that can return its minimum quickly.
    def __init__(self):  # Initialize an empty stack.
        self.stack = []  # Store pairs of (value, minimum so far).
    def push(self, value):  # Add a new value.
        current_min = value if not self.stack else min(value, self.stack[-1][1])  # Compute the new minimum.
        self.stack.append((value, current_min))  # Store the value and its associated minimum.
    def pop(self):  # Remove and return the top value.
        return self.stack.pop()[0]  # Discard the stored minimum and return the value.
    def top(self):  # Read the top value without removing it.
        return self.stack[-1][0]  # Return the value part of the last pair.
    def get_min(self):  # Read the current minimum in constant time.
        return self.stack[-1][1]  # Return the minimum saved with the top entry.

stack = MinStack()  # Create an empty minimum stack.
stack.push(3)  # Push the first value.
stack.push(1)  # Push a new minimum.
stack.push(2)  # Push a value above the minimum.
print(stack.get_min())  # Print 1.
```

## 69. Queue Using Two Stacks

**Pattern:** Amortized queue. One stack accepts incoming items; another serves outgoing items.
- Amortized enqueue/dequeue: `O(1)`.
- Space: `O(n)`.

```python
class QueueWithStacks:  # Implement FIFO behavior using two LIFO stacks.
    def __init__(self):  # Initialize the two stacks.
        self.in_stack = []  # Hold newly enqueued items.
        self.out_stack = []  # Hold items ready to dequeue.
    def enqueue(self, value):  # Add an item to the back of the queue.
        self.in_stack.append(value)  # Push it onto the input stack.
    def dequeue(self):  # Remove the oldest item from the queue.
        if not self.out_stack:  # Refill only when the output stack is empty.
            while self.in_stack:  # Move every waiting item across.
                self.out_stack.append(self.in_stack.pop())  # Reverse the order into FIFO order.
        if not self.out_stack:  # Check whether both stacks are empty.
            raise IndexError("dequeue from empty queue")  # Report an empty-queue error.
        return self.out_stack.pop()  # Pop the oldest item.

queue = QueueWithStacks()  # Create an empty queue.
queue.enqueue("first")  # Enqueue the first item.
queue.enqueue("second")  # Enqueue the second item.
print(queue.dequeue())  # Print first.
```

This simple version uses extra lists for readability.
- Returns the kth smallest value where k is one-based.

```python
def kth_smallest(nums, k):  # Find the kth smallest value using partitioning.
    if not 1 <= k <= len(nums):  # Validate one-based k.
        raise ValueError("k must be between 1 and len(nums)")  # Explain the allowed range.
    pivot = nums[len(nums) // 2]  # Choose a pivot value.
    lower = [x for x in nums if x < pivot]  # Collect values below the pivot.
    equal = [x for x in nums if x == pivot]  # Collect values equal to the pivot.
    upper = [x for x in nums if x > pivot]  # Collect values above the pivot.
    if k <= len(lower):  # Check whether the answer lies in the lower partition.
        return kth_smallest(lower, k)  # Search the lower partition with the same rank.
    if k <= len(lower) + len(equal):  # Check whether the rank falls among pivot-equal values.
        return pivot  # The pivot is the kth smallest value.
    return kth_smallest(upper, k - len(lower) - len(equal))  # Adjust rank and search the upper partition.

print(kth_smallest([7, 10, 4, 3, 20, 15], 3))  # Print 7.
```

## 72. Counting Sort

**Pattern:** Count occurrences when keys are integers in a manageable range.
- Time: `O(n + range)`.
- Space: `O(range + n)` for this output-producing version.
Best when the value range is not enormous compared with the input size.

```python
def counting_sort(nums):  # Sort a list of integers using frequency counts.
    if not nums:  # Handle empty input.
        return []  # An empty list is already sorted.
    low, high = min(nums), max(nums)  # Find the inclusive range of values.
    counts = [0] * (high - low + 1)  # Allocate one counter for every possible value.
    for num in nums:  # Count each input value.
        counts[num - low] += 1  # Shift by low so negative integers are supported.
    result = []  # Store the sorted output.
    for offset, count in enumerate(counts):  # Visit values from low to high.
        result.extend([offset + low] * count)  # Add each value as many times as it appeared.
    return result  # Return the sorted list.

print(counting_sort([4, -1, 2, 4, 0, -1]))  # Print [-1, -1, 0, 2, 4, 4].
```

## 73. Bucket Sort for Values in [0, 1)

**Pattern:** Distribute values into buckets, sort each bucket, then concatenate.
- Expected time can be `O(n)` for well-distributed input.
- This implementation uses Python's sorting inside each bucket.
Input values must satisfy `0 <= x < 1`.

```python
def bucket_sort_unit_interval(nums):  # Sort real values in the interval [0, 1).
    if any(not 0 <= x < 1 for x in nums):  # Verify every input belongs to the expected range.
        raise ValueError("all values must be in [0, 1)")  # Explain the bucket-sort requirement.
    if not nums:  # Handle an empty list.
        return []  # Return no values.
    buckets = [[] for _ in nums]  # Create n buckets for n input values.
    for num in nums:  # Distribute each value into a bucket.
        index = int(num * len(nums))  # Map the value to a bucket index.
        buckets[index].append(num)  # Add the value to its bucket.
    result = []  # Store the sorted output.
    for bucket in buckets:  # Process buckets from low index to high index.
        result.extend(sorted(bucket))  # Sort each bucket and append its values.
    return result  # Return all values in sorted order.

print(bucket_sort_unit_interval([0.42, 0.32, 0.23, 0.52, 0.25]))  # Print values in ascending order.
```

## 74. Heap Sort

**Pattern:** Heap-based sorting.
- Time: `O(n log n)`.
- Space: `O(n)` for this straightforward output-producing version.
Python's `heapq` is a min-heap.

```python
import heapq  # Import the standard-library heap implementation.

def heap_sort(nums):  # Sort values by repeatedly extracting the smallest.
    heap = nums.copy()  # Copy the input so it remains unchanged.
    heapq.heapify(heap)  # Transform the list into a min-heap in linear time.
    result = []  # Store values in ascending order.
    while heap:  # Continue until every value has been removed.
        result.append(heapq.heappop(heap))  # Extract and append the smallest remaining value.
    return result  # Return the sorted output.

print(heap_sort([5, 1, 9, 2, 6]))  # Print [1, 2, 5, 6, 9].
```

## 75. Exponential Search

**Pattern:** Find a range by doubling the index, then binary-search inside it.
- Time: `O(log i)` where i is the target position; worst case `O(log n)`.
- Requirement: sorted ascending input.

```python
def exponential_search(nums, target):  # Search a sorted list by expanding the candidate range.
    if not nums:  # Handle empty input.
        return -1  # The target cannot be present.
    if nums[0] == target:  # Check the first element before expanding.
        return 0  # Return the first index immediately.
    bound = 1  # Start with the first index after zero.
    while bound < len(nums) and nums[bound] < target:  # Double the range while values are too small.
        bound *= 2  # Expand the search boundary exponentially.
    left = bound // 2  # Start at the previous boundary.
    right = min(bound, len(nums) - 1)  # Clamp the right boundary to the list.
    while left <= right:  # Binary-search the narrowed range.
        mid = (left + right) // 2  # Find the midpoint.
        if nums[mid] == target:  # Check for a match.
            return mid  # Return the matching index.
        if nums[mid] < target:  # If the midpoint is too small,
            left = mid + 1  # search right.
        else:  # If the midpoint is too large,
            right = mid - 1  # search left.
    return -1  # Return -1 if the target is absent.

print(exponential_search([2, 4, 6, 8, 10, 12, 14], 12))  # Print 5.
```

## 76. Jump Search

**Pattern:** Jump by blocks, then linearly search the matching block.
- Time: `O(√n)`.
- Requirement: sorted ascending input.

```python
from math import isqrt  # Import integer square root for the jump size.

def jump_search(nums, target):  # Search a sorted list by jumping through blocks.
    n = len(nums)  # Store the number of values.
    if n == 0:  # Handle empty input.
        return -1  # The target cannot be found.
    step = max(1, isqrt(n))  # Choose a block size near the square root of n.
    start = 0  # Mark the start of the current block.
    end = step  # Mark the end boundary of the current block.
    while start < n and nums[min(end, n) - 1] < target:  # Skip blocks whose last value is too small.
        start = end  # Move the block start forward.
        end += step  # Move the block end forward by one jump.
        if start >= n:  # Check whether all blocks were skipped.
            return -1  # The target is larger than every value.
    for index in range(start, min(end, n)):  # Search the candidate block linearly.
        if nums[index] == target:  # Check for the target value.
            return index  # Return its index.
        if nums[index] > target:  # Sorted order means the target cannot appear later in this block.
            break  # Stop scanning the block.
    return -1  # The target was not found.

print(jump_search([1, 3, 5, 7, 9, 11, 13], 9))  # Print 4.
```

## 77. Boyer-Moore String Search (Bad-Character Rule)

**Pattern:** Align the pattern and skip ahead based on the mismatched text character.
- Average performance can be strong; worst-case depends on variant and input.
- This educational version uses the bad-character heuristic.

```python
def boyer_moore_bad_character(text, pattern):  # Search using right-to-left pattern comparison.
    if pattern == "":  # Define empty pattern behavior.
        return 0  # Empty pattern matches at index zero.
    last = {char: i for i, char in enumerate(pattern)}  # Store each character's last pattern position.
    shift = 0  # Store the current alignment of pattern against text.
    while shift <= len(text) - len(pattern):  # Continue while the pattern fits in the text.
        j = len(pattern) - 1  # Start comparison at the pattern's rightmost character.
        while j >= 0 and pattern[j] == text[shift + j]:  # Move left while characters match.
            j -= 1  # Check the preceding pattern character.
        if j < 0:  # All pattern characters matched.
            return shift  # Return the alignment where the match begins.
        bad_char = text[shift + j]  # Read the mismatching text character.
        shift += max(1, j - last.get(bad_char, -1))  # Shift to align the last matching pattern occurrence.
    return -1  # Return -1 when no alignment matches.

print(boyer_moore_bad_character("HERE IS A SIMPLE EXAMPLE", "EXAMPLE"))  # Print 17.
```

## 78. Rolling Hash / Rabin-Karp

**Pattern:** Compare rolling hashes of text windows to find pattern candidates.
- Expected time: `O(n + m)` with a suitable hash/modulus; worst case can be `O(nm)` due to collisions.
- Space: `O(1)` beyond input for this single-pattern version.
A real match is verified after a hash match.

```python
def rabin_karp(text, pattern):  # Find a pattern using a rolling polynomial hash.
    n, m = len(text), len(pattern)  # Store text and pattern lengths.
    if m == 0:  # Handle an empty pattern.
        return 0  # Empty pattern matches at index zero.
    if m > n:  # A longer pattern cannot fit.
        return -1  # Report no match.
    base, mod = 256, 1_000_000_007  # Choose a character base and a large prime modulus.
    highest = pow(base, m - 1, mod)  # Precompute the leading character's multiplier.
    pattern_hash = 0  # Initialize the pattern's hash.
    window_hash = 0  # Initialize the first text window's hash.
    for i in range(m):  # Hash each character of the first window and pattern.
        pattern_hash = (pattern_hash * base + ord(pattern[i])) % mod  # Add pattern character to its hash.
        window_hash = (window_hash * base + ord(text[i])) % mod  # Add text character to its hash.
    for start in range(n - m + 1):  # Slide the pattern-sized window across the text.
        if pattern_hash == window_hash and text[start:start + m] == pattern:  # Verify possible hash matches.
            return start  # Return the first verified occurrence.
        if start < n - m:  # Update the hash only if another window exists.
            window_hash = (window_hash - ord(text[start]) * highest) % mod  # Remove the outgoing character.
            window_hash = (window_hash * base + ord(text[start + m])) % mod  # Add the incoming character.
    return -1  # No verified occurrence was found.

print(rabin_karp("abracadabra", "cada"))  # Print 4.
```

## 79. Maximum Product Subarray

**Pattern:** Track both maximum and minimum products ending at each index because a negative value can swap them.
- Time: `O(n)`.
- Space: `O(1)`.

```python
def max_product_subarray(nums):  # Find the maximum product of a non-empty contiguous subarray.
    if not nums:  # Require at least one number.
        raise ValueError("nums must not be empty")  # Explain invalid input.
    max_ending = min_ending = best = nums[0]  # Initialize max, min, and global best from the first value.
    for num in nums[1:]:  # Process every later value.
        if num < 0:  # A negative number swaps the roles of maximum and minimum products.
            max_ending, min_ending = min_ending, max_ending  # Swap the two running products.
        max_ending = max(num, max_ending * num)  # Choose a new subarray or extend the maximum product.
        min_ending = min(num, min_ending * num)  # Choose a new subarray or extend the minimum product.
        best = max(best, max_ending)  # Update the best product found anywhere.
    return best  # Return the maximum product.

print(max_product_subarray([2, 3, -2, 4]))  # Print 6.
```

## 80. Word Break

**Pattern:** Dynamic programming over string prefixes.
- Time: commonly `O(n²)` substring checks plus set lookup costs.
- Space: `O(n)` for the DP table, excluding the dictionary.
Checks whether a string can be segmented into dictionary words.

```python
def word_break(text, words):  # Determine whether text can be segmented into words.
    word_set = set(words)  # Use a set for average constant-time word lookup.
    dp = [False] * (len(text) + 1)  # dp[i] means text[:i] can be segmented.
    dp[0] = True  # The empty prefix can always be segmented.
    for end in range(1, len(text) + 1):  # Consider every non-empty prefix ending position.
        for start in range(end):  # Try every possible final word start.
            if dp[start] and text[start:end] in word_set:  # Check a segment after a valid prefix.
                dp[end] = True  # Mark this prefix segmentable.
                break  # No other split is needed for this end position.
    return dp[len(text)]  # Return whether the full string is segmentable.

print(word_break("leetcode", ["leet", "code"]))  # Print True.
```

## 81. House Robber

**Pattern:** Dynamic programming with two rolling states.
- Time: `O(n)`.
- Space: `O(1)`.
Choose non-adjacent values to maximize the total.

```python
def house_robber(nums):  # Find the maximum sum of non-adjacent values.
    two_back = 0  # Best total before the previous position.
    one_back = 0  # Best total up through the previous position.
    for value in nums:  # Consider each house value.
        current = max(one_back, two_back + value)  # Skip this house or take it after skipping the previous.
        two_back = one_back  # Shift the older state forward.
        one_back = current  # Save the best total through the current house.
    return one_back  # Return the best total for the entire list.

print(house_robber([2, 7, 9, 3, 1]))  # Print 12.
```

## 82. Climbing Stairs

**Pattern:** Fibonacci-style dynamic programming.
- Time: `O(n)`.
- Space: `O(1)`.
Assumes each move can climb one or two steps.

```python
def climbing_stairs(n):  # Count ways to climb n steps using one-step or two-step moves.
    if n < 0:  # Negative stair counts are invalid.
        return 0  # There are no ways to climb a negative number of steps.
    previous, current = 1, 1  # There is one way for zero steps and one step.
    for _ in range(2, n + 1):  # Calculate counts from step two through n.
        previous, current = current, previous + current  # Each step combines the prior two counts.
    return current if n >= 1 else 1  # Return one for zero steps, otherwise the computed count.

print(climbing_stairs(5))  # Print 8.
```

## 83. Pascal's Triangle

**Pattern:** Build each row from the previous row.
- Time and output space: `O(n²)` for n rows.
Each interior value is the sum of the two values above it.

```python
def pascal_triangle(rows):  # Generate the first rows of Pascal's triangle.
    triangle = []  # Store all generated rows.
    for row_index in range(rows):  # Build each row from index zero onward.
        row = [1] * (row_index + 1)  # Every row begins and ends with one.
        for col in range(1, row_index):  # Visit only interior positions.
            row[col] = triangle[row_index - 1][col - 1] + triangle[row_index - 1][col]  # Sum the two values above.
        triangle.append(row)  # Add the completed row to the triangle.
    return triangle  # Return all rows.

print(pascal_triangle(5))  # Print the first five rows.
```

## 86. Spiral Matrix Traversal

**Pattern:** Maintain top, bottom, left, and right boundaries.
- Time: `O(rows × columns)`.
- Space: `O(rows × columns)` for the output.

```python
def spiral_order(matrix):  # Return matrix values in clockwise spiral order.
    if not matrix or not matrix[0]:  # Handle empty matrices.
        return []  # There are no values to visit.
    top, bottom = 0, len(matrix) - 1  # Set the first and last row boundaries.
    left, right = 0, len(matrix[0]) - 1  # Set the first and last column boundaries.
    result = []  # Store values in spiral order.
    while top <= bottom and left <= right:  # Continue while an unvisited rectangle remains.
        for col in range(left, right + 1):  # Traverse the top row from left to right.
            result.append(matrix[top][col])  # Append each top-row value.
        top += 1  # Exclude the row just visited.
        for row in range(top, bottom + 1):  # Traverse the right column downward.
            result.append(matrix[row][right])  # Append each right-column value.
        right -= 1  # Exclude the column just visited.
        if top <= bottom:  # Check whether a bottom row remains.
            for col in range(right, left - 1, -1):  # Traverse the bottom row from right to left.
                result.append(matrix[bottom][col])  # Append each bottom-row value.
            bottom -= 1  # Exclude the row just visited.
        if left <= right:  # Check whether a left column remains.
            for row in range(bottom, top - 1, -1):  # Traverse the left column upward.
                result.append(matrix[row][left])  # Append each left-column value.
            left += 1  # Exclude the column just visited.
    return result  # Return the collected spiral sequence.

print(spiral_order([[1, 2, 3], [4, 5, 6], [7, 8, 9]]))  # Print [1, 2, 3, 6, 9, 8, 7, 4, 5].
```

## 87. Binary Search: Find Peak Element

**Pattern:** Binary search on a slope.
- Time: `O(log n)`.
- A peak is a value greater than its existing neighbors.
This returns one peak index; the input need not be sorted.

```python
def find_peak(nums):  # Find an index of a peak element.
    if not nums:  # Check for empty input.
        return -1  # No peak exists in an empty list.
    left, right = 0, len(nums) - 1  # Search the full index range.
    while left < right:  # Stop when both boundaries meet.
        mid = (left + right) // 2  # Choose the midpoint.
        if nums[mid] < nums[mid + 1]:  # An upward slope means a peak lies to the right.
            left = mid + 1  # Keep the rising half.
        else:  # A downward slope means a peak exists at mid or to its left.
            right = mid  # Keep the left half including mid.
    return left  # The converged index is a peak.

values = [1, 2, 1, 3, 5, 6, 4]  # Define a list with peaks.
index = find_peak(values)  # Find one peak position.
print(index, values[index])  # Print the index and its peak value.
```

## 88. Search in Rotated Sorted Array

**Pattern:** Modified binary search. At least one half is sorted at each step.
- Time: `O(log n)` when values are distinct.
- Assumes a rotated ascending array with distinct values.

```python
def search_rotated(nums, target):  # Find target in a rotated sorted array with distinct values.
    left, right = 0, len(nums) - 1  # Start with the full array.
    while left <= right:  # Continue while a candidate range exists.
        mid = (left + right) // 2  # Choose the midpoint.
        if nums[mid] == target:  # Check for a match.
            return mid  # Return the matching index.
        if nums[left] <= nums[mid]:  # Check whether the left half is sorted.
            if nums[left] <= target < nums[mid]:  # Check whether target lies in that sorted half.
                right = mid - 1  # Search the left half.
            else:  # Otherwise target must lie in the other half.
                left = mid + 1  # Search the right half.
        else:  # The right half must be sorted.
            if nums[mid] < target <= nums[right]:  # Check whether target lies in the sorted right half.
                left = mid + 1  # Search the right half.
            else:  # Otherwise target lies in the left half.
                right = mid - 1  # Search the left half.
    return -1  # Return -1 if target is absent.

print(search_rotated([4, 5, 6, 7, 0, 1, 2], 0))  # Print 4.
```

## 89. Find First and Last Position

**Pattern:** Two binary searches for the lower and upper boundaries.
- Time: `O(log n)`.
- Requirement: sorted ascending input.

```python
def search_range(nums, target):  # Return the first and last indices of target.
    def bound(find_first):  # Search one boundary of the target range.
        left, right = 0, len(nums)  # Use a half-open search interval.
        while left < right:  # Continue until the insertion boundary is found.
            mid = (left + right) // 2  # Calculate the midpoint.
            if nums[mid] > target or (find_first and nums[mid] == target):  # Move left for the requested boundary.
                right = mid  # Keep the left portion including mid.
            else:  # The current value is before the desired boundary.
                left = mid + 1  # Move the left boundary past mid.
        return left  # Return the boundary index.
    first = bound(True)  # Find the first position not less than target.
    if first == len(nums) or nums[first] != target:  # Check whether target appears at all.
        return [-1, -1]  # Return the conventional not-found result.
    return [first, bound(False) - 1]  # The last position is before the upper boundary.

print(search_range([5, 7, 7, 8, 8, 10], 8))  # Print [3, 4].
```

## 90. Shortest Path in an Unweighted Graph

**Pattern:** BFS while recording each node's predecessor.
- Time: `O(V + E)`.
- Space: `O(V)`.
Returns a shortest path in edge count, or `None` if unreachable.

```python
from collections import deque  # Import the queue required for BFS.

def shortest_unweighted_path(graph, start, goal):  # Find a minimum-edge path between two nodes.
    queue = deque([start])  # Begin BFS from the start node.
    parent = {start: None}  # Mark the start visited and give it no predecessor.
    while queue:  # Continue until the queue is empty or the goal is discovered.
        node = queue.popleft()  # Process the next node in BFS order.
        if node == goal:  # Check whether the goal has been reached.
            path = []  # Prepare to reconstruct the path.
            while node is not None:  # Follow predecessors back to the start.
                path.append(node)  # Add this node to the reversed path.
                node = parent[node]  # Move to the node that discovered this one.
            return path[::-1]  # Reverse the path into start-to-goal order.
        for neighbor in graph.get(node, []):  # Examine each neighbor.
            if neighbor not in parent:  # Only discover each node once.
                parent[neighbor] = node  # Record the predecessor for reconstruction.
                queue.append(neighbor)  # Schedule the neighbor for BFS.
    return None  # Return None if the goal cannot be reached.

graph = {"A": ["B", "C"], "B": ["D"], "C": ["D"], "D": []}  # Define an unweighted graph.
print(shortest_unweighted_path(graph, "A", "D"))  # Print one shortest path: A-B-D or A-C-D.
```

## 91. Kruskal Minimum Spanning Tree

**Pattern:** Sort edges by weight and use Union-Find to avoid cycles.
- Time: `O(E log E)` for sorting edges.
- Assumes an undirected weighted graph.
Returns the selected edges and total weight; a disconnected graph yields a forest.

```python
def kruskal(n, edges):  # Build a minimum spanning forest for n vertices.
    uf = UnionFind(n)  # Create disjoint sets for all vertices.
    chosen = []  # Store edges selected for the forest.
    total_weight = 0  # Accumulate selected edge weights.
    for weight, u, v in sorted(edges):  # Process edges from lowest weight to highest.
        if uf.union(u, v):  # Select the edge only if it connects different components.
            chosen.append((u, v, weight))  # Record the chosen edge.
            total_weight += weight  # Add its weight to the total.
    return chosen, total_weight  # Return the forest edges and total weight.

edges = [(1, 0, 1), (3, 1, 2), (2, 0, 2), (4, 2, 3)]  # Define (weight, endpoint, endpoint) edges.
print(kruskal(4, edges))  # Print a minimum spanning tree and its total weight.
```

## 92. Bellman-Ford Shortest Paths

**Pattern:** Relax every edge repeatedly.
- Time: `O(V × E)`.
- Supports negative edge weights and can detect reachable negative cycles.
Edges use `(source, destination, weight)` tuples.

```python
def bellman_ford(nodes, edges, start):  # Find shortest distances from start, allowing negative edges.
    distance = {node: float("inf") for node in nodes}  # Initialize every node as unreachable.
    distance[start] = 0  # Set the starting distance to zero.
    for _ in range(len(nodes) - 1):  # A shortest simple path uses at most V-1 edges.
        changed = False  # Track whether this pass improved any distance.
        for u, v, weight in edges:  # Inspect every directed edge.
            if distance[u] != float("inf") and distance[u] + weight < distance[v]:  # Check whether this edge improves v.
                distance[v] = distance[u] + weight  # Save the shorter distance.
                changed = True  # Record that the pass made progress.
        if not changed:  # Stop if a full pass makes no improvements.
            break  # Further passes cannot improve these distances.
    for u, v, weight in edges:  # Check for an additional possible relaxation.
        if distance[u] != float("inf") and distance[u] + weight < distance[v]:  # A reachable negative cycle remains.
            raise ValueError("reachable negative-weight cycle detected")  # Report that shortest paths are undefined.
    return distance  # Return all computed distances.

print(bellman_ford(["A", "B", "C"], [("A", "B", 4), ("A", "C", 5), ("B", "C", -2)], "A"))  # Print A:0, B:4, C:2.
```

## 93. Floyd-Warshall All-Pairs Shortest Paths

**Pattern:** Dynamic programming over allowed intermediate vertices.
- Time: `O(V³)`.
- Space: `O(V²)`.
Use `None` for no direct edge; negative cycles require an additional check on the diagonal.

```python
def floyd_warshall(matrix):  # Compute all-pairs shortest distances from an adjacency matrix.
    dist = [row.copy() for row in matrix]  # Copy the matrix so the input is unchanged.
    n = len(dist)  # Store the number of vertices.
    for k in range(n):  # Allow vertex k as an intermediate point.
        for i in range(n):  # Choose each source vertex.
            for j in range(n):  # Choose each destination vertex.
                if dist[i][k] is not None and dist[k][j] is not None:  # Require both path segments to exist.
                    candidate = dist[i][k] + dist[k][j]  # Calculate the path through k.
                    if dist[i][j] is None or candidate < dist[i][j]:  # Check whether this path is better.
                        dist[i][j] = candidate  # Save the shorter distance.
    return dist  # Return the all-pairs shortest-distance matrix.

matrix = [[0, 3, None], [None, 0, 2], [1, None, 0]]  # Define direct edge costs and missing edges.
print(floyd_warshall(matrix))  # Print shortest distances between all vertex pairs.
```

## 94. Prim Minimum Spanning Tree

**Pattern:** Grow a connected tree using the lightest edge leaving the current tree.
- Time: `O(E log V)` with a heap and adjacency lists.
- Assumes an undirected weighted graph; this example starts at one node.

```python
import heapq  # Import the min-heap used to choose the next lightest edge.

def prim(graph, start):  # Build a minimum spanning tree for the component containing start.
    visited = {start}  # Add the start node to the growing tree.
    heap = []  # Store candidate edges as (weight, source, destination).
    for neighbor, weight in graph.get(start, []):  # Inspect edges leaving the start.
        heapq.heappush(heap, (weight, start, neighbor))  # Add each edge to the priority queue.
    tree_edges = []  # Store the selected tree edges.
    total = 0  # Accumulate the tree's weight.
    while heap:  # Continue while candidate edges remain.
        weight, source, target = heapq.heappop(heap)  # Select the lightest candidate edge.
        if target in visited:  # Ignore edges leading to an already included node.
            continue  # Such an edge would create a cycle.
        visited.add(target)  # Add the new node to the tree.
        tree_edges.append((source, target, weight))  # Record the selected edge.
        total += weight  # Add its weight to the total.
        for neighbor, next_weight in graph.get(target, []):  # Inspect edges from the newly added node.
            if neighbor not in visited:  # Only consider edges leaving the current tree.
                heapq.heappush(heap, (next_weight, target, neighbor))  # Schedule the candidate edge.
    return tree_edges, total  # Return the spanning tree for this connected component.

graph = {"A": [("B", 1), ("C", 3)], "B": [("A", 1), ("C", 2)], "C": [("A", 3), ("B", 2)]}  # Define an undirected graph.
print(prim(graph, "A"))  # Print a minimum spanning tree with total weight 3.
```

## 95. Matrix Chain Multiplication

**Pattern:** Interval dynamic programming.
- Time: `O(n³)`.
- Space: `O(n²)`.
Find the minimum number of scalar multiplications needed to multiply a chain of matrices.

```python
def matrix_chain_cost(dimensions):  # dimensions=[p0,p1,...,pn] defines n matrices.
    n = len(dimensions) - 1  # The number of matrices is one fewer than the dimension count.
    if n <= 1:  # Zero or one matrix needs no multiplication.
        return 0  # Return zero operations.
    dp = [[0] * n for _ in range(n)]  # dp[i][j] stores the best cost for matrices i through j.
    for length in range(2, n + 1):  # Increase the length of the matrix chain being solved.
        for i in range(n - length + 1):  # Choose the starting matrix index.
            j = i + length - 1  # Calculate the ending matrix index.
            dp[i][j] = float("inf")  # Start with an upper bound on the cost.
            for split in range(i, j):  # Try every possible final split point.
                cost = dp[i][split] + dp[split + 1][j] + dimensions[i] * dimensions[split + 1] * dimensions[j + 1]  # Add subproblem costs and final multiplication.
                dp[i][j] = min(dp[i][j], cost)  # Keep the cheapest split.
    return dp[0][n - 1]  # Return the minimum cost for the complete chain.

print(matrix_chain_cost([10, 30, 5, 60]))  # Print 4500.
```

## 96. Subset Sum

**Pattern:** Boolean dynamic programming over achievable sums.
- Time: `O(n × target)`.
- Space: `O(target)`.
Assumes non-negative integers and a non-negative target.

```python
def subset_sum(nums, target):  # Check whether some subset sums exactly to target.
    if target < 0 or any(num < 0 for num in nums):  # Restrict this version to non-negative values.
        raise ValueError("values and target must be non-negative")  # Explain the supported input.
    possible = [False] * (target + 1)  # possible[s] means sum s can be formed.
    possible[0] = True  # The empty subset forms sum zero.
    for num in nums:  # Process each number once.
        for total in range(target, num - 1, -1):  # Iterate downward to avoid reusing the same number.
            possible[total] = possible[total] or possible[total - num]  # Add num to any achievable prior sum.
    return possible[target]  # Return whether the target sum is achievable.

print(subset_sum([3, 34, 4, 12, 5, 2], 9))  # Print True because 4 + 5 = 9.
```

## 97. Combination Sum

**Pattern:** Backtracking with reuse of candidates.
- Output can be exponential.
- Assumes positive candidates and a non-negative target.
The same candidate can be selected multiple times; combinations are returned in nondecreasing order.

```python
def combination_sum(candidates, target):  # Find combinations that sum to target.
    candidates = sorted(set(candidates))  # Sort candidates and remove duplicates.
    if any(value <= 0 for value in candidates) or target < 0:  # Validate positive candidates and target.
        raise ValueError("candidates must be positive and target non-negative")  # Explain constraints.
    result = []  # Store completed combinations.
    path = []  # Store the current partial combination.
    def backtrack(start, remaining):  # Search using candidates from start onward.
        if remaining == 0:  # Check whether the target has been reached exactly.
            result.append(path.copy())  # Save a copy of the completed combination.
            return  # Stop this branch.
        for i in range(start, len(candidates)):  # Try each candidate without changing combination order.
            value = candidates[i]  # Read the candidate value.
            if value > remaining:  # Sorted order means later candidates are also too large.
                break  # Stop exploring this branch.
            path.append(value)  # Include the candidate.
            backtrack(i, remaining - value)  # Reuse this candidate if needed.
            path.pop()  # Undo the choice before trying the next candidate.
    backtrack(0, target)  # Begin with all candidates available.
    return result  # Return every valid combination.

print(combination_sum([2, 3, 6, 7], 7))  # Print [[2, 2, 3], [7]].
```

## 98. Decode Ways

**Pattern:** Dynamic programming over string positions.
- Time: `O(n)`.
- Space: `O(n)` in this table-based version.
Digits map 1 through 26 to letters; strings starting with invalid zero codes have no decoding.

```python
def decode_ways(digits):  # Count ways to decode a digit string using codes 1 through 26.
    if not digits:  # Treat empty input as having no decoding in this example.
        return 0  # Return zero for an empty string.
    dp = [0] * (len(digits) + 1)  # dp[i] counts decodings of the first i digits.
    dp[0] = 1  # The empty prefix provides one base case for transitions.
    for i in range(1, len(digits) + 1):  # Process each prefix length.
        if digits[i - 1] != "0":  # A non-zero digit can be decoded by itself.
            dp[i] += dp[i - 1]  # Add the one-digit decoding possibilities.
        if i >= 2 and "10" <= digits[i - 2:i] <= "26":  # Check whether the final two digits form a valid code.
            dp[i] += dp[i - 2]  # Add the two-digit decoding possibilities.
    return dp[len(digits)]  # Return the number of ways to decode the full string.

print(decode_ways("226"))  # Print 3.
```

## 99. Min Cost Climbing Stairs

**Pattern:** Dynamic programming with rolling states.
- Time: `O(n)`.
- Space: `O(1)`.
At each step, pay the cost of the step you land on; the top is beyond the final index.

```python
def min_cost_climbing_stairs(cost):  # Find the minimum cost to reach the top.
    one_before = 0  # Minimum cost to reach the position before the starting step.
    two_before = 0  # Minimum cost to reach the position two steps before the current one.
    for i in range(2, len(cost) + 1):  # Calculate cost to reach each position through the top.
        current = min(two_before + cost[i - 2], one_before + cost[i - 1])  # Arrive from one or two steps back.
        two_before = one_before  # Shift the older state forward.
        one_before = current  # Save the best cost for this position.
    return one_before  # Return the minimum cost to reach the top.

print(min_cost_climbing_stairs([10, 15, 20]))  # Print 15.
```

## 100. Bit Counting

**Pattern:** Dynamic programming on binary representations.
- Time: `O(n)`.
- Space: `O(n)` for the output.
The number of set bits in i equals the count for `i >> 1` plus its last bit.

```python
def count_bits(n):  # Count set bits for every integer from zero through n.
    if n < 0:  # This example expects a non-negative upper bound.
        raise ValueError("n must be non-negative")  # Explain the input constraint.
    result = [0] * (n + 1)  # Allocate one count for each integer.
    for i in range(1, n + 1):  # Calculate counts in increasing order.
        result[i] = result[i >> 1] + (i & 1)  # Remove the last bit, then add that bit's value.
    return result  # Return all bit counts.

print(count_bits(5))  # Print [0, 1, 1, 2, 1, 2].
```
