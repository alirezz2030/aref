## 1. NumPy Arrays (ndarray)

NumPy's main structure is the **`ndarray`**, an efficient array that stores values of a consistent data type. It is useful for numeric analysis because operations can run over entire arrays without writing a Python loop.

**Key idea:** `np.array(...)` converts a Python sequence into an array. NumPy is commonly imported with the alias `np`.

```python
import numpy as np  # Import NumPy and use its standard short alias.

daily_sales = np.array([120, 150, 90, 175, 200])  # Convert five daily sales values into a NumPy array.
print(daily_sales)  # Display the complete array.
print(type(daily_sales))  # Show the object's Python type.
print(daily_sales + 10)  # Add 10 to every element without writing a loop.
```

## 2. Creating Arrays and Choosing Data Types

NumPy provides tools for building arrays: `zeros`, `ones`, `full`, `arange`, and `linspace`. A numeric array usually stores one data type, such as integers or floating-point numbers.

- `arange(start, stop, step)` uses a step size and excludes `stop`.
- `linspace(start, stop, num)` creates a chosen number of evenly spaced values and includes the endpoints by default.
- Use an explicit `dtype` when a specific numeric representation is needed.

```python
zeros = np.zeros(4)  # Create four floating-point zeros.
ones = np.ones(3, dtype=int)  # Create three integer ones.
filled = np.full(4, 7.5)  # Fill an array with the value 7.5.
sequence = np.arange(0, 10, 2)  # Create 0, 2, 4, 6, and 8.
even_points = np.linspace(0, 1, 5)  # Create five equally spaced values from 0 to 1.
print("zeros:", zeros)  # Display the zero array.
print("ones:", ones)  # Display the integer array.
print("filled:", filled)  # Display the constant array.
print("arange:", sequence)  # Display the stepped sequence.
print("linspace:", even_points)  # Display the evenly spaced values.
```

## 3. Shape, Dimensions, Size, and Data Type

Before analyzing an array, inspect its structure.

- `shape` gives the size of each dimension.
- `ndim` gives the number of dimensions.
- `size` counts all elements.
- `dtype` tells you the stored data type.

These checks are especially useful when preparing a feature matrix for statistics or machine learning.

```python
sales_table = np.array([[120, 3], [150, 4], [90, 2]])  # Create a 2D array with sales and units columns.
print("Array:\n", sales_table)  # Display the two-dimensional array.
print("Shape:", sales_table.shape)  # Show its row and column counts.
print("Dimensions:", sales_table.ndim)  # Show that this is a two-dimensional array.
print("Total elements:", sales_table.size)  # Count all stored values.
print("Data type:", sales_table.dtype)  # Display the numeric type used by the array.
```

## 4. Indexing and Slicing

Indexing gets a specific value; slicing gets a range of values. Python indexing starts at zero, and the end position of a slice is excluded.

For a two-dimensional array, use `array[row, column]`. A colon (`:`) means “select all” along that axis.

```python
temperatures = np.array([18, 20, 21, 19, 23, 25, 24])  # Store seven daily temperatures.
print("First day:", temperatures[0])  # Read the first element.
print("Last day:", temperatures[-1])  # Read the final element using negative indexing.
print("First three:", temperatures[:3])  # Select indices 0, 1, and 2.
print("Every second value:", temperatures[::2])  # Select every other value.
matrix = np.array([[10, 20, 30], [40, 50, 60]])  # Create a two-row, three-column array.
print("Row 2, column 3:", matrix[1, 2])  # Read the element at zero-based position [1, 2].
print("First column:", matrix[:, 0])  # Select every row from the first column.
```

## 5. Reshaping and Flattening

`reshape()` changes an array's dimensions without changing its element order. The new shape must have the same total number of elements.

- `flatten()` returns a copy as a one-dimensional array.
- `ravel()` returns a one-dimensional view when possible.
- `reshape(-1, 1)` is commonly used to turn one feature into a column vector.

```python
values = np.arange(1, 13)  # Create twelve sequential values from 1 through 12.
grid = values.reshape(3, 4)  # Arrange the twelve values into three rows and four columns.
column = values.reshape(-1, 1)  # Convert the values into a 12-by-1 column vector.
flat_copy = grid.flatten()  # Return a flattened copy of the grid.
print("3 x 4 grid:\n", grid)  # Display the reshaped matrix.
print("Column shape:", column.shape)  # Show the column vector's dimensions.
print("Flattened:", flat_copy)  # Display the one-dimensional copy.
```

## 6. Vectorized Arithmetic

Vectorization means applying operations to whole arrays instead of writing explicit loops. When shapes match, NumPy performs element-by-element arithmetic efficiently.

Operations like `+`, `-`, `*`, `/`, `**`, and comparisons work element-wise. This is one of the main reasons NumPy is useful for numeric analysis.

```python
price = np.array([10.0, 15.0, 8.0, 20.0])  # Store the prices for four products.
quantity = np.array([3, 2, 5, 1])  # Store the quantity sold for each product.
revenue = price * quantity  # Multiply corresponding prices and quantities.
discounted_price = price * 0.9  # Apply a 10 percent discount to every price.
price_difference = price - 12  # Subtract 12 from every price.
is_expensive = price >= 15  # Create a Boolean array for prices of at least 15.
print("Revenue per product:", revenue)  # Display revenue values.
print("Discounted prices:", discounted_price)  # Display discounted prices.
print("Difference from 12:", price_difference)  # Display how each price differs from 12.
print("Price >= 15:", is_expensive)  # Display the Boolean condition results.
```

## 7. Broadcasting

**Broadcasting** lets NumPy combine arrays of compatible shapes without manually copying values. For example, a row of product prices can be applied to every row of a sales matrix.

Dimensions are compatible when they are equal or one of them is `1`. Broadcasting is powerful, but always inspect shapes so an unintended operation does not silently produce a large result.

```python
monthly_units = np.array([[2, 3, 1], [4, 1, 2], [3, 2, 5]])  # Store units sold across three months and three products.
unit_prices = np.array([100, 25, 60])  # Store one price per product.
monthly_revenue = monthly_units * unit_prices  # Broadcast each product price across the month rows.
print("Units shape:", monthly_units.shape)  # Show the sales matrix shape.
print("Prices shape:", unit_prices.shape)  # Show the prices vector shape.
print("Revenue by month and product:\n", monthly_revenue)  # Display the element-wise revenue matrix.
print("Monthly totals:", monthly_revenue.sum(axis=1))  # Add product revenue across each row.
print("Product totals:", monthly_revenue.sum(axis=0))  # Add monthly revenue down each column.
```

## 8. Boolean Filtering

A Boolean mask selects values that meet a condition. Create the mask with comparisons such as `>`, `<=`, or `==`, then apply it to the array.

Use `&` for AND and `|` for OR between NumPy conditions. Put each condition in parentheses; Python's `and` and `or` are not suitable for element-wise array conditions.

```python
orders = np.array([45, 120, 80, 230, 95, 150])  # Store six order values.
large_mask = orders >= 100  # Mark orders with a value of at least 100.
large_orders = orders[large_mask]  # Keep only values whose mask is True.
middle_orders = orders[(orders >= 80) & (orders <= 150)]  # Select orders between 80 and 150 inclusive.
print("Mask:", large_mask)  # Display which orders satisfy the first condition.
print("Large orders:", large_orders)  # Display orders valued at least 100.
print("Middle-range orders:", middle_orders)  # Display orders in the requested range.
print("Count of large orders:", np.sum(large_mask))  # Count True values in the Boolean mask.
```

## 9. Conditional Values with np.where

`np.where(condition, value_if_true, value_if_false)` applies an if/else choice element-wise. It is useful for creating flags, categories, or simple business rules.

For multiple categories, nested `where` expressions can become hard to read; consider `np.select()` or clearer step-by-step logic in a real project.

```python
scores = np.array([42, 67, 88, 51, 93, 74])  # Store six assessment scores.
passed = np.where(scores >= 60, "Pass", "Fail")  # Label scores of 60 or higher as Pass.
performance = np.where(scores >= 85, "High", np.where(scores >= 60, "Medium", "Low"))  # Assign three performance bands.
bonus = np.where(scores >= 80, 100, 0)  # Assign a 100-unit bonus to scores of at least 80.
print("Pass status:", passed)  # Display pass/fail labels.
print("Performance band:", performance)  # Display high, medium, or low labels.
print("Bonus:", bonus)  # Display the bonus assigned to each score.
```

## 10. Aggregations and the axis Parameter

Aggregations summarize values using functions such as `sum`, `mean`, `median`, `min`, `max`, and `std`.

For a 2D array:
- `axis=0` combines values down rows, producing one result per column.
- `axis=1` combines values across columns, producing one result per row.
- Omitting `axis` aggregates all elements.

```python
weekly_sales = np.array([[100, 120, 90], [80, 110, 130], [140, 100, 150]])  # Store three weeks of sales across three products.
print("All-sales total:", weekly_sales.sum())  # Add every value in the entire matrix.
print("Total per product:", weekly_sales.sum(axis=0))  # Sum down rows to get each product's total.
print("Total per week:", weekly_sales.sum(axis=1))  # Sum across columns to get each week's total.
print("Average per product:", weekly_sales.mean(axis=0).round(2))  # Calculate each product's mean weekly sales.
print("Highest value:", weekly_sales.max())  # Find the largest single entry.
print("Median per week:", np.median(weekly_sales, axis=1))  # Calculate each week's median across products.
```

## 11. Missing Numeric Values with NaN

NumPy uses `np.nan` to represent a missing floating-point number. Standard functions like `np.mean()` often return NaN if a value is missing, while `np.nanmean()` ignores missing values.

Do not fill missing values blindly. Decide whether a missing value means unknown, not applicable, zero, or a collection error. `np.nan` is mainly used with floating-point arrays.

```python
readings = np.array([12.0, 15.0, np.nan, 18.0, np.nan, 20.0])  # Store numeric readings with two missing values.
missing_mask = np.isnan(readings)  # Identify positions containing NaN.
valid_readings = readings[~missing_mask]  # Keep only values that are not missing.
mean_ignoring_missing = np.nanmean(readings)  # Calculate the mean using available values.
filled_readings = np.where(missing_mask, mean_ignoring_missing, readings)  # Fill missing positions with the observed mean for demonstration.
print("Missing mask:", missing_mask)  # Display the locations of missing readings.
print("Valid readings:", valid_readings)  # Display only available readings.
print("Mean ignoring NaN:", mean_ignoring_missing)  # Display the mean computed from available values.
print("Mean-filled example:", filled_readings)  # Display the illustrative imputed array.
```

## 12. Sorting and Ranking

`np.sort()` returns sorted values, while `np.argsort()` returns the indices that would sort the array. Index-based sorting is useful when you need to retrieve associated records or features.

For descending order, reverse the sorted array or index sequence. `argsort()` is often helpful for finding a ranked list of product or customer values.

```python
product_names = np.array(["Laptop", "Tablet", "Phone", "Monitor"])  # Store product labels.
product_revenue = np.array([5200, 2100, 4800, 1700])  # Store revenue associated with each product.
sorted_revenue = np.sort(product_revenue)  # Return revenue values from low to high.
rank_indices = np.argsort(product_revenue)[::-1]  # Get indices ordered by revenue from high to low.
ranked_names = product_names[rank_indices]  # Apply the ranking indices to product names.
ranked_revenue = product_revenue[rank_indices]  # Apply the same indices to revenue values.
print("Sorted revenue ascending:", sorted_revenue)  # Display values in ascending order.
print("Products ranked high to low:", ranked_names)  # Display the product ranking.
print("Ranked revenue:", ranked_revenue)  # Display revenue in matching order.
```

## 13. Unique Values and Counts

`np.unique()` returns the distinct values in an array. With `return_counts=True`, it also returns how often each value appears.

This is handy for checking category codes, discovering repeated labels, or summarizing the frequency of discrete numeric values. For larger analytics workflows, pandas has additional categorical tools, but NumPy's method is useful for arrays.

```python
regions = np.array(["North", "South", "North", "East", "West", "South", "North"])  # Store region labels for records.
unique_regions, counts = np.unique(regions, return_counts=True)  # Find each distinct label and its frequency.
print("Unique regions:", unique_regions)  # Display category names in sorted order.
print("Counts:", counts)  # Display the frequency of each category.
print("Frequency map:", dict(zip(unique_regions.tolist(), counts.tolist())))  # Combine category labels and counts into a dictionary.
```

## 14. Percentiles, Quartiles, and IQR

Percentiles describe the position of values in a distribution. The 25th percentile is the first quartile (Q1); the 75th percentile is the third quartile (Q3).

The **interquartile range** is `IQR = Q3 − Q1`. It describes the spread of the middle 50% of values and is less sensitive to extremes than the range or standard deviation.

```python
delivery_days = np.array([2, 3, 3, 4, 4, 5, 6, 7, 12, 18])  # Store delivery times in days.
q1, median, q3 = np.percentile(delivery_days, [25, 50, 75])  # Calculate the first quartile, median, and third quartile.
iqr = q3 - q1  # Calculate the interquartile range.
p90 = np.percentile(delivery_days, 90)  # Calculate the 90th percentile.
print("Q1:", q1)  # Display the 25th percentile.
print("Median:", median)  # Display the 50th percentile.
print("Q3:", q3)  # Display the 75th percentile.
print("IQR:", iqr)  # Display the middle-50-percent spread.
print("90th percentile:", p90)  # Display the value below which about 90 percent of observations fall.
```

## 15. Detecting Potential Outliers with IQR

A common rule flags values below `Q1 − 1.5 × IQR` or above `Q3 + 1.5 × IQR`. This is a **screening rule**, not proof that values are errors.

An outlier might be a valid rare event, a data-entry mistake, or a different kind of customer. Investigate unusual records using business context before removing or changing them.

```python
transaction_values = np.array([20, 22, 25, 28, 30, 31, 33, 35, 38, 220])  # Create sample transaction values, including one extreme value.
q1, q3 = np.percentile(transaction_values, [25, 75])  # Calculate the first and third quartiles.
iqr = q3 - q1  # Calculate the interquartile range.
lower_bound = q1 - 1.5 * iqr  # Calculate the lower outlier-screening threshold.
upper_bound = q3 + 1.5 * iqr  # Calculate the upper outlier-screening threshold.
outlier_mask = (transaction_values < lower_bound) | (transaction_values > upper_bound)  # Identify values outside the thresholds.
print("Lower bound:", lower_bound)  # Display the lower screening threshold.
print("Upper bound:", upper_bound)  # Display the upper screening threshold.
print("Potential outliers:", transaction_values[outlier_mask])  # Display flagged values for review.
print("Original values unchanged:", transaction_values)  # Confirm that screening did not modify the data.
```

## 16. Min-Max Normalization

Min-max normalization rescales values to a chosen interval, commonly 0 to 1, using:

`(x − min) / (max − min)`

This is useful when features need comparable ranges. A constant-valued array has `max == min`, so the denominator is zero; handle that case explicitly.

```python
customer_spend = np.array([100.0, 250.0, 400.0, 550.0])  # Store four customer spending values.
minimum = customer_spend.min()  # Find the smallest spending value.
maximum = customer_spend.max()  # Find the largest spending value.
if maximum == minimum:  # Check whether every value is the same.
    normalized = np.zeros_like(customer_spend)  # Avoid dividing by zero for a constant array.
else:  # Handle the usual case where the values vary.
    normalized = (customer_spend - minimum) / (maximum - minimum)  # Scale the values to the 0-to-1 interval.
print("Original:", customer_spend)  # Display the original values.
print("Normalized:", normalized.round(3))  # Display the normalized values.
```

## 17. Standardization with Z-Scores

A z-score measures how many standard deviations a value is above or below the mean:

`z = (x − mean) / standard_deviation`

Standardization creates values with mean near 0 and standard deviation near 1, using the same standard-deviation convention in both steps. It is often useful before distance-based methods. It is sensitive to outliers, and in machine learning the mean and standard deviation should be calculated from training data only.

```python
exam_scores = np.array([55.0, 60.0, 65.0, 70.0, 90.0])  # Store five exam scores.
mean_score = exam_scores.mean()  # Calculate the average score.
std_score = exam_scores.std()  # Calculate the population standard deviation using NumPy's default ddof=0.
if std_score == 0:  # Check whether there is any variation.
    z_scores = np.zeros_like(exam_scores)  # Use zeros when every score is the same.
else:  # Handle the usual case.
    z_scores = (exam_scores - mean_score) / std_score  # Convert each score to its z-score.
print("Mean:", round(mean_score, 2))  # Display the mean.
print("Population standard deviation:", round(std_score, 2))  # Display the standard deviation used.
print("Z-scores:", z_scores.round(2))  # Display standardized values.
```

## 18. Cumulative Sums and Differences

Cumulative functions reveal how a quantity builds up over time.

- `np.cumsum()` computes a running total.
- `np.diff()` computes differences between neighboring values.
- Differences can reveal daily changes, but the first value is lost because there is no previous value to compare against.

```python
daily_revenue = np.array([100, 120, 90, 150, 180, 170])  # Store revenue by day in chronological order.
running_total = np.cumsum(daily_revenue)  # Calculate cumulative revenue through each day.
daily_change = np.diff(daily_revenue)  # Calculate each day's change from the previous day.
print("Daily revenue:", daily_revenue)  # Display the original time series.
print("Running total:", running_total)  # Display cumulative revenue.
print("Day-to-day changes:", daily_change)  # Display absolute changes between consecutive days.
print("Total revenue:", running_total[-1])  # Read the final cumulative value as the total.
```

## 19. Growth Rates and Percentage Changes

Percentage change measures relative movement between consecutive values:

`(current − previous) / previous × 100`

It is often used for sales, traffic, and financial time series. A previous value of zero makes the percentage change undefined, so handle such cases rather than assuming every denominator is valid.

```python
monthly_revenue = np.array([1000.0, 1100.0, 990.0, 1200.0, 1320.0])  # Store revenue for five sequential months.
absolute_change = np.diff(monthly_revenue)  # Calculate absolute change from one month to the next.
previous = monthly_revenue[:-1]  # Select every month except the final one as the denominator.
current = monthly_revenue[1:]  # Select every month except the first one as the numerator's current value.
growth_rate = np.divide(current - previous, previous, out=np.full_like(current, np.nan), where=previous != 0) * 100  # Calculate percent change while avoiding division by zero.
print("Absolute monthly changes:", absolute_change)  # Display numeric differences.
print("Monthly growth (%):", growth_rate.round(2))  # Display percentage changes rounded to two decimals.
```

## 20. Moving Average

A moving average smooths short-term fluctuations by averaging a fixed-size window of consecutive values. It is often used to reveal a trend in noisy time-series data.

This version uses `np.convolve()` with a uniform window. The result is shorter than the original series when using `mode="valid"` because it only reports complete windows.

```python
daily_orders = np.array([12, 15, 13, 20, 18, 22, 19, 25, 23, 28])  # Store daily order counts.
window_size = 3  # Choose a window containing three days.
weights = np.ones(window_size) / window_size  # Create equal weights that add up to one.
moving_average = np.convolve(daily_orders, weights, mode="valid")  # Calculate the average of each complete three-day window.
print("Daily orders:", daily_orders)  # Display the raw time series.
print("3-day moving average:", moving_average.round(2))  # Display the smoothed series.
```

## 21. Correlation and Covariance

**Covariance** indicates whether two variables tend to move together; its magnitude depends on their units. **Correlation** standardizes this relationship, usually between -1 and 1.

A strong correlation does not establish causation. It can also be misleading with small samples, outliers, or non-linear relationships. For `np.corrcoef()`, rows are treated as variables by default when `rowvar=True`; this example transposes the two-column data so each variable is a row.

```python
advertising_spend = np.array([10, 12, 15, 18, 20, 25], dtype=float)  # Store advertising spend for six periods.
sales = np.array([100, 105, 130, 135, 150, 180], dtype=float)  # Store sales measured in the same periods.
correlation_matrix = np.corrcoef(advertising_spend, sales)  # Calculate the 2-by-2 Pearson correlation matrix.
covariance_matrix = np.cov(advertising_spend, sales)  # Calculate the sample covariance matrix.
correlation = correlation_matrix[0, 1]  # Read the relationship between spend and sales from the matrix.
covariance = covariance_matrix[0, 1]  # Read the covariance between the two variables.
print("Correlation matrix:\n", correlation_matrix.round(3))  # Display standardized associations.
print("Sample covariance matrix:\n", covariance_matrix.round(2))  # Display covariance in the original units.
print("Spend-sales correlation:", round(correlation, 3))  # Display the single correlation coefficient.
print("Interpret carefully: correlation does not prove causation.")  # Remind the analyst not to infer causality from association alone.
```

## 22. Reproducible Random Sampling

Random sampling can help inspect a subset of records or build simulations. `np.random.default_rng(seed)` creates a modern random-number generator; a fixed seed makes the example reproducible.

For records with associated fields, sample indices first and use those indices to select all corresponding arrays so fields stay aligned.

```python
rng = np.random.default_rng(42)  # Create a seeded random generator for reproducible output.
customer_ids = np.arange(1001, 1021)  # Create 20 customer IDs.
spend = rng.integers(20, 500, size=len(customer_ids))  # Generate one illustrative spend value per customer.
sample_indices = rng.choice(len(customer_ids), size=5, replace=False)  # Select five distinct row indices.
sampled_ids = customer_ids[sample_indices]  # Select matching customer IDs.
sampled_spend = spend[sample_indices]  # Select spend values using the same indices.
print("Sampled customer IDs:", sampled_ids)  # Display the sampled IDs.
print("Their spend values:", sampled_spend)  # Display the corresponding spend values.
print("Aligned records:\n", np.column_stack((sampled_ids, sampled_spend)))  # Combine the sampled fields into two columns.
```

## 23. Simulation with Random Numbers

Simulation generates many possible outcomes to estimate a quantity when an exact answer is difficult or unnecessary. A Monte Carlo estimate uses random draws and summarizes their outcomes.

The example estimates the probability of rolling at least 5 on a fair six-sided die. More simulations generally reduce random variation, but they do not eliminate the need for sound assumptions.

```python
rng = np.random.default_rng(7)  # Create a reproducible random generator for the simulation.
number_of_rolls = 100_000  # Choose how many simulated dice rolls to generate.
rolls = rng.integers(1, 7, size=number_of_rolls)  # Simulate fair six-sided die outcomes from 1 through 6.
successful_rolls = np.sum(rolls >= 5)  # Count rolls that resulted in 5 or 6.
estimated_probability = successful_rolls / number_of_rolls  # Estimate the success probability using the observed fraction.
theoretical_probability = 2 / 6  # Calculate the exact probability of rolling either 5 or 6.
print("Estimated probability:", round(estimated_probability, 4))  # Display the Monte Carlo estimate.
print("Theoretical probability:", round(theoretical_probability, 4))  # Display the known exact probability for comparison.
```

## 24. Histograms and Binning

A histogram groups numeric values into intervals (bins) and counts how many values fall into each interval. It helps analyze distributions, including skew and concentration.

The chosen bin edges affect the appearance of a histogram. For meaningful comparisons, use consistent bins across groups. `np.histogram()` returns the count per bin and the bin-edge array.

```python
rng = np.random.default_rng(21)  # Create a random generator for reproducible sample data.
response_minutes = rng.gamma(shape=2.0, scale=4.0, size=500)  # Simulate a positively skewed response-time distribution.
counts, bin_edges = np.histogram(response_minutes, bins=8)  # Divide the values into eight bins and count observations.
print("Counts per bin:", counts)  # Display the number of observations in each bin.
print("Bin edges:", bin_edges.round(2))  # Display the numeric boundaries of the bins.
print("Total observations counted:", counts.sum())  # Confirm that all in-range observations were counted.
```

## 25. Top-K Values and Their Indices

A common analytics task is to find the top few values without sorting or displaying the entire dataset. `np.argpartition()` separates the k largest elements efficiently, though the selected elements are not internally sorted.

Sort only the selected subset if you need the final top-k values in rank order. Take care with ties: more than k records may legitimately tie at the cutoff.

```python
rng = np.random.default_rng(5)  # Create a reproducible random generator.
scores = rng.integers(0, 1000, size=20)  # Generate 20 illustrative scores.
k = 5  # Choose how many top scores to find.
candidate_indices = np.argpartition(scores, -k)[-k:]  # Select indices of the five largest values without fully sorting all values.
ordered_indices = candidate_indices[np.argsort(scores[candidate_indices])[::-1]]  # Sort only the selected indices by descending score.
print("All scores:", scores)  # Display the original scores.
print("Top scores:", scores[ordered_indices])  # Display the largest scores in descending order.
print("Top-score indices:", ordered_indices)  # Display their positions in the original array.
```

## 26. Combining and Splitting Arrays

`concatenate()` joins arrays along an existing axis. `vstack()` adds rows, and `hstack()` joins columns for compatible shapes. `split()` separates an array into parts.

Before combining arrays, confirm that their dimensions match on all axes except the axis being joined. This is useful when combining batches or separating feature groups.

```python
january = np.array([[10, 20], [30, 40]])  # Store two rows of two measurements.
february = np.array([[50, 60], [70, 80]])  # Store another two rows with the same column count.
combined_rows = np.concatenate((january, february), axis=0)  # Join the arrays by adding rows.
combined_columns = np.concatenate((january, february), axis=1)  # Join the arrays by adding columns.
first_half, second_half = np.split(combined_rows, 2, axis=0)  # Split the combined rows into two equal parts.
print("Combined by rows:\n", combined_rows)  # Display the vertically combined data.
print("Combined by columns:\n", combined_columns)  # Display the horizontally combined data.
print("First split:\n", first_half)  # Display the first half.
print("Second split:\n", second_half)  # Display the second half.
```

## 27. Weighted Average

A weighted average gives each value a specified level of influence. It is useful for metrics such as average price when different products have different quantities sold.

The weights should match the values in shape and, for common weighted means, have a nonzero total. Always ensure that the weighting scheme matches the business definition.

```python
unit_prices = np.array([10.0, 20.0, 50.0])  # Store three product prices.
units_sold = np.array([100, 50, 10])  # Store the quantity sold at each price.
simple_average = unit_prices.mean()  # Give all products equal weight.
weighted_average = np.average(unit_prices, weights=units_sold)  # Weight each price by the number of units sold.
total_revenue = np.sum(unit_prices * units_sold)  # Calculate total revenue across products.
total_units = units_sold.sum()  # Calculate total quantity sold.
print("Unweighted average price:", round(simple_average, 2))  # Display the simple mean of the product prices.
print("Quantity-weighted average price:", round(weighted_average, 2))  # Display average price paid per unit.
print("Revenue / units:", round(total_revenue / total_units, 2))  # Confirm the weighted average by the ratio of totals.
```

## 28. Matrix Multiplication and Dot Products

The dot product multiplies corresponding elements and adds the products. Matrix multiplication combines rows of one matrix with columns of another.

`@` is Python's matrix-multiplication operator for NumPy arrays. Shapes must align: if A has shape `(m, n)`, B must have shape `(n, p)`, and the result has shape `(m, p)`.

```python
features = np.array([[1, 10], [1, 20], [1, 30]])  # Create a feature matrix with a leading intercept column.
coefficients = np.array([5, 2])  # Store a simple intercept and slope.
predicted_values = features @ coefficients  # Multiply the feature matrix by the coefficient vector.
simple_dot = np.dot(np.array([2, 3, 4]), np.array([10, 20, 30]))  # Compute a vector dot product.
print("Feature matrix shape:", features.shape)  # Display the design matrix dimensions.
print("Predicted values:", predicted_values)  # Display the linear predictions.
print("Dot product:", simple_dot)  # Display the sum of pairwise products.
```

## 29. Simple Linear Fit with NumPy

`np.polyfit(x, y, deg=1)` estimates the slope and intercept of a straight line by least squares. It is a convenient introductory tool for examining a linear trend.

A fitted relationship is not automatically causal or reliable for extrapolation. For production modeling, use a train/test split and evaluate errors on data not used to fit the model.

```python
months = np.array([1, 2, 3, 4, 5, 6], dtype=float)  # Represent six months as a numeric time index.
sales = np.array([100, 115, 125, 140, 150, 165], dtype=float)  # Store observed sales across those months.
slope, intercept = np.polyfit(months, sales, deg=1)  # Fit a least-squares straight line to the observations.
fitted_sales = slope * months + intercept  # Calculate the fitted values from the estimated line.
residuals = sales - fitted_sales  # Calculate observed minus fitted values.
mae = np.mean(np.abs(residuals))  # Calculate mean absolute fitting error for this illustrative dataset.
print("Estimated slope:", round(slope, 2))  # Display the average change per month according to the line.
print("Estimated intercept:", round(intercept, 2))  # Display the fitted value at month zero.
print("Fitted values:", fitted_sales.round(2))  # Display predictions from the fitted line.
print("In-sample MAE:", round(mae, 2))  # Display the mean absolute error on the same data used to fit the line.
```
