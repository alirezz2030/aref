# Job Preparation for Data Analysts

A practical career-preparation notebook for junior data analyst and business intelligence roles. It covers role targeting, resume and LinkedIn, portfolio presentation, job applications, interviews, business cases, and professional communication.

Each lesson has two cells: a Markdown explanation and a Python example with `#` comments. Some lessons create reusable text templates or checklists; the notebook is a working preparation workbook, not a job guarantee.

## Before you start
- Customize every template with your real skills and achievements.
- Never invent work experience, business impact, certifications, or project results.
- Use portfolio projects as independent projects and disclose synthetic practice data.
- Prepare to explain each line of SQL/Python, every KPI definition, and the limitations of your conclusions.
- Apply for roles when you can demonstrate the core requirements; do not wait until every optional skill is mastered.

**Requirements:** Python and pandas. The A/B test example also uses SciPy. In Anaconda, these are usually available; if needed, run `%pip install pandas numpy scipy matplotlib` in a notebook cell.


## 1. Choose a Specific Target Role

“Data Analyst” covers several job types. Some roles emphasize SQL and dashboards; others emphasize product analytics, finance, operations, or experimentation. Pick one primary role and a secondary alternative so your resume and portfolio can be focused.

Read several real job descriptions and identify repeated responsibilities, tools, and domain terms. Do not claim a skill just because it appears in a job ad; mark it as strong, working knowledge, learning, or not yet learned.

```python
target_role = "Junior Data Analyst"  # Choose the main role you are targeting.
secondary_role = "Business Intelligence Analyst"  # Choose a related alternative role.
role_focus = {  # Record what this job family commonly needs.
    "core_tools": ["SQL", "Excel", "Python/pandas", "Power BI"],  # List tools to demonstrate with evidence.
    "core_work": ["clean data", "define KPIs", "analyze trends", "communicate recommendations"],  # List work outputs rather than only tool names.
    "evidence": ["portfolio projects", "clear README files", "well-defined metrics", "business recommendations"],  # List evidence an employer can inspect.
}  # Finish the role profile.
print("Primary target:", target_role)  # Display the primary role.
print("Secondary target:", secondary_role)  # Display the related role.
for category, items in role_focus.items():  # Iterate through role requirements.
    print(f"\n{category.replace('_', ' ').title()}:")  # Print each category label.
    for item in items:  # Iterate through each item.
        print(" -", item)  # Print each requirement or evidence type.
```

## 3. Resume Structure for a Junior Data Analyst

A clear one-page resume is often enough for an early-career analyst, though length depends on relevant experience. Put the most relevant evidence near the top: headline/summary, skills, selected projects, experience, and education or relevant training.

Use simple formatting, consistent dates, recognizable section headings, and searchable text. Avoid skill bars, decorative charts, complicated columns, and graphics that can be difficult for resume systems to parse.

```python
resume_outline = [  # Store a practical resume order.
    "Name and professional contact details",  # Make contact information visible and professional.
    "Headline: Junior Data Analyst | SQL | Python | Power BI",  # State the target role and verified core tools.
    "Summary: 2-3 lines focused on analysis, tools, and business value",  # Avoid vague claims such as hardworking expert.
    "Technical skills grouped by category",  # Group SQL, Python, BI, spreadsheets, and methods.
    "Selected analytical projects with links and measurable outputs",  # Put strong evidence where it can be found quickly.
    "Relevant work experience with outcomes and transferable skills",  # Include only truthful responsibilities and accomplishments.
    "Education, relevant courses, or certifications",  # Include the highest completed education and relevant training.
]  # Finish the resume outline.
for position, section in enumerate(resume_outline, start=1):  # Number each recommended section.
    print(f"{position}. {section}")  # Display the resume structure.
```

## 4. Write Evidence-Based Resume Bullets

A useful bullet often follows **action + analysis/tool + scope + result**. Use real, verifiable results. If you do not know the business impact, describe the deliverable and validation honestly instead of inventing a percentage.

For portfolio work, label it as a project—not employment. A clear project bullet can still demonstrate practical skills when it names the question, method, and output.

```python
bullet_formula = {  # Define a formula for a strong accomplishment bullet.
    "action": "Analyzed",  # Start with a precise action verb.
    "method": "sales and margin trends using SQL and pandas",  # State how the work was performed.
    "scope": "across four regions and three sales channels",  # State the analysis scope.
    "result": "identified categories for margin investigation and summarized findings in a dashboard",  # Describe a defensible output without inventing savings.
}  # Finish the bullet components.
example_bullet = (  # Combine the components into one resume-style sentence.
    f'{bullet_formula["action"]} {bullet_formula["method"]} '  # Add action and tools.
    f'{bullet_formula["scope"]}; {bullet_formula["result"]}.'  # Add scope and outcome.
)  # Finish the example sentence.
print(example_bullet)  # Display the example bullet.
print("\nOnly add numeric impact when you measured it and can explain the calculation.")  # Reinforce truthfulness.
```

## 5. Tailor a Resume to a Job Description

Tailoring means emphasizing the most relevant truthful evidence, not copying a job ad or stuffing keywords. Match your skills and project bullets to the employer's stated responsibilities and tools when you can substantiate the match.

Keep a master resume and create a targeted version for each role family. Reorder bullets, clarify relevant tools, and use the employer's terminology where it accurately describes your work.

```python
job_description_terms = ["SQL", "Power BI", "KPI reporting", "data quality", "stakeholder communication"]  # Example requirements extracted from a job ad.
resume_evidence = {  # Map each term to truthful evidence.
    "SQL": "SQL project: joins, GROUP BY, CTEs, and window functions",  # Name where SQL is demonstrated.
    "Power BI": "Portfolio dashboard with a star schema and DAX measures",  # Name a dashboard artifact.
    "KPI reporting": "Sales project: revenue, gross profit, gross margin, and order value",  # Name KPI evidence.
    "data quality": "Notebook: missing-value, duplicate, and key-integrity checks",  # Name data-quality evidence.
    "stakeholder communication": "One-page executive summary with findings and recommendations",  # Name communication evidence.
}  # Finish the mapping.
for term in job_description_terms:  # Review each requirement.
    evidence = resume_evidence.get(term, "No clear evidence yet; build a small proof task.")  # Retrieve evidence or flag a gap.
    print(f"{term}: {evidence}")  # Show how each keyword is supported.
print("\nDo not list tools or achievements that you cannot explain in an interview.")  # State the integrity rule.
```

Do not publish a notebook alone without context. Include a concise README, clear screenshots, a small data dictionary, SQL where relevant, and a statement that practice data are synthetic.

## 10. Prioritize Which Jobs to Apply For

Apply with focus rather than sending the same application to every role. A simple score can consider match to core skills, relevance of responsibilities, level fit, location/work eligibility, and interest.

The score is a triage tool, not an objective measure of hiring probability. Apply when you meet the central requirements even if you do not match every preferred qualification.

```python
job_options = pd.DataFrame({  # Create illustrative opportunities for prioritization.
    "role": ["Junior Data Analyst", "Senior BI Analyst", "Operations Reporting Analyst"],  # Name possible roles.
    "skill_match_1_to_5": [4, 2, 4],  # Rate how closely your demonstrated skills match.
    "level_fit_1_to_5": [5, 1, 4],  # Rate the seniority fit.
    "interest_1_to_5": [5, 3, 4],  # Rate personal interest.
    "practical_fit_1_to_5": [4, 3, 4],  # Rate location, schedule, and eligibility fit.
})  # Finish the role list.
job_options["priority_score"] = job_options[["skill_match_1_to_5", "level_fit_1_to_5", "interest_1_to_5", "practical_fit_1_to_5"]].mean(axis=1)  # Calculate an unweighted mean score.
print(job_options.sort_values("priority_score", ascending=False).round(2).to_string(index=False))  # Rank roles to help decide where to spend customization time.
```

## 13. Use STAR for Behavioral Questions

STAR stands for **Situation, Task, Action, Result**. It structures a real story so the interviewer can understand the context, your responsibility, your specific actions, and the outcome.

For a portfolio project, be clear that it was a self-directed or practice project. The result can be an insight or deliverable; it does not need a fabricated financial impact.

```python
star_story = {  # Draft a STAR answer based on a portfolio project.
    "Situation": "A practice sales dataset showed that revenue alone did not explain profitability.",  # Describe the context.
    "Task": "Identify which categories and channels deserved further investigation.",  # State the objective.
    "Action": "Cleaned the transaction data, checked order grain, calculated gross margin, and compared groups in SQL and pandas.",  # Explain what you personally did.
    "Result": "Produced a dashboard-ready KPI table and recommendations for reviewing discounting and product mix.",  # State the defensible result.
}  # Finish the STAR answer.
for stage, text in star_story.items():  # Display each STAR section.
    print(f"{stage}: {text}")  # Print the response section.
print("\nAdd real details from your own work and clearly label practice data as synthetic.")  # Remind the user to personalize the story.
```

## 17. Statistics and A/B Test Questions

Interviewers may ask you to interpret distributions, confidence intervals, p-values, correlation, sampling bias, and experiment results. The key skill is explaining what the numbers do and do not establish.

A p-value is not the probability the null hypothesis is true. A non-significant result is not proof of no effect. State the effect size, uncertainty, assumptions, and business relevance.

```python
statistics_questions = [  # Prepare concise statistical interview questions.
    "When would you report median instead of mean?",  # Check robust descriptive statistics.
    "What is the difference between standard deviation and standard error?",  # Check the concept of sampling uncertainty.
    "How would you explain a 95% confidence interval?",  # Check interval interpretation.
    "What does a p-value mean, and what does it not mean?",  # Check hypothesis-test interpretation.
    "Why does correlation not establish causation?",  # Check causal reasoning.
    "How can selection bias affect an experiment?",  # Check experimental design.
    "What would you do if an A/B test is inconclusive?",  # Check decision judgment.
]  # Finish the question list.
for i, question in enumerate(statistics_questions, start=1):  # Number each question.
    print(f"{i}. {question}")  # Display a practice prompt.
```

## 18. Solve a Business Case

A business case often begins with a broad symptom: revenue is down, costs are rising, churn increased, or delivery is late. A useful approach is to clarify the metric and time period, decompose the problem into drivers, segment the data, test plausible explanations, then recommend an action.

Do not jump to a solution before checking definitions and comparisons. Establish whether the change is real, where it occurs, how much it matters, and what evidence could distinguish competing explanations.

```python
business_case_steps = [  # Define a repeatable framework for an analyst case interview.
    "Clarify the decision, metric definition, scope, and time period.",  # Make the question precise.
    "Check data quality and validate the reported change.",  # Verify the symptom.
    "Decompose the KPI into drivers with a formula or tree.",  # Organize plausible explanations.
    "Segment by time, product, region, channel, and customer type.",  # Localize the change.
    "Compare with a suitable baseline and test alternative explanations.",  # Separate signal from noise.
    "Estimate impact and uncertainty; identify missing data.",  # Quantify importance and limits.
    "Recommend a next action, owner, success metric, and guardrail.",  # Turn analysis into a decision.
]  # Finish the case framework.
for step_number, step in enumerate(business_case_steps, start=1):  # Iterate through each step.
    print(f"{step_number}. {step}")  # Display the framework in order.
```

## 19. Explain an Analysis to a Nontechnical Person

A useful explanation starts with the conclusion, gives a few pieces of evidence, and finishes with the implication or decision. Avoid leading with implementation details or using undefined acronyms.

A concise structure is **finding → evidence → implication → recommendation**. Include the comparison period and caveat if they change how a listener should interpret the result.

```python
executive_summary = {  # Draft an example four-part explanation.
    "Finding": "Gross margin is lower in the Online channel than in the Store channel.",  # State the key observation.
    "Evidence": "The aggregated channel table shows a lower gross-profit-to-revenue ratio online.",  # Point to the metric and comparison.
    "Implication": "Discounting or product mix may be reducing online profitability.",  # Identify plausible explanations without claiming proof.
    "Recommendation": "Validate discounts, product mix, returns, and fulfillment cost before changing promotions.",  # Give a bounded action.
}  # Finish the executive summary.
for section, sentence in executive_summary.items():  # Iterate through the summary.
    print(f"{section}: {sentence}")  # Display the message in an executive-friendly format.
```

## 20. Prepare for a Take-Home Assignment

Before analysis, read the prompt carefully and note the expected decision, deliverables, deadline, and constraints. Plan time for data understanding, cleaning, analysis, validation, visualization, recommendations, and final review.

Avoid overengineering. A small, reproducible analysis with clear reasoning is usually stronger than many charts with unclear definitions. Make your assumptions explicit and document any trade-off caused by time limits.

```python
take_home_plan = pd.DataFrame({  # Plan a five-hour take-home exercise.
    "phase": ["Understand prompt", "Profile and clean", "Analyze", "Visualize", "Write recommendations", "QA and package"],  # List the phases.
    "minutes": [25, 55, 115, 60, 45, 20],  # Allocate an example amount of time to each phase.
})  # Finish the schedule.
take_home_plan["cumulative_minutes"] = take_home_plan["minutes"].cumsum()  # Calculate elapsed time after each phase.
print(take_home_plan.to_string(index=False))  # Display the plan.
print("Total planned minutes:", take_home_plan["minutes"].sum())  # Check the total time allocated.
```
